> ### <span style="color:orange">📖 Story section</span>
>

> "Sir Modulus's work on number remainders has led to some fascinating ways of hiding messages – difficult, or even impossible, to break without forbidden magic! One method in particular stands out: the PM (Power Modulus) technique.
>
> [<span style="color:lime">This technique</span>](https://youtu.be/cbGB__V8MNk) allows for incredibly rapid calculations of remainders when numbers are raised to massive powers and then divided by any number.
>
> The method uses a very clever trick that allows the bypassing of many multiplications and remainder calculations along the way. What if I devised a function that doesn't allow for such a shortcut? I shall name it `eni`, after my name, Enigmatus!"
> <br/><br/>

The eni function takes three arguments in this order `eni(N, EXP, MOD)`:

- The number `N` is multiplied by itself many times.
- The parameter `EXP` says how many times the number `N `is multiplied by itself.
- The number `MOD` is used to take the remainder after each multiplication.

The algorithm works as follows:

- Start with `1` as the initial score, and prepare an empty list to store remainders.
- Multiply the score by the number `N`.
- Take the remainder after dividing the score by `MOD`.
- Add this number to the start of the list of remainders.
- Repeat the multiplication, remainder, and appending steps `EXP` times.
- The final result is effectively a number formed by joining the remainders list as a single number.

For example,` eni(2,4,5) = 1342`, because:

<pre style="border: solid grey 1px;border-radius:1px;padding:2px;width:fit-content">
          score       remainders list 
               1
 1 * 2 mod 5 = 2                    2
 2 * 2 mod 5 = 4                  4,2
 4 * 2 mod 5 = 3                3,4,2
 3 * 2 mod 5 = 1              1,3,4,2
</pre>

Similarly, `eni(3,5,16) = 311193`, because:

<pre style="border: solid grey 1px;border-radius:1px;padding:2px;width:fit-content">
score     remainders list 
                 1
  1 * 3 mod 16 = 3                  3
  3 * 3 mod 16 = 9                9,3
 9 * 3 mod 16 = 11             11,9,3
 11 * 3 mod 16 = 1           1,11,9,3
  1 * 3 mod 16 = 3         3,1,11,9,3
</pre>

```python
from test_utilities import test

eni_tests = [
    {
        'name': 'Example eni(2,4,5) = 1342',
        'n': 2,
        'exp': 4,
        'mod': 5,
        'expected': 1342,
    },
    {
        'name': 'Example eni(3,5,16) = 311193',
        'n': 3,
        'exp': 5,
        'mod': 16,
        'expected': 311193,
    },
]


@test(tests=eni_tests)
def eni(n: int, exp: int, mod: int) -> int:
    score = 1
    result = []

    for _ in range(exp):
        score = score * n % mod
        result.append(score)

    return int(''.join(map(str, reversed(result))))
```

```python
import re
from itertools import batched

from test_utilities import test

tests = [
    {
        'name': 'Example',
        's': """
            A=4 B=4 C=6 X=3 Y=4 Z=5 M=11
            A=8 B=4 C=7 X=8 Y=4 Z=6 M=12
            A=2 B=8 C=6 X=2 Y=4 Z=5 M=13
            A=5 B=9 C=6 X=8 Y=6 Z=8 M=14
            A=5 B=9 C=7 X=6 Y=6 Z=8 M=15
            A=8 B=8 C=8 X=6 Y=9 Z=6 M=16
        """,
        'expected': 11611972920,
    },
]


def small_exercise(A: int, B: int, C: int, X: int, Y: int, Z: int, M: int) -> int:
    return eni(A, X, M) + eni(B, Y, M) + eni(C, Z, M)


@test(tests=tests)
def part_I(s: str) -> int:
    return max(
        small_exercise(*parameters)
        for parameters in batched(map(int, re.findall(r'\d+', s)), n=7, strict=True)
    )
```

```python
from test_utilities import test

eni_tests = [
    {
        'name': 'Example eni(2,7,5) = 34213',
        'n': 2,
        'exp': 7,
        'mod': 5,
        'expected': 34213,
    },
    {
        'name': 'Example eni(3,8,16) = 111931',
        'n': 3,
        'exp': 8,
        'mod': 16,
        'expected': 111931,
    },
]


def square_and_multiply(base: int, exponent: int, modulus: int) -> int:
    result = 1
    base %= modulus

    while exponent > 0:
        if exponent & 1 == 1:  # If the current bit is 1
            result = (result * base) % modulus

        exponent >>= 1
        base = (base * base) % modulus

    return result


@test(tests=eni_tests)
def eni(n: int, exp: int, mod: int) -> int:  # noqa: F811
    score = square_and_multiply(n, exp - 4, mod)
    result = [score]

    for _ in range(4):
        score = (score * n) % mod
        result.append(score)

    return int(''.join(map(str, reversed(result))))
```

```python
from test_utilities import test

eni_tests = [
    {
        'name': 'Example eni(2,7,5) = 19',
        'n': 2,
        'exp': 7,
        'mod': 5,
        'expected': 19,
    },
    {
        'name': 'Example eni(3,8,16) = 48',
        'n': 3,
        'exp': 8,
        'mod': 16,
        'expected': 48,
    },
]


def slow_modular_geometric_sum(base: int, exp: int, mod: int) -> int:
    total_sum = 0
    current_term = 1  # This represents base^0

    for _ in range(1, exp + 1):
        current_term = (current_term * base) % mod
        total_sum += current_term

    return total_sum


def detect_cycle(base: int, exp: int, mod: int) -> tuple[int, int, list[int]]:
    """returns start of of cycle, end of cycle, and the list of prefix sums.
    if no cycle detected the start of the cycle and the end of the cycle are set to -1"""
    seen = {}  # Maps a remainder to the loop index it appeared at
    prefix_sums = [0]  # Quick lookup for cumulative sum of the sequence

    base = base % mod

    current_term = 1  # Represents base^0

    for i in range(1, exp + 1):
        current_term = (current_term * base) % mod

        if current_term in seen:  # Cycle detected!
            return seen[current_term], i - 1, prefix_sums

        # Record state tracking data
        seen[current_term] = i
        prefix_sums.append(prefix_sums[-1] + current_term)

    return -1, -1, prefix_sums


def large_exponent_modular_sum(base: int, exp: int, mod: int) -> int:
    # Step 1: Detect the cycle
    cycle_start, cycle_end, prefix_sums = detect_cycle(base, exp, mod)

    if cycle_start < 0:  # No cycle detected
        return prefix_sums[-1]

    # Step 2: Split the exponent into pre-cycle, full-cycles, and remainder
    pre_cycle_len = cycle_start - 1
    cycle_len = (cycle_end - cycle_start) + 1
    remaining_exp = exp - pre_cycle_len

    full_cycles = remaining_exp // cycle_len
    leftover_elements = remaining_exp % cycle_len

    # Step 3: Math out the totals using our prefix sum lookups
    pre_cycle_sum = prefix_sums[pre_cycle_len]

    single_cycle_sum = prefix_sums[cycle_end] - prefix_sums[pre_cycle_len]
    all_cycles_sum = full_cycles * single_cycle_sum

    leftover_sum = prefix_sums[pre_cycle_len + leftover_elements] - prefix_sums[pre_cycle_len]

    # Step 4: Combine parts
    return pre_cycle_sum + all_cycles_sum + leftover_sum


@test(tests=eni_tests)
def eni(n: int, exp: int, mod: int) -> int:  # noqa: F811
    return large_exponent_modular_sum(n, exp, mod)
```
