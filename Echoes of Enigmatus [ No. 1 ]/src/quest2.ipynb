Let's first build the trees without any swaps. The command to add new nodes always has a unique identifier for the tangled pair and two nodes: one for the left tree and one for the right.

A node is represented by a pair of values in the format: <code> [rank,symbol] </code>. The rank is an integer unique to both trees and determines the tree's structure by indicating where the node should be added. The symbol can be any character and is used only for information. For example, symbols can hide a message within the tree's structure.

<span style="color:ivory">Nodes are added to the trees following traditional binary tree rules</span>, namely:

- If the tree is empty, the node becomes the root;
- Otherwise, compare the new node's rank with the rank of the current node (starting from the root);
- If the new node's rank is smaller, move left;
- If it is larger, move right;
- Repeat the process until an empty spot is found, where the node is added.

Each tree – left and right – works independently, following the same rules. However, nodes are always added in pairs: one node goes to one tree, and the other simultaneously goes to the other tree.
Below is a list of commands for adding several pairs of nodes ( <span style="color:orange">your notes</span> ). Based on these commands, build two binary trees – left and right. <span style="color:ivory">Then, in each tree, find the level with the most nodes. Read the symbols from these levels as a single message, starting with the left tree and then the right.</span>

```python
import re
from dataclasses import dataclass, fields

from IPython.display import HTML, display
from test_utilities import test

tests = [
    {
        'name': 'Example',
        's': """
            ADD id=1 left=[10,A] right=[30,H]
            ADD id=2 left=[15,D] right=[25,I]
            ADD id=3 left=[12,F] right=[31,J]
            ADD id=4 left=[5,B] right=[27,L]
            ADD id=5 left=[3,C] right=[28,M]
            ADD id=6 left=[20,G] right=[32,K]
            ADD id=7 left=[4,E] right=[21,N]
        """,
        'expected': 'CFGNLK',
    },
    {
        'name': 'Slightly More Complex Example',
        's': """
            ADD id=1 left=[160,E] right=[175,S]
            ADD id=2 left=[140,W] right=[224,D]
            ADD id=3 left=[122,U] right=[203,F]
            ADD id=4 left=[204,N] right=[114,G]
            ADD id=5 left=[136,V] right=[256,H]
            ADD id=6 left=[147,G] right=[192,O]
            ADD id=7 left=[232,I] right=[154,K]
            ADD id=8 left=[118,E] right=[125,Y]
            ADD id=9 left=[102,A] right=[210,D]
            ADD id=10 left=[183,Q] right=[254,E]
            ADD id=11 left=[146,E] right=[148,C]
            ADD id=12 left=[173,Y] right=[299,S]
            ADD id=13 left=[190,B] right=[277,B]
            ADD id=14 left=[124,T] right=[142,N]
            ADD id=15 left=[153,R] right=[133,M]
            ADD id=16 left=[252,D] right=[276,M]
            ADD id=17 left=[258,I] right=[245,P]
            ADD id=18 left=[117,O] right=[283,!]
            ADD id=19 left=[212,O] right=[127,R]
            ADD id=20 left=[278,A] right=[169,C]
        """,
        'expected': 'EVERYBODYCODES',
    },
]


@dataclass(unsafe_hash=True)
class Node:
    rank: int
    symbol: str
    left: Node | None = None
    right: Node | None = None

    def get_levels(self) -> list[list[Node]]:
        level = [self]
        levels = []

        while level:
            next_level = []
            for node in level:
                if node.left:
                    next_level.append(node.left)
                if node.right:
                    next_level.append(node.right)

            levels.append(level)
            level = next_level

        return levels

    def add_node(self, node: Node) -> None:
        if node.rank < self.rank:
            if self.left is None:
                self.left = node
            else:
                self.left.add_node(node)
        else:
            if self.right is None:
                self.right = node
            else:
                self.right.add_node(node)


@dataclass
class TangledTree:
    left: Node | None = None
    right: Node | None = None

    def get_message(self, debug=False) -> str:
        levels_left = self.left.get_levels() if self.left else []
        levels_right = self.right.get_levels() if self.right else []

        max_left = max(levels_left, key=len)
        max_right = max(levels_right, key=len)

        if debug:
            print(f'levels_left = {[[node.symbol for node in level] for level in levels_left]}')
            print(f'{max_left=}')
            print(f'levels_right={[[node.symbol for node in level] for level in levels_right]}')
            print(f'{max_right=}')

        return ''.join(n.symbol for n in max_left) + ''.join(n.symbol for n in max_right)

    def display(self) -> None:
        display(
            HTML(f"""
                     <div style="display:grid; grid-template-columns:1fr 1fr; gap:20px;justify-items: center;">
                        <div>{self.tree_to_svg(self.left)}</div>
                        <div>{self.tree_to_svg(self.right)}</div>
                     </div>
             """)
        )

    @classmethod
    def layout_tree(cls, root: Node | None, x=0, y=0, x_step=60, y_step=80):
        # 1. Assign x,y positions using an in-order traversal
        positions = {}  # node -> (x, y)
        current_x = x

        def dfs(node: Node | None, depth: int):
            nonlocal current_x
            if not node:
                return
            dfs(node.left, depth + 1)
            positions[node] = (current_x, depth * y_step)
            current_x += x_step
            dfs(node.right, depth + 1)

        dfs(root, 0)
        return positions

    @classmethod
    def tree_to_svg(cls, root: Node | None, radius=18) -> str:
        # 2. Build SVG elements
        if not root:
            return "<svg xmlns='http://www.w3.org/2000/svg'></svg>"

        positions = cls.layout_tree(root)
        nodes_svg: list[str] = []
        edges_svg: list[str] = []

        # edges
        for node, (x, y) in positions.items():
            for child in (node.left, node.right):
                if child:
                    cx, cy = positions[child]
                    edges_svg.append(
                        f"<line x1='{x}' y1='{y}' x2='{cx}' y2='{cy}' "
                        f"stroke='ivory' stroke-width='2' />"
                    )

        # nodes
        for node, (x, y) in positions.items():
            nodes_svg.append(
                f"<circle cx='{x}' cy='{y}' r='{radius}' fill='black' stroke='lime' stroke-width='4'/>"
            )
            nodes_svg.append(
                f"<text x='{x}' y='{y - 2}' stroke='ivory' text-anchor='middle' "
                f"font-family='monospace' font-size='12'>{node.rank}</text>"
            )
            nodes_svg.append(
                f"<text x='{x}' y='{y + 9}' stroke='ivory' text-anchor='middle' "
                f"font-family='monospace' font-size='12'>{node.symbol}</text>"
            )

        # compute bounds
        xs = [p[0] for p in positions.values()]
        ys = [p[1] for p in positions.values()]
        min_x, max_x = min(xs) - 40, max(xs) + 40
        min_y, max_y = min(ys) - 40, max(ys) + 40
        width = max_x - min_x
        height = max_y - min_y

        svg_content = '\n  '.join(edges_svg + nodes_svg)
        svg = (
            f"<svg xmlns='http://www.w3.org/2000/svg' "
            f"viewBox='{min_x} {min_y} {width} {height}' "
            f"width='{width}' height='{height}'>\n  {svg_content}\n</svg>"
        )
        return svg

    @classmethod
    def build_from_str(cls, s: str) -> TangledTree:
        pattern = re.compile(r'\[(\d+,\S+)\]')

        nodes = [(int(node.split(',')[0]), node.split(',')[-1]) for node in re.findall(pattern, s)]

        if not nodes:
            return cls()

        left_root, *left_nodes = nodes[::2]
        right_root, *right_nodes = nodes[1::2]

        tt = cls(Node(*left_root), Node(*right_root))

        assert tt.left and tt.right

        for node in left_nodes:
            tt.left.add_node(Node(*node))

        for node in right_nodes:
            tt.right.add_node(Node(*node))

        return tt


@test(tests=tests[:])
def part_I(s: str, debug=False) -> str:
    tt = TangledTree.build_from_str(s)

    if debug:
        tt.display()

    return tt.get_message()
```

```python
from more_itertools import first
from util import hexcolor_str

tests = [
    {
        'name': 'Example Part I',
        's': """
            ADD id=1 left=[10,A] right=[30,H]
            ADD id=2 left=[15,D] right=[25,I]
            ADD id=3 left=[12,F] right=[31,J]
            ADD id=4 left=[5,B] right=[27,L]
            ADD id=5 left=[3,C] right=[28,M]
            ADD id=6 left=[20,G] right=[32,K]
            ADD id=7 left=[4,E] right=[21,N]
        """,
        'expected': 'CFGNLK',
    },
    {
        'name': 'Slightly More Complex Example Part I',
        's': """
            ADD id=1 left=[160,E] right=[175,S]
            ADD id=2 left=[140,W] right=[224,D]
            ADD id=3 left=[122,U] right=[203,F]
            ADD id=4 left=[204,N] right=[114,G]
            ADD id=5 left=[136,V] right=[256,H]
            ADD id=6 left=[147,G] right=[192,O]
            ADD id=7 left=[232,I] right=[154,K]
            ADD id=8 left=[118,E] right=[125,Y]
            ADD id=9 left=[102,A] right=[210,D]
            ADD id=10 left=[183,Q] right=[254,E]
            ADD id=11 left=[146,E] right=[148,C]
            ADD id=12 left=[173,Y] right=[299,S]
            ADD id=13 left=[190,B] right=[277,B]
            ADD id=14 left=[124,T] right=[142,N]
            ADD id=15 left=[153,R] right=[133,M]
            ADD id=16 left=[252,D] right=[276,M]
            ADD id=17 left=[258,I] right=[245,P]
            ADD id=18 left=[117,O] right=[283,!]
            ADD id=19 left=[212,O] right=[127,R]
            ADD id=20 left=[278,A] right=[169,C]
        """,
        'expected': 'EVERYBODYCODES',
    },
    {
        'name': 'challenge Part I',
        's': notes1,
        'expected': 'QUACK!ZTGWTBWL',
    },
    {
        'name': 'Example Part II',
        's': """
            ADD id=1 left=[10,A] right=[30,H]
            ADD id=2 left=[15,D] right=[25,I]
            ADD id=3 left=[12,F] right=[31,J]
            ADD id=4 left=[5,B] right=[27,L]
            ADD id=5 left=[3,C] right=[28,M]
            SWAP 1
            SWAP 5
            ADD id=6 left=[20,G] right=[32,K]
            ADD id=7 left=[4,E] right=[21,N]
        """,
        'expected': 'MGFLNK',
    },
]


class NodeWithSwap(Node):
    def swap(self, other: NodeWithSwap) -> None:
        self.rank, other.rank = other.rank, self.rank
        self.symbol, other.symbol = other.symbol, self.symbol


class TangledTreeWithSwap(TangledTree):
    debug: bool = False

    @classmethod
    def build_from_str(cls, s: str, node_cls: type = NodeWithSwap) -> TangledTree:
        pattern = re.compile(r'\[(\d+,\S+)\]')
        integers = re.compile(r'\d+')

        lines = [line.strip() for line in s.strip().splitlines()]

        tt = cls()
        ids: list[list[type]] = [[]]

        for line in lines:
            if line.startswith('ADD'):
                left, right = re.findall(pattern, line)

                node_left = node_cls(int(left.split(',')[0]), left.split(',')[-1])
                if tt.left is None:
                    tt.left = node_left
                else:
                    tt.left.add_node(node_left)

                node_right = node_cls(int(right.split(',')[0]), right.split(',')[-1])
                if tt.right is None:
                    tt.right = node_right
                else:
                    tt.right.add_node(node_right)

                ids.append([node_left, node_right])

                if cls.debug:
                    print(hexcolor_str('#00FF00', '**** ADD ****'))
                    tt.display()

            else:
                swap_id = int(first(re.findall(integers, line)))
                left_node, right_node = ids[swap_id]
                left_node.swap(right_node)

                if cls.debug:
                    print(hexcolor_str('#ef03b2', '**** SWAP ****'))
                    tt.display()

        return tt


@test(tests=tests[:])
def part_II(s: str, debug=False) -> str:
    tt = TangledTreeWithSwap.build_from_str(s)

    if debug:
        tt.display()

    return tt.get_message()
```

```python
tests = [
    {
        'name': 'Example Part III',
        's': """
            ADD id=1 left=[10,A] right=[30,H]
            ADD id=2 left=[15,D] right=[25,I]
            ADD id=3 left=[12,F] right=[31,J]
            ADD id=4 left=[5,B] right=[27,L]
            ADD id=5 left=[3,C] right=[28,M]
            SWAP 1
            SWAP 5
            ADD id=6 left=[20,G] right=[32,K]
            ADD id=7 left=[4,E] right=[21,N]
            SWAP 2
        """,
        'expected': 'DJMGL',
    },
    {
        'name': 'Slightly More Complex Example Part I',
        's': """
            ADD id=1 left=[10,A] right=[30,H]
            ADD id=2 left=[15,D] right=[25,I]
            ADD id=3 left=[12,F] right=[31,J]
            ADD id=4 left=[5,B] right=[27,L]
            ADD id=5 left=[3,C] right=[28,M]
            SWAP 1
            SWAP 5
            ADD id=6 left=[20,G] right=[32,K]
            ADD id=7 left=[4,E] right=[21,N]
            SWAP 2
            SWAP 5
        """,
        'expected': 'DJCGL',
    },
]


class NodeWithTotalSwap(NodeWithSwap):
    def swap(self, other: NodeWithTotalSwap) -> None:
        for f in fields(self):
            # Swap values in place
            self_val = getattr(self, f.name)
            other_val = getattr(other, f.name)
            setattr(self, f.name, other_val)
            setattr(other, f.name, self_val)


@test(tests=tests[:])
def part_III(s: str, debug=False) -> str:
    tt = TangledTreeWithSwap.build_from_str(s, node_cls=NodeWithTotalSwap)

    if debug:
        tt.display()

    return tt.get_message()
```
