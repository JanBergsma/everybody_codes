After <code> 100 </code>days, the snails are at the following positions:
<code> [x=1 y=2], [x=6 y=2], [x=2 y=3], [x=1 y=6] </code>

When we use the formula <code> x + (100 \* y) </code> for each snail, the sum is therefore:
<code style="color:ivory;font-weight:bolder"> 201 + 206 + 302 + 601 = 1310 </span>

<span style="color:orange">What is the sum of all snails' positions after <code style="color:orange;border: dotted orange 1px;border-radius:2px;padding:2px;width:fit-content"> 100 </code> days?</span>


```python
import re
from functools import partial
from itertools import batched

from test_utilities import test

tests = [
    {
        'name': 'Example',
        's': """
            x=1 y=2
            x=2 y=3
            x=3 y=4
            x=4 y=4
        """,
        'expected': 1310,
    },
]


def anti_diagonals_pos_after_days(snail: tuple[int, int], days: int, debug: bool = False) -> int:
    x, y = snail

    x, y = x - 1, y - 1
    k = x + y
    mod = k + 1

    x = (x + days) % mod  # x is the pos on the anti-diagonal we go from bottom to top so add days
    y = k - x
    x, y = x + 1, y + 1

    if debug:
        print(f'{snail=} after {days} days => [{x=},{y=}]={x + 100 * y}')

    return x + 100 * y


@test(tests=tests[:])
def part_I(s: str, days: int = 100) -> int:
    snails = batched(map(int, re.findall(r'\d+', s)), n=2, strict=True)
    after_100_days = partial(anti_diagonals_pos_after_days, days=days)
    return sum(map(after_100_days, snails))  # type: ignore
```

<span style="color:orange">After how many days will all the snails align on the golden line for the first time?</span>


```python
from sympy.ntheory.modular import crt
from test_utilities import test

tests = [
    {
        'name': 'Example I',
        's': """
            x=12 y=2
            x=8 y=4
            x=7 y=1
            x=1 y=5
            x=1 y=3
        """,
        'expected': 14,
    },
    {
        'name': 'Example II',
        's': """
            x=3 y=1
            x=3 y=9
            x=1 y=5
            x=4 y=10
            x=5 y=3
        """,
        'expected': 13659,
    },
]


def solve_crt(inverses: list[int], moduli: list[int], debug: bool = False) -> int:
    result = crt(moduli, inverses)

    if result is None:
        raise ValueError('No Solution')

    if debug:
        sol, mod = result
        print(f'Solution: {sol}, Modulo: {mod}')

    return result[0]


@test(tests=tests[:])
def part_II(s: str, debug: bool = False) -> int:
    snails = batched(map(int, re.findall(r'\d+', s)), n=2, strict=True)
    inverses, moduli = [], []

    for x, y in snails:
        x, y = x - 1, y - 1

        m = x + y + 1
        x_inv = -x % m

        inverses.append(x_inv)
        moduli.append(m)

    return solve_crt(inverses, moduli, debug) - 1
```
