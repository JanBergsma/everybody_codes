The colour of each dragonduck scale is stored as three colour components: red, green, and blue. Each component is written as six letters, either uppercase or lowercase: <code style="border: 0.1pt solid DimGray;border-radius: 5px;padding:0.1rem 1rem;width: max-content;">Rr</code> for red, <code style="border: 0.1pt solid DimGray;border-radius: 5px;padding:0.1rem 1rem;width: max-content;">Gg</code> for green, and <code style="border: 0.1pt solid DimGray;border-radius: 5px;padding:0.1rem 1rem;width: max-content;">Bb</code> for blue. An example colour can be written as <code style="border: 0.1pt solid DimGray;border-radius: 5px;padding:0.1rem 1rem;width: max-content;">RrRRrR GgggGG bBbBBb</code>.

Veronica noticed that if we convert each component into a binary number and mix the colours in these proportions, we get the colour of the analysed scale! We only need to treat uppercase letters as ones and lowercase letters as zeros:

<pre style="border: 0.1pt solid DimGray;border-radius: 5px;padding:1rem;width: max-content;">
Value   Binary   Red      Green    Blue
     0   000000   rrrrrr   gggggg   bbbbbb 
     1   000001   rrrrrR   gggggG   bbbbbB
     2   000010   rrrrRr   ggggGg   bbbbBb
     3   000011   rrrrRR   ggggGG   bbbbBB
     4   000100   rrrRrr   gggGgg   bbbBbb
     5   000101   rrrRrR   gggGgG   bbbBbB
     6   000110   rrrRRr   gggGGg   bbbBBb
     7   000111   rrrRRR   gggGGG   bbbBBB
     8   001000   rrRrrr   ggGggg   bbBbbb
     9   001001   rrRrrR   ggGggG   bbBbbB
                                      etc.
</pre>

To practise understanding the rules, the nerdmaster gives you a list with scale identifiers and colour components. Your first task is to find all **scales where the green value is dominant (greater than red and greater than blue), and then give the sum of the identifiers of those scales.**

**Example based on the following notes:**

<pre style="border: 2px dotted lime;border-radius: 5px;padding:1rem;width: max-content;">
2456:rrrrrr ggGgGG bbbbBB
7689:rrRrrr ggGggg bbbBBB
3145:rrRrRr gggGgg bbbbBB
6710:rrrRRr ggGGGg bbBBbB
</pre>

After converting the components, we get the following values:

<pre style="border: 1pt solid DimGray;border-radius: 5px;padding:1rem;width: max-content;">
2456:   rrrrrr:  0   ggGgGG: 11   bbbbBB:  3 
7689:   rrRrrr:  8   ggGggg:  8   bbbBBB:  7
3145:   rrRrRr: 10   gggGgg:  4   bbbbBB:  3
6710:   rrrRRr:  6   ggGGGg: 14   bbBBbB: 13
</pre>

Green has the highest value in scales <code style="border: 0.1pt solid DimGray;border-radius: 5px;padding:0.1rem 1rem;width: max-content;">2456</code> and <code style="border: 0.1pt solid DimGray;border-radius: 5px;padding:0.1rem 1rem;width: max-content;">6710</code>, so the answer for this example is <code style="border: 0.1pt solid DimGray;border-radius: 5px;padding:0.1rem 1rem;width: max-content;">9166</code>.

<span style="color:#eed699">What is the sum of the identifiers of the scales where green is the dominant colour<span>


```python
import re
from itertools import batched

from test_utilities import test

test_cases = [
    # cSpell:disable
    {
        'name': 'Example I',
        'notes': """
            2456:rrrrrr ggGgGG bbbbBB
            7689:rrRrrr ggGggg bbbBBB
            3145:rrRrRr gggGgg bbbbBB
            6710:rrrRRr ggGGGg bbBBbB
         """,
        'expected': 9166,
    },
    # cSpell:enable
]


def convert(s: str) -> int:
    return sum(2**i for i, ch in enumerate(reversed(s)) if ch.isupper())


@test(tests=test_cases[:])
def partI(notes: str, debug: bool = False) -> int:
    scales = [int(scale) for scale in re.findall(r'\d+', notes)]
    rgbs = batched(map(convert, re.findall(r'[RGBrgb]+', notes)), 3, strict=True)

    return sum(
        scale for scale, rgb in zip(scales, rgbs) if max(range(3), key=lambda i: rgb[i]) == 1
    )
```

```python
with open('./inputs/everybody_codes_e3_q01_p1.txt') as f:
    notes1 = f.read()

print(f'Part I: {partI(notes1)}')
```

## <span style="color:#eed699">Part II</span>
In no time, you find the required scales, which brings a familiar smile of admiration to Veronica’s face. Without any delay, you begin the next part of the analysis.

Besides colour, each scale also has a level of shine. It is encoded in the same way as the colour components, but using the letters <code style="border: 0.1pt solid DimGray;border-radius: 5px;padding:0.1rem 1rem;width: max-content;">Ss</code> . The new list for analysis includes the shine component as the last attribute, placed just after the blue component. Veronica asks you to find the scales of individuals with the highest level of shine, and among them choose the darkest one (the one with the lowest total sum of colour components).

**Example based on the following notes:**

<pre style="border: 2px dotted lime;border-radius: 5px;padding:1rem;width: max-content;">
2456:rrrrrr ggGgGG bbbbBB sSsSsS
7689:rrRrrr ggGggg bbbBBB ssSSss
3145:rrRrRr gggGgg bbbbBB sSsSsS
6710:rrrRRr ggGGGg bbBBbB ssSSss
</pre>

After converting the components, we get the following values:

<pre style="border: 0.1pt solid DimGray;border-radius: 5px;padding:1rem;width: max-content;">
2456:   rrrrrr:  0   ggGgGG: 11   bbbbBB:  3   sSsSsS: 21 
7689:   rrRrrr:  8   ggGggg:  8   bbbBBB:  7   ssSSss: 12
3145:   rrRrRr: 10   gggGgg:  4   bbbbBB:  3   sSsSsS: 21
6710:   rrrRRr:  6   ggGGGg: 14   bbBBbB: 13   ssSSss: 12
</pre>

The scales with identifiers <code style="border: 0.1pt solid DimGray;border-radius: 5px;padding:0.1rem 1rem;width: max-content;">2456</code> and <code style="border: 0.1pt solid DimGray;border-radius: 5px;padding:0.1rem 1rem;width: max-content;">3145</code> have the highest shine.

The scale <code style="border: 0.1pt solid DimGray;border-radius: 5px;padding:0.1rem 1rem;width: max-content;">2456</code> has a colour sum of `0 + 11 + 3 = 14`, while <code style="border: 0.1pt solid DimGray;border-radius: 5px;padding:0.1rem 1rem;width: max-content;">3145</code> has a colour sum of `10 + 4 + 3 = 17`.

The darkest scale among the most shiny ones is therefore <code style="border: 0.1pt solid DimGray;border-radius: 5px;padding:0.1rem 1rem;width: max-content;">2456</code>.

<span style="color:#eed699">What is the identifier of the darkest scale among the most shiny ones?</span>


```python
from more_itertools import last
from test_utilities import test

test_cases = [
    # cSpell:disable
    {
        'name': 'Example I',
        'notes': """
            2456:rrrrrr ggGgGG bbbbBB sSsSsS
            7689:rrRrrr ggGggg bbbBBB ssSSss
            3145:rrRrRr gggGgg bbbbBB sSsSsS
            6710:rrrRRr ggGGGg bbBBbB ssSSss
         """,
        'expected': 2456,
    },
    # cSpell:enable
]


@test(tests=test_cases[:])
def partII(notes: str, debug: bool = False) -> int:
    scales = [int(scale) for scale in re.findall(r'\d+', notes)]
    rgbs = list(batched(map(convert, re.findall(r'[RrGgBbSs]+', notes)), 4, strict=True))

    max_shine = max(shine for *_, shine in rgbs)

    return last(
        min((sum(rgb[:-1]), scale) for scale, rgb in zip(scales, rgbs) if rgb[-1] == max_shine)
    )
```

```python
with open('./inputs/everybody_codes_e3_q01_p2.txt') as f:
    notes2 = f.read()


print(f'Part II: {partII(notes2)}')
```

For the next stage of the research, you must **find the largest group of similar scales**, based on colour and shine. For this reason, the entire collection of scales must be grouped by their dominant colour and their shine. This will create six possible groups of scales: red matte, red shiny, green matte, green shiny, blue matte, and blue shiny.

A scale is considered matte if its shine value is <code style="border: 0.1pt solid DimGray;border-radius: 5px;padding:0.1rem 1rem;width: max-content;">30</code> or lower, and shiny if this value is <code style="border: 0.1pt solid DimGray;border-radius: 5px;padding:0.1rem 1rem;width: max-content;">33</code> or higher.

For the remaining values, it is difficult to decide whether the scale is shiny or not, so such scales should be ignored during grouping. Similarly, if it is not possible to determine a dominant colour (one that has a higher value than the others), that scale should also be ignored.

<span style="color:#eed699">What is the sum of the identifiers of the scales in the largest group?</span>


```python
from collections import defaultdict

from tabulate import tabulate
from test_utilities import test

test_cases = [
    # cSpell:disable
    {
        'name': 'Example I',
        'notes': """
            15437:rRrrRR gGGGGG BBBBBB sSSSSS
            94682:RrRrrR gGGggG bBBBBB ssSSSs
            56513:RRRrrr ggGGgG bbbBbb ssSsSS
            76346:rRRrrR GGgggg bbbBBB ssssSs
            87569:rrRRrR gGGGGg BbbbbB SssSss
            44191:rrrrrr gGgGGG bBBbbB sSssSS
            49176:rRRrRr GggggG BbBbbb sSSssS
            85071:RRrrrr GgGGgg BBbbbb SSsSss
            44303:rRRrrR gGggGg bBbBBB SsSSSs
            94978:rrRrRR ggGggG BBbBBb SSSSSS
            26325:rrRRrr gGGGgg BBbBbb SssssS
            43463:rrrrRR gGgGgg bBBbBB sSssSs
            15059:RRrrrR GGgggG bbBBbb sSSsSS
            85004:RRRrrR GgGgGG bbbBBB sSssss
            56121:RRrRrr gGgGgg BbbbBB sSsSSs
            80219:rRRrRR GGGggg BBbbbb SssSSs
         """,
        'expected': 292320,
    },
    # cSpell:enable
]


@test(tests=test_cases[:])
def partIII(notes: str, debug: bool = False) -> int:
    scales = [int(scale) for scale in re.findall(r'\d+', notes)]
    rgbs = list(batched(map(convert, re.findall(r'[RrGgBbSs]+', notes)), 4, strict=True))

    groups = defaultdict(list)
    for scale, (*rgb, shine) in zip(scales, rgbs):
        if 30 < shine < 33:
            continue

        max_color = max(rgb)
        if rgb.count(max_color) > 1:
            continue

        color = rgb.index(max_color)
        groups[(shine <= 30, color)].append(scale)

    if debug:
        print(tabulate(groups, headers=groups.keys()))  # type: ignore

    return sum(max(groups.values(), key=len))
```
