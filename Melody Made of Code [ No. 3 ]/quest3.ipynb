### <span style="color:#eed699">Part I</span>

> ## <span style="color:#eed699"> <img src='./images/book.svg' height=40 width=40 style="display:inline-block height:fit-content;vertical-align: middle; margin: 5px; ">Story section</span>
>
> The final missing element of the sounds produced by dragonducks is melody. Veronica has discovered that the scales contain certain components which suggest that they are, quite simply… musical notes. However, just like with the vocal bones, it appears that the scales do not store the melody itself, but rather a precise recipe for how to construct something that can later reveal it.
>
> Veronica shows her notes, filled with strange elements that at first glance seem to connect with one another in a chaotic way. It is precisely these elements which, when properly arranged, will reveal the sequence of sounds that forms the melody.
>
> <br/>

The previous part of the analysis went exceptionally smoothly, so Veronica suggests to start again with a simplified version of the target mechanism.

The melody is defined as a **list of elements which Veronica has called “nodes”.** An example definition of a single node is shown below:

<pre style="border: 0.1pt solid DimGray;border-radius: 5px;padding:1rem;width: max-content;">
id=1, plug=BLUE HEXAGON, leftSocket=GREEN CIRCLE, rightSocket=BLUE PENTAGON, data=...
</pre>

The first three steps follow the sequence: up, right, down, without any obstacles. Af

Each node consists of a unique identifier `id`, a single plug , and two sockets: `leftSocket` and `rightSocket`, each with a specific colour and shape. In addition, at the end of the node definition there is some `data`, where (according to the nerdmaster’s suspicions) the melody itself is hidden, but you can completely ignore this part. A sample node like the one above can therefore be imagined as follows:

Nodes are constructed one by one according to the list encoded on the scale and are connected into a larger structure called a “tree”. **The rules for building the tree are as follow**s. The first node in the list becomes the root of the tree.
Each subsequent node circles the existing tree in a clockwise direction, starting from the root. Note that in Veronica’s notation the root node is located at the bottom of the tree; therefore, when moving clockwise from the root, the first socket encountered is the left socket of the root. The traversal follows the outer boundary of the tree, and sockets are encountered in the order in which they appear along this clockwise path. **When a node encounters a socket whose shape and colour match its plug, it connects at that exact point, ending its journey**.

Once all nodes from the list have found their target positions, the reading phase starts. Reading consists of constructing a special reading node, without a plug or sockets, which circles the tree in exactly the same way as the other nodes, but whose task is to read the data of all nodes in the order of its journey. **Reading happens when the node passes the upper part of the node being read**, marked in the diagram above as `*`, which is located exactly between the left and right socket.

Veronica proposes a quick exercise using a short list of nodes. The list is guaranteed to connect all nodes into a single tree. Your task is to **build the tree and read the node ids** in the order produced by the reading node. Then **calculate the checksum** using the resulting list of ids, by multiplying each id by its position in that list (starting from 1) and adding everything together.

**Example based on the following notes:**

<pre style="border: 2px dotted lime;border-radius: 5px;padding:1rem;width: max-content;">
id=1, plug=BLUE HEXAGON, leftSocket=GREEN CIRCLE, rightSocket=BLUE PENTAGON, data=?
id=2, plug=GREEN CIRCLE, leftSocket=BLUE HEXAGON, rightSocket=BLUE CIRCLE, data=?
id=3, plug=BLUE PENTAGON, leftSocket=BLUE CIRCLE, rightSocket=BLUE CIRCLE, data=?
id=4, plug=BLUE CIRCLE, leftSocket=RED HEXAGON, rightSocket=BLUE HEXAGON, data=?
id=5, plug=RED HEXAGON, leftSocket=GREEN CIRCLE, rightSocket=RED HEXAGON, data=?
</pre>

```python
from dataclasses import dataclass
from typing import Self

from IPython.display import HTML, display
from test_utilities import test

test_cases = [
    # cSpell:disable
    {
        'name': 'Example I',
        'notes': """
            id=1, plug=BLUE HEXAGON, leftSocket=GREEN CIRCLE, rightSocket=BLUE PENTAGON, data=?
            id=2, plug=GREEN CIRCLE, leftSocket=BLUE HEXAGON, rightSocket=BLUE CIRCLE, data=?
            id=3, plug=BLUE PENTAGON, leftSocket=BLUE CIRCLE, rightSocket=BLUE CIRCLE, data=?
            id=4, plug=BLUE CIRCLE, leftSocket=RED HEXAGON, rightSocket=BLUE HEXAGON, data=?
            id=5, plug=RED HEXAGON, leftSocket=GREEN CIRCLE, rightSocket=RED HEXAGON, data=?
         """,
        'expected': 43,
    },
    # cSpell:enable
]

type Plug = tuple[str, str]


@dataclass
class Node:
    id: int
    plug: Plug
    left_preference: Plug
    right_preference: Plug
    left_socket: Node | None = None
    right_socket: Node | None = None
    data: str | None = None

    def add(self, node: Node) -> Node | None:
        if not self.left_socket:
            if node.plug == self.left_preference:
                self.left_socket = node
                return
        else:
            if not self.left_socket.add(node):
                return

        if not self.right_socket:
            if node.plug == self.right_preference:
                self.right_socket = node
                return
        else:
            if not self.right_socket.add(node):
                return

        return node

    def checksum(self) -> int:
        acc = []
        self.in_order_traversal(acc)
        return sum(i * id for i, id in enumerate(acc, start=1))

    def in_order_traversal(self, acc: list[int]) -> None:
        if self.left_socket:
            self.left_socket.in_order_traversal(acc)

        acc.append(self.id)

        if self.right_socket:
            self.right_socket.in_order_traversal(acc)

    def __hash__(self):
        return self.id

    def display(self) -> None:
        display(HTML(f'<div>{self.tree_to_svg(self)}</div>'))

    @classmethod
    def create_tree(cls, notes: str, debug: bool = False) -> Self:
        nodes = cls.parse(notes)
        root = nodes[0]

        if debug:
            root.display()

        for node in nodes[1:]:
            while node:
                node = root.add(node)

            if debug:
                root.display()

        return root

    @classmethod
    def parse(cls, notes: str) -> list[Self]:
        nodes = []

        for line in notes.strip().splitlines():
            obj = {}
            for field in line.strip().split(','):
                name, data = field.strip().split('=')
                if data.isdigit():
                    obj[name] = int(data)
                else:
                    obj[name] = data

            nodes.append(
                cls(
                    id=obj['id'],
                    plug=tuple(obj['plug'].split(' ')),
                    left_preference=tuple(obj['leftSocket'].split(' ')),
                    right_preference=tuple(obj['rightSocket'].split(' ')),
                    data=obj['data'],
                )
            )

        return nodes

    @classmethod
    def layout_tree(cls, root: Node | None, x=0, y=0, x_step=60, y_step=80):
        # 1. Assign x,y positions using an in-order traversal
        positions = {}  # node -> (x, y)
        current_x = x

        def dfs(node: Node | None, depth: int):
            nonlocal current_x
            if not node:
                return
            dfs(node.left_socket, depth + 1)
            positions[node] = (current_x, depth * y_step)
            current_x += x_step
            dfs(node.right_socket, depth + 1)

        dfs(root, 0)
        return positions

    @classmethod
    def tree_to_svg(cls, root: Node | None, radius=52, width=200) -> str:
        # 2. Build SVG elements
        if not root:
            return "<svg xmlns='http://www.w3.org/2000/svg'></svg>"

        positions = cls.layout_tree(root)
        nodes_svg: list[str] = []
        edges_svg: list[str] = []

        # edges
        for node, (x, y) in positions.items():
            for child in (node.left_socket, node.right_socket):
                if child:
                    cx, cy = positions[child]
                    edges_svg.append(
                        f"<line x1='{x}' y1='{y}' x2='{cx}' y2='{cy}' "
                        f"stroke='ivory' stroke-width='2' />"
                    )

        # nodes
        for node, (x, y) in positions.items():
            nodes_svg.append(
                f"<circle cx='{x}' cy='{y}' r='{radius}' fill='black' stroke='lime' stroke-width='4'/>"
            )
            nodes_svg.append(
                f"<text x='{x}' y='{y - 2}' stroke='white' text-anchor='middle' "
                f"font-family='monospace' font-size='12'>{node.id}</text>"
            )
            nodes_svg.append(
                f"<text x='{x}' y='{y + 9}' stroke='orange' text-anchor='middle' "
                f"font-family='monospace' font-size='10'>{node.plug}</text>"
            )

        # compute bounds
        xs = [p[0] for p in positions.values()]
        ys = [p[1] for p in positions.values()]
        min_x, max_x = min(xs) - width, max(xs) + width
        min_y, max_y = min(ys) - width, max(ys) + width
        width = max_x - min_x
        height = max_y - min_y

        svg_content = '\n  '.join(edges_svg + nodes_svg)
        svg = (
            f"<svg xmlns='http://www.w3.org/2000/svg' "
            f"viewBox='{min_x} {min_y} {width} {height}' "
            f"width='{width}' height='{height}'>\n  {svg_content}\n</svg>"
        )
        return svg


@test(tests=test_cases[:])
def partI(notes: str, debug: bool = False) -> int:
    root = Node.create_tree(notes, debug)
    return root.checksum()
```

```python
with open('./inputs/everybody_codes_e3_q03_p1.txt') as f:
    notes1 = f.read()

print(f'Part I: {partI(notes1)}')
```

## <span style="color:#eed699">Part II</span>

The real process of building the tree may create two types of bonds: strong and weak. A **strong bond** is one where both the colour and the shape match. All bonds in Part I were therefore strong bonds. A **weak bond** is created when only the colour matches or only the shape matches between the plug and the socket. A node connects to the first socket encountered during the clockwise traversal that forms either a strong bond or a weak bond with its plug.

Veronica suggests practising this change using a slightly longer list of nodes. The list is guaranteed to connect all nodes into a single tree again. Your task is still to **build the tree and provide the checksum** in exactly the same way as before.

-Node `1` is first on the list, so it becomes the root of the tree.
-Node `2` creates a weak bond with the left socket of node `1`, based on the shape.
-Node `3` creates a weak bond with the left socket of node `2`, based on the colour.
-Node `4` creates a strong bond with the right socket of node `1`.
-Node `5` creates a weak bond with the left socket of node `4`, based on the shape.

After reading the ids of all nodes in the correct order, we obtain the following list: `3, 2, 1, 5, 4`.
The checksum is calculated the same way as in Part I: `1 * 3 + 2 * 2 + 3 * 1 + 4 * 5 + 5 * 4 = `<code style="color:white"> 50</code>.

<span style="color:#eed699">What is the checksum of the tree constructed from the list of nodes?<span>


```python
from test_utilities import test

test_cases = [
    # cSpell:disable
    {
        'name': 'Example I',
        'notes': """
            id=1, plug=RED TRIANGLE, leftSocket=RED TRIANGLE, rightSocket=RED TRIANGLE, data=?
            id=2, plug=GREEN TRIANGLE, leftSocket=BLUE CIRCLE, rightSocket=GREEN CIRCLE, data=?
            id=3, plug=BLUE PENTAGON, leftSocket=BLUE CIRCLE, rightSocket=GREEN CIRCLE, data=?
            id=4, plug=RED TRIANGLE, leftSocket=BLUE PENTAGON, rightSocket=GREEN PENTAGON, data=?
            id=5, plug=RED PENTAGON, leftSocket=GREEN CIRCLE, rightSocket=GREEN CIRCLE, data=?
         """,
        'expected': 50,
    },
    # cSpell:enable
]


class NodeII(Node):
    def add(self, node: Node) -> Node | None:
        if not self.left_socket:
            if sum(1 for a, b in zip(node.plug, self.left_preference) if a == b) > 0:
                self.left_socket = node
                return
        else:
            if not self.left_socket.add(node):
                return

        if not self.right_socket:
            if sum(1 for a, b in zip(node.plug, self.right_preference) if a == b) > 0:
                self.right_socket = node
                return
        else:
            if not self.right_socket.add(node):
                return

        return node


@test(tests=test_cases[:])
def partII(notes: str, debug: bool = False) -> int:
    root = NodeII.create_tree(notes, debug)
    return root.checksum()
```

```python
with open('./inputs/everybody_codes_e3_q03_p2.txt') as f:
    notes2 = f.read()


print(f'Part II: {partII(notes2)}')
```

## <span style="color:#eed699">Part III</span>

You’ve now mastered all the rules for connecting nodes into trees! The real process differs from what you’ve practised in just one small detail: **strong bonds can break weak bonds.**

If a node circling the tree encounters a socket with which it can form a strong bond, and that socket is currently occupied by another node via a weak bond, the weak bond is broken and replaced by a strong bond with the new node. **The detached node continues circling the tree to find a new place to connect**. It resumes its search from the next position on the clockwise traversal, immediately after the newly formed strong bond.

The list is guaranteed to connect all nodes into a single tree. Your task is again to **build the tree and provide the checksum** in exactly the same way as before.

It finds a matching socket at the right socket of node `1`, creating a weak bond based on the shape.

- Node `1` is first on the list, so it becomes the root of the tree.
- Node `2` creates a weak bond with the left socket of node `1`, based on the shape.
- Node `3` creates a weak bond with the left socket of node `2`, based on the colour.
- Node `4` creates a weak bond with the right socket of node `1`, based on the shape.
- Node `5` creates a strong bond with the left socket of node `1`. Node 2 is detached and creates a strong bond with the right socket of node `1`. As a result, node `4` is detached and creates a weak bond with the right socket of node `5`. Note that node `4` circles the tree starting from the point where it was detached, and must pass by the root node as well in order to find its target position.
- Node `6` creates a weak bond with the left socket of node `4`, based on the colour.

After reading the ids of all nodes in the correct order, we obtain the following list: `5, 6, 4, 1, 3, 2`.
The checksum is calculated the same way as in previous parts: `1 * 5 + 2 * 6 + 3 * 4 + 4 * 1 + 5 * 3 + 6 * 2 = `<code style="color:white"> 60</code>.

<span style="color:#eed699">What is the checksum of the tree constructed from the list of nodes?<span>


```python
from test_utilities import test

test_cases = [
    # cSpell:disable
    {
        'name': 'Example I',
        'notes': """
            id=1, plug=RED TRIANGLE, leftSocket=RED TRIANGLE, rightSocket=RED TRIANGLE, data=?
            id=2, plug=GREEN TRIANGLE, leftSocket=BLUE CIRCLE, rightSocket=GREEN CIRCLE, data=?
            id=3, plug=BLUE PENTAGON, leftSocket=BLUE CIRCLE, rightSocket=GREEN CIRCLE, data=?
            id=4, plug=RED TRIANGLE, leftSocket=BLUE PENTAGON, rightSocket=GREEN PENTAGON, data=?
            id=5, plug=RED PENTAGON, leftSocket=GREEN CIRCLE, rightSocket=GREEN CIRCLE, data=?
         """,
        'expected': 38,
    },
    {
        'name': 'Example II',
        'notes': """
            id=1, plug=RED TRIANGLE, leftSocket=BLUE TRIANGLE, rightSocket=GREEN TRIANGLE, data=?
            id=2, plug=GREEN TRIANGLE, leftSocket=BLUE CIRCLE, rightSocket=GREEN CIRCLE, data=?
            id=3, plug=BLUE PENTAGON, leftSocket=BLUE CIRCLE, rightSocket=GREEN CIRCLE, data=?
            id=4, plug=RED TRIANGLE, leftSocket=BLUE PENTAGON, rightSocket=GREEN PENTAGON, data=?
            id=5, plug=BLUE TRIANGLE, leftSocket=GREEN CIRCLE, rightSocket=RED CIRCLE, data=?
            id=6, plug=BLUE TRIANGLE, leftSocket=GREEN CIRCLE, rightSocket=RED CIRCLE, data=?
         """,
        'expected': 60,
    },
    # cSpell:enable
]


class NodeIII(Node):
    def add(self, node: Node) -> Node | None:
        if not self.left_socket:
            if sum(1 for a, b in zip(node.plug, self.left_preference) if a == b) > 0:
                self.left_socket = node
                return

        elif sum(1 for a, b in zip(node.plug, self.left_preference) if a == b) > sum(
            1 for a, b in zip(self.left_socket.plug, self.left_preference) if a == b
        ):
            node1 = self.left_socket
            self.left_socket = node
            node = node1
        else:
            node1 = self.left_socket.add(node)
            if not node1:
                return
            node = node1

        if not self.right_socket:
            if sum(1 for a, b in zip(node.plug, self.right_preference) if a == b) > 0:
                self.right_socket = node
                return

        elif sum(1 for a, b in zip(node.plug, self.right_preference) if a == b) > sum(
            1 for a, b in zip(self.right_socket.plug, self.right_preference) if a == b
        ):
            node1 = self.right_socket
            self.right_socket = node
            node = node1
        else:
            node1 = self.right_socket.add(node)
            if not node1:
                return

            node = node1

        return node


@test(tests=test_cases[:])
def partIII(notes: str, debug: bool = False) -> int:
    root = NodeIII.create_tree(notes, debug)
    return root.checksum()
```

```python
with open('./inputs/everybody_codes_e3_q03_p3.txt') as f:
    notes3 = f.read()


print(f'Part III: {partIII(notes3)}')
```
