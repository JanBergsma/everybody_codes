As a quick reminder, sparkballs are a favourite trick of pyromancers. They launch glowing spheres filled with beetles whose wings have been coated with a special luminescent powder. The sphere bursts high in the sky, releasing the beetles, which fly off in every direction and scatter the glowing powder, creating a spectacular display overhead.

Your goal is to prepare the coolest possible sparkball launch plan! To make planning easier, you decide to treat the sky as an infinitely large grid of sky squares. **This allows every sky square to be described by two coordinates**, `X` and `Y`. Your plan is to launch one giant sparkball at the coordinates `START`, together with three beetle beacons: `A`, `B`, and `C`. The beacons are configured to attract the beetles according to the sequence of ` MOVES` .

For Sir Sierpiński's method to work correctly, each beacon in the sequence **must stay active until the swarm has travelled exactly halfway towards it**, measured from the moment that step of the sequence began. At that point, all beacons briefly switch off so the beetles stop and illuminate one sky square with the powder. The next beacon in the sequence is then activated. After the final step, no further signals are sent, and the beetles scatter in every direction.

To simplify the calculations, all coordinates are treated as integers. **Any fractional values are always rounded down to the nearest integer**. To see how this works in practice, you decide to test a small launch plan <span style="color:#eed699">( your notes )</span> and determine **how many unique sky squares become illuminated**.

Following the whole sequence makes the swarm visit these sky squares: `[5,0], [2,0], [6,0], [5,5], [5,7], [7,3], [3,1], [6,0], [5,5], [2,2]`
The sky squares `[6,0]` and `[5,5]` appear twice, **but each illuminated sky square is counted only once**. Once the entire illumination plan has been completed, there are <code style="color:white"> 8</code> illuminated sky squares.

<span style="color:#eed699">How many unique sky squares become illuminated by your launch plan?<span>


```python
import math
import re
from itertools import batched

import matplotlib.animation as animation
import matplotlib.pyplot as plt
from IPython.display import HTML, display
from matplotlib.patches import Circle
from more_itertools import minmax
from test_utilities import test
from util import Str, hexcolor_str

test_cases = [
    # cSpell:disable
    {
        'name': 'Example I',
        'notes': """
            START=[5,0]
            A=[0,0]
            B=[10,0]
            C=[5,10]
            MOVES=ABCCBABCA
         """,
        'simulate': False,
        'expected': 8,
    },
    # cSpell:enable
]

type Point = tuple[int, int]


class LaunchPlan(Str):
    def __init__(self, notes: str) -> None:
        self.beacons: dict[str, Point] = dict(
            zip(
                ('START', 'A', 'B', 'C'),
                batched(map(int, re.findall(r'\d+', notes)), 2, strict=True),
            )
        )
        self.start = self.beacons['START']
        del self.beacons['START']
        self.moves = notes.strip().splitlines()[-1].strip().removeprefix('MOVES=')

    def calculate(self, debug=False) -> int:
        illuminated = set()
        swarm = self.start

        if debug:
            print(self.debug(0, '_', swarm, illuminated))

        for step, beacon in enumerate(self.moves, start=1):
            illuminated.add(swarm)
            x_s, y_s = swarm
            x_b, y_b = self.beacons[beacon]
            swarm = (x_s + x_b) // 2, (y_s + y_b) // 2

            if debug:
                print(self.debug(step, beacon, swarm, illuminated))

            illuminated.add(swarm)

        return len(illuminated)

    def simulate(self) -> int:
        step = 0
        beacon = '-'
        steps_data = [
            {'step': step, 'beacon': beacon, 'swarm': self.start},
        ]

        illuminated = set()
        swarm = self.start

        for step, beacon in enumerate(self.moves, start=1):
            illuminated.add(swarm)
            x_s, y_s = swarm
            x_b, y_b = self.beacons[beacon]
            swarm = (x_s + x_b) // 2, (y_s + y_b) // 2

            illuminated.add(swarm)
            steps_data.append({'step': step, 'beacon': beacon, 'swarm': swarm})

        self.show_simulation(steps_data, set())
        return len(illuminated)

    def show_simulation(self, steps_data, fireflies):
        # 3. Plot setup
        fig, ax = plt.subplots(figsize=(7, 7))

        min_x, max_x = minmax(x for x, _ in self.beacons.values())
        min_y, max_y = minmax(y for _, y in self.beacons.values())

        min_x, max_x = min_x - 1, max_x + 1
        min_y, max_y = min_y - 1, max_y + 1

        ax.set_xlim(min_x - 1, max_x + 2)
        ax.set_ylim(min_y - 1, max_y + 2)

        ax.set_aspect('equal')
        ax.grid(True, which='both', linestyle='--', alpha=0.5)

        # Beacons
        for name, (x, y) in self.beacons.items():
            ax.scatter(x, y, color='blue', s=180, zorder=3, label='Beacon' if name == 'A' else '')
            ax.text(x, y + 0.35, name, color='blue', fontsize=12, ha='center', weight='bold')

        # Ring highlight for the active beacon
        beacon_highlights = {}
        for name, (x, y) in self.beacons.items():
            ring = Circle(
                (x, y), 0.55, facecolor='none', edgecolor='gold', linewidth=2.5, alpha=0.0, zorder=4
            )
            ax.add_patch(ring)
            beacon_highlights[name] = ring

        # Trail and swarm
        trail_x, trail_y = [], []
        (trail_plot,) = ax.plot([], [], 'rx', markersize=7, label='Trail')
        (swarm_plot,) = ax.plot([], [], 'go', markersize=12, markeredgewidth=2, label='Swarm')
        (fireflies_plot,) = ax.plot([], [], 'y+', markersize=7, label='Fireflies')

        # Status text
        title_text = ax.text(
            0.05, 0.95, '', transform=ax.transAxes, fontsize=11, verticalalignment='top'
        )

        ax.legend(loc='upper right')

        # 4. Animation update
        def update(frame):
            data = steps_data[frame]
            current_swarm = data['swarm']
            active_beacon = data['beacon']

            # Add previous swarm position to trail
            if frame > 0:
                prev_swarm = steps_data[frame - 1]['swarm']
                prev_point = (prev_swarm[0], prev_swarm[1])
                if prev_point not in list(zip(trail_x, trail_y)):
                    trail_x.append(prev_swarm[0])
                    trail_y.append(prev_swarm[1])

            # Update beacon highlight
            for name, ring in beacon_highlights.items():
                ring.set_alpha(0.0 if name != active_beacon else 1.0)

            # Distance to selected beacon
            if active_beacon != '-':
                bx, by = self.beacons[active_beacon]
                distance = math.hypot(current_swarm[0] - bx, current_swarm[1] - by)
            else:
                distance = None

            # Update plot data
            trail_plot.set_data(trail_x, trail_y)
            swarm_plot.set_data([current_swarm[0]], [current_swarm[1]])
            if 'fireflies' in data:
                fireflies_plot.set_data([x for x, _ in fireflies], [y for _, y in fireflies])

            # Status text
            if distance is None:
                distance_text = 'N/A'
            else:
                distance_text = f'{distance:.2f}'

            title_text.set_text(
                f'Step: {data["step"]}\n'
                f'Beacon: {active_beacon}\n'
                f'Swarm: {current_swarm}\n'
                f'Distance: {distance_text}'
            )

            return trail_plot, swarm_plot, fireflies_plot, title_text, *beacon_highlights.values()

        # 5. Create animation
        ani = animation.FuncAnimation(
            fig, update, frames=len(steps_data), interval=1000, blit=True, repeat=False
        )

        plt.close(fig)

        # 6. Render in notebook
        display(HTML(ani.to_jshtml()))

    def debug(self, step: int, beacon: str, swarm: Point, illuminated: set[Point]) -> str:
        min_x, max_x = minmax(x for x, _ in self.beacons.values())
        min_y, max_y = minmax(y for _, y in self.beacons.values())

        min_x, max_x = min_x - 1, max_x + 1
        min_y, max_y = min_y - 1, max_y + 1

        width_y = len(str(max_y))

        points_to_beacons = {v: k for k, v in self.beacons.items()}
        return '\n'.join(
            (
                f'step: {step} {"(init)" if step == 0 else ""}',
                f'beacon: {beacon}',
                f'swarm: [{swarm[0], swarm[1]}]',
                '',
                '\n'.join(
                    f'{y:.>{width_y}}'
                    + ''.join(
                        hexcolor_str('#00FF00', '#')
                        if (x, y) == swarm
                        else hexcolor_str('#008080', 'X')
                        if (x, y) in illuminated
                        else hexcolor_str(
                            '#00BFFF' if (x, y) in points_to_beacons else '#FFFFFF',
                            points_to_beacons.get((x, y), '.'),
                        )
                        for x in range(min_x, max_x + 1)
                    )
                    for y in range(max_y, min_y - 1, -1)
                ),
                '.' * width_y + ''.join(f'{x}'[-1] for x in range(min_x, max_x + 1)),
                '.' * width_y
                + ''.join(
                    f'{x // 10}'[-1] if x // 10 and not x % 10 else '.'
                    for x in range(min_x, max_x + 1)
                ),
                '',
            )
        )


@test(tests=test_cases[:])
def partI(notes: str, debug: bool = False, simulate=False) -> int:
    lp = LaunchPlan(notes)
    if simulate:
        return lp.simulate()
    return lp.calculate(debug)
```

```python
from test_utilities import test

test_cases = [
    # cSpell:disable
    {
        'name': 'Example I',
        'notes': """
            START=[5,0]
            A=[0,0]
            B=[10,0]
            C=[5,10]
            MOVES=ABCCBABCA
         """,
        'debug': False,
        'simulate': False,
        'plot_result': True,
        'expected': 25,
    },
    {
        'name': 'Example II',
        'notes': """
            START=[5,0]
            A=[0,0]
            B=[10,0]
            C=[5,10]
            MOVES=BABCAABBCABCCCBBABCCCAAACABABCBCBBCAABBABBCACCBAABCBCBBBCBBBBBCCCAACAACB
         """,
        'debug': False,
        'simulate': False,
        'plot_result': True,
        'expected': 46,
    },
    # cSpell:enable
]


class LaunchPlan2(LaunchPlan):
    AROUND = (-1, 0), (0, 1), (1, 0), (0, -1)

    def calculate(self, debug=False, plot_result=False) -> int:
        illuminated = set()
        fireflies = set()
        swarm = self.start
        step = 0
        beacon = '_'

        if debug:
            print(self.debug(step, beacon, swarm, illuminated))

        for step, beacon in enumerate(self.moves, start=1):
            illuminated.add(swarm)
            x_s, y_s = swarm
            fireflies |= {(x_s + dx, y_s + dy) for dx, dy in self.AROUND}

            x_b, y_b = self.beacons[beacon]
            swarm = (x_s + x_b) // 2, (y_s + y_b) // 2

            if debug:
                print(self.debug(step, beacon, swarm, illuminated))

        illuminated.add(swarm)
        x_s, y_s = swarm
        fireflies |= {(x_s + dx, y_s + dy) for dx, dy in self.AROUND}
        fireflies -= illuminated

        if debug:
            print(self.debug(step, beacon, (-1, -1), illuminated, fireflies))
        if plot_result:
            self.plot(illuminated, fireflies)

        return len(fireflies)

    def simulate(self) -> int:
        step = 0
        beacon = '-'

        steps_data = [
            {'step': step, 'beacon': beacon, 'swarm': self.start},
        ]

        illuminated = set()
        fireflies = set()

        swarm = self.start

        for step, beacon in enumerate(self.moves, start=1):
            illuminated.add(swarm)
            x_s, y_s = swarm
            fireflies |= {(x_s + dx, y_s + dy) for dx, dy in self.AROUND}

            x_b, y_b = self.beacons[beacon]
            swarm = (x_s + x_b) // 2, (y_s + y_b) // 2

            steps_data.append({'step': step, 'beacon': beacon, 'swarm': swarm})

        illuminated.add(swarm)

        x_s, y_s = swarm
        fireflies |= {(x_s + dx, y_s + dy) for dx, dy in self.AROUND}
        fireflies -= illuminated

        steps_data.append({'step': step, 'beacon': beacon, 'swarm': swarm, 'fireflies': fireflies})

        self.show_simulation(steps_data, fireflies)

        return len(fireflies)

    def plot(self, illuminated: set[Point], fireflies: set[Point] | None = None) -> None:
        if fireflies is None:
            fireflies = set()

        points_to_beacons = {v: k for k, v in self.beacons.items()}
        beacon_points = set(points_to_beacons)
        points = beacon_points | illuminated | fireflies

        min_x, max_x = minmax(x for x, _ in points)
        min_y, max_y = minmax(y for _, y in points)
        fig, ax = plt.subplots(figsize=(7, 7))
        ax.set_xlim(min_x - 0.5, max_x + 0.5)
        ax.set_ylim(min_y - 0.5, max_y + 0.5)
        ax.set_xticks(range(min_x, max_x + 1))
        ax.set_yticks(range(min_y, max_y + 1))
        ax.set_aspect('equal', adjustable='box')
        ax.grid(True, linestyle='--', alpha=0.5)
        ax.set_axisbelow(True)
        ax.set_xlabel('X')
        ax.set_ylabel('Y')
        ax.set_title('Sierpiński triangle')

        visible_illuminated = illuminated
        visible_beacons = beacon_points - illuminated
        visible_fireflies = fireflies - illuminated - beacon_points

        for marker_points, marker, color, label in (
            (visible_fireflies, '+', 'gold', 'Fireflies'),
            (visible_illuminated, 'x', 'crimson', 'Illuminated'),
            (visible_beacons, 'o', 'deepskyblue', 'Beacons'),
        ):
            ax.scatter(
                [x for x, _ in marker_points],
                [y for _, y in marker_points],
                marker=marker,
                color=color,
                label=label,
                zorder=2,
            )

        for position, name in points_to_beacons.items():
            if position in visible_beacons:
                ax.annotate(name, position, xytext=(4, 4), textcoords='offset points')

        ax.legend(loc='best')
        plt.show()

    def debug(
        self,
        step: int,
        beacon: str,
        swarm: Point,
        illuminated: set[Point],
        fireflies: set[Point] | None = None,
    ) -> str:
        min_x, max_x = minmax(x for x, _ in self.beacons.values())
        min_y, max_y = minmax(y for _, y in self.beacons.values())

        min_x, max_x = min_x - 1, max_x + 1
        min_y, max_y = min_y - 1, max_y + 1

        width_y = len(str(max_y))

        if fireflies is None:
            fireflies = set()

        points_to_beacons = {v: k for k, v in self.beacons.items()}
        return '\n'.join(
            (
                f'step: {step} {"(init)" if step == 0 else ""}',
                f'beacon: {beacon}',
                f'swarm: [{swarm[0], swarm[1]}]',
                '',
                '\n'.join(
                    f'{y:.>{width_y}}'
                    + ''.join(
                        hexcolor_str('#00FF00', '#')
                        if (x, y) == swarm
                        else hexcolor_str('#DC143C', 'X')
                        if (x, y) in illuminated
                        else hexcolor_str(
                            '#00BFFF' if (x, y) in points_to_beacons else '#FFFFFF',
                            points_to_beacons.get((x, y), '.'),
                        )
                        if (x, y) not in fireflies
                        else hexcolor_str('#FFFF00', 'F')
                        for x in range(min_x, max_x + 1)
                    )
                    for y in range(max_y, min_y - 1, -1)
                ),
                '.' * width_y + ''.join(f'{x}'[-1] for x in range(min_x, max_x + 1)),
                '.' * width_y
                + ''.join(
                    f'{x // 10}'[-1] if x // 10 and not x % 10 else '.'
                    for x in range(min_x, max_x + 1)
                ),
                '',
            )
        )


@test(tests=test_cases[:])
def partII(notes: str, debug: bool = False, simulate=False, plot_result=False) -> int:
    lp = LaunchPlan2(notes)
    if simulate:
        return lp.simulate()
    return lp.calculate(debug, plot_result)
```

```python
from collections import deque
from itertools import chain

from test_utilities import test

test_cases = [
    # cSpell:disable
    {
        'name': 'Example I',
        'notes': """
            START=[5,0]
            A=[0,0]
            B=[10,0]
            C=[5,10]
         """,
        'debug': True,
        'simulate': False,
        'plot_result': False,
        'expected': 42,
    },
    {
        'name': 'Example II',
        'notes': """
            START=[0,0]
            A=[0,0]
            B=[80,15]
            C=[5,30]
         """,
        'debug': True,
        'simulate': False,
        'plot_result': False,
        'expected': 432,
    },
    # cSpell:enable
]


def dist(a: Point, b: Point) -> int:
    return abs(a[0] - b[0]) + abs(a[1] - b[1])


class LaunchPlan3(LaunchPlan2):
    def __init__(self, notes: str) -> None:
        super().__init__(notes)
        self.moves = None

    def calculate(self, debug=False, plot_result=False) -> int:
        illuminated = set()
        beacons = self.beacons.values()

        stack = deque([self.start])

        while stack:
            point = stack.pop()
            if point in illuminated:
                continue

            illuminated.add(point)
            x, y = point

            stack.extend(((x + xb) // 2, (y + yb) // 2) for xb, yb in beacons)

        fireflies = {
            (x + dx, y + dy)
            for x, y in illuminated
            for dx, dy in self.AROUND
            if (x + dx, y + dy) not in illuminated
        }

        if plot_result:
            self.plot(illuminated, fireflies)

        if debug:
            print(self.ascii_plot(illuminated, fireflies))

        return len(fireflies)

    def simulate(self) -> int:
        step = 0
        beacon = '-'

        steps_data = [
            {'step': step, 'beacon': beacon, 'swarm': self.start},
        ]

        illuminated = set()
        fireflies = set()

        swarm = self.start

        stack = deque([(beacon, swarm)])

        while stack:
            beacon, swarm = stack.popleft()
            # beacon, swarm = stack.pop()
            # choose between popleft and pop,
            # queue or stack in simulation you can see the difference

            if swarm in illuminated:
                continue

            step += 1

            steps_data.append({'step': step, 'beacon': beacon, 'swarm': swarm})

            illuminated.add(swarm)
            x, y = swarm

            stack.extend(
                (b, ((x + xb) // 2, (y + yb) // 2)) for b, (xb, yb) in self.beacons.items()
            )

        fireflies = {
            (x + dx, y + dy)
            for x, y in illuminated
            for dx, dy in self.AROUND
            if (x + dx, y + dy) not in illuminated
        }

        steps_data.append({'step': step, 'beacon': beacon, 'swarm': swarm, 'fireflies': fireflies})

        self.show_simulation(steps_data, fireflies)

        return len(fireflies)

    def ascii_plot(
        self,
        illuminated: set[Point],
        fireflies: set[Point] | None = None,
    ) -> str:
        if fireflies is None:
            fireflies = set()

        min_x, max_x = minmax(x for x, _ in chain(self.beacons.values(), illuminated, fireflies))
        min_y, max_y = minmax(y for _, y in chain(self.beacons.values(), illuminated, fireflies))

        min_x, max_x = min_x - 1, max_x + 1
        min_y, max_y = min_y - 1, max_y + 1

        width_y = len(str(max_y))

        points_to_beacons = {v: k for k, v in self.beacons.items()}
        return '\n'.join(
            (
                '\n'.join(
                    f'{y:.>{width_y}}'
                    + ''.join(
                        hexcolor_str('#DC143C', 'X')
                        if (x, y) in illuminated
                        else hexcolor_str(
                            '#00BFFF' if (x, y) in points_to_beacons else '#FFFFFF',
                            points_to_beacons.get((x, y), '.'),
                        )
                        if (x, y) not in fireflies
                        else hexcolor_str('#FFFF00', 'F')
                        for x in range(min_x, max_x + 1)
                    )
                    for y in range(max_y, min_y - 1, -1)
                ),
                '.' * width_y + ''.join(f'{x}'[-1] for x in range(min_x, max_x + 1)),
                '.' * width_y
                + ''.join(
                    f'{x // 10}'[-1] if x // 10 and not x % 10 else '.'
                    for x in range(min_x, max_x + 1)
                ),
                '',
            )
        )


@test(tests=test_cases[:])
def partIII(notes: str, debug: bool = False, simulate=False, plot_result=False) -> int:
    lp = LaunchPlan3(notes)

    if debug:
        print(lp)

    if simulate:
        return lp.simulate()

    return lp.calculate(debug=debug, plot_result=plot_result)
```
