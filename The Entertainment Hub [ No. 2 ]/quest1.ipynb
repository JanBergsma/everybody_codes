### <span style="color:#eed699">Part I</span>

> ## <span style="color:#eed699"> <img src='./images/book.svg' height=40 width=40 style="display:inline-block height:fit-content;vertical-align: middle; margin: 5px; ">Story section</span>
>
> The Order has a tendency to meticulously inspect the mechanisms of my machines. I have a plan to exploit this inspection process to keep them busy for a while. The machine I'm working on will have a gazillion gears inside, which ultimately do absolutely nothing. The entire mechanism determining the outcome of the game will be packed into small wooden tokens used for the game. Before those nosy knights come up with the idea of checking this element of the game, my pouches will already be stuffed to the brim with coins!
>
> The machine is essentially just ordinary wooden boards with nails driven in a specific way. The spaces between these nails are large enough for wooden tokens, which are dropped from the top, to move between them. The tokens bounce off the nails to the right or left, falling to lower and lower levels. The place from which the token falls and where the tokens ultimately land determines the number of coins won.
>
> <br/>

**Each machine has its own individual nail pattern**, which can be very quickly changed in case of an inspection by the Order. Nails on each diagram are marked as `*`, and the rest is just empty space. At the top of the machine, at every odd column, there are toss slots, where the tokens are inserted. The slots are numbered in order from left to right: 1, 2, 3, etc. At the bottom of the machine, there are final slots, directly above the toss slots, numbered in the same way. And now my favorite part - the game tokens. When the falling token hits a nail, it bounces to the left or right by exactly one column on the schematic. At first glance, they appear to be ordinary wooden tokens, but the mechanism I embedded in them allows me to control the gameplay to a large extent. **In each token, there is a sequence of behaviors**, and it is this sequence, not blind fate, that determines whether the token will bounce to the left `L` or right `R` upon hitting the next nail in its path. It is worth adding here that if the nail is near the wall, the token can also bounce off that wall. This means that you can only bounce off such nails in one direction, opposite to the wall.

```python
import re

from tabulate import tabulate
from test_utilities import test
from util import hexcolor_str

test_cases = [
    # cSpell:disable
    {
        'name': 'Example',
        'notes': """
             *.*.*.*.*
            .*.*.*.*.
            *...*.*.*
            .*.*...*.
            *.*...*..
            .*.*.*.*.

            RLLRLR
            RLLRLR
            RLLRLR
            RLLRLR
            RLLRLR
         """,
        'expected': 15,
    },
    {
        'name': 'Example I',
        'notes': """
             *.*.*.*.*.*.*.*.*
             .*.*.*.*.*.*.*.*.
             *.*.*...*.*...*..
             .*.*.*.*.*...*.*.
             *.*.....*...*.*.*
             .*.*.*.*.*.*.*.*.
             *...*...*.*.*.*.*
             .*.*.*.*.*.*.*.*.
             *.*.*...*.*.*.*.*
             .*...*...*.*.*.*.
             *.*.*.*.*.*.*.*.*
             .*.*.*.*.*.*.*.*.

             RRRLRLRRRRRL
             LLLLRLRRRRRR
             RLLLLLRLRLRL
             LRLLLRRRLRLR
             LLRLLRLLLRRL
             LRLRLLLRRRRL
             LRLLLLLLRLLL
             RRLLLRLLRLRR
             RLLLLLRLLLRL
         """,
        'expected': 26,
    },
    # cSpell:enable
]


@test(tests=test_cases[:])
def partI(notes: str, debug: bool = False, debug_slots=False) -> int:
    diagram, tokens = re.split(r'(?:\r?\n){2,}', notes.strip())
    diagram = [line.strip() for line in diagram.splitlines()]
    tokens = [token.strip() for token in tokens.splitlines()]

    assert all(len(token) == len(diagram) for token in tokens)

    total_coins_won = 0

    if debug:
        print(tabulate(diagram, tablefmt='plain'))
        print()
        print(tabulate(tokens, tablefmt='plain'))
        print()
        print(f'{" " * (len(tokens[0]) - 5)}Token   Toss Slot   Final Slot   Coins Won')

    for toss_number, token in enumerate(tokens, start=1):
        col = 2 * (toss_number - 1)

        if debug_slots:
            visited = {(0, col)}
            token_indici = []

        current_bounce = 0

        for row in range(len(diagram)):
            bounce = token[current_bounce]

            if debug_slots:
                visited.add((row, col))  # type: ignore
                token_indici.append(current_bounce if diagram[row][col] == '*' else -1)  # type: ignore

            if diagram[row][col] == '*':
                col += 1 if bounce == 'R' else -1

                if col < 0:
                    col = 1
                elif col == len(diagram[0]):
                    col = len(diagram[0]) - 2

                current_bounce += 1

            if debug_slots:
                visited.add((row, col))  # type: ignore

        if debug_slots:
            print()
            print(
                '\n'.join(
                    '\t'.join(
                        (
                            ''.join(
                                hexcolor_str('#00ff00', ch) if (r, c) in visited else ch  # type: ignore
                                for c, ch in enumerate(row)
                            ),
                            f'{token[token_indici[r]] if token_indici[r] != -1 else ""}',  # type: ignore
                        )
                    )
                    for r, row in enumerate(diagram)
                )
            )
            print()

        final_slot = col // 2 + 1
        coins_won = max(0, 2 * final_slot - toss_number)
        total_coins_won += coins_won

        if debug:
            print(f'{token} {toss_number:>{11}d} {(final_slot):>{11}d} {coins_won:>{10}d}')

    return total_coins_won
```

```python
# OO version

test_cases += [
    {
        'name': 'Part I',
        'notes': notes1,
        'expected': 51,
    },
]


class Diagram_I:
    def __init__(self, notes: str, debug_total: bool = False, debug_slots: bool = False) -> None:
        diagram, tokens = re.split(r'(?:\r?\n){2,}', notes.strip())
        self.diagram = [line.strip() for line in diagram.splitlines()]
        self.tokens = [token.strip() for token in tokens.splitlines()]

        assert all(len(token) == len(self.diagram) for token in self.tokens)

        self.debug_total_number_of_coins = debug_total
        self.debug_total_number_of_coins_per_token = debug_slots

    def total_number_of_coins(self) -> int:
        total_coins_won = 0

        if self.debug_total_number_of_coins:
            self.debug_total_number_of_coins_header()

        for toss_number, token in enumerate(self.tokens, start=1):
            col = self.total_number_of_coins_per_token(toss_number, token)

            final_slot = col // 2 + 1
            coins_won = max(0, 2 * final_slot - toss_number)
            total_coins_won += coins_won

            if self.debug_total_number_of_coins:
                self.debug_total_number_of_coins_totals(toss_number, token, final_slot, coins_won)

        return total_coins_won

    def debug_total_number_of_coins_totals(self, toss_number, token, final_slot, coins_won):
        print(f'{token} {toss_number:>{11}d} {(final_slot):>{11}d} {coins_won:>{10}d}')

    def debug_total_number_of_coins_header(self):
        print(tabulate(self.diagram, tablefmt='plain'))
        print()
        print(tabulate(self.tokens, tablefmt='plain'))
        print()
        print(f'{" " * (len(self.tokens[0]) - 5)}Token   Toss Slot   Final Slot   Coins Won')

    def total_number_of_coins_per_token(self, toss_number, token):
        col = 2 * (toss_number - 1)

        if self.debug_total_number_of_coins_per_token:
            visited, token_indici = self.debug_total_number_of_coins_per_token_initialize(col)

        current_bounce = 0

        for row in range(len(self.diagram)):
            bounce = token[current_bounce]

            if self.debug_total_number_of_coins_per_token:
                visited.add((row, col))  # type: ignore
                token_indici.append(current_bounce if self.diagram[row][col] == '*' else -1)  # type: ignore

            if self.diagram[row][col] == '*':
                col += 1 if bounce == 'R' else -1

                if col < 0:
                    col = 1
                elif col == len(self.diagram[0]):
                    col = len(self.diagram[0]) - 2

                current_bounce += 1

            if self.debug_total_number_of_coins_per_token:
                visited.add((row, col))  # type: ignore

        if self.debug_total_number_of_coins_per_token:
            self.debug_total_number_of_coins_per_token_diagram(token, visited, token_indici)  # type: ignore

        return col

    def debug_total_number_of_coins_per_token_diagram(self, token, visited, token_indici):
        print()
        print(
            '\n'.join(
                '\t'.join(
                    (
                        ''.join(
                            hexcolor_str('#00ff00', ch) if (r, c) in visited else ch  # type: ignore
                            for c, ch in enumerate(row)
                        ),
                        f'{token[token_indici[r]] if token_indici[r] != -1 else ""}',  # type: ignore
                    )
                )
                for r, row in enumerate(self.diagram)
            )
        )
        print()

    def debug_total_number_of_coins_per_token_initialize(self, col):
        visited = {(0, col)}
        token_indici = []
        return visited, token_indici


@test(tests=test_cases[:])
def partI_OO(notes: str, debug: bool = False, debug_slots=False) -> int:
    diagram = Diagram_I(notes, debug_total=debug, debug_slots=debug_slots)
    return diagram.total_number_of_coins()
```

```python
from functools import cache

from more_itertools import minmax
from test_utilities import test

test_cases = [
    # cSpell:disable
    {
        'name': 'Example I',
        'notes': """
            *.*.*.*.*.*.*.*.*
            .*.*.*.*.*.*.*.*.
            *.*.*...*.*...*..
            .*.*.*.*.*...*.*.
            *.*.....*...*.*.*
            .*.*.*.*.*.*.*.*.
            *...*...*.*.*.*.*
            .*.*.*.*.*.*.*.*.
            *.*.*...*.*.*.*.*
            .*...*...*.*.*.*.
            *.*.*.*.*.*.*.*.*
            .*.*.*.*.*.*.*.*.

            RRRLRLRRRRRL
            LLLLRLRRRRRR
            RLLLLLRLRLRL
            LRLLLRRRLRLR
            LLRLLRLLLRRL
            LRLRLLLRRRRL
         """,
        'expected': '13 43',
    },
    {
        'name': 'Example II',
        'notes': """
            *.*.*.*.*.*.*.*.*.*.*.*.*
            .*.*.*.*.*.*.*.*.*.*.*.*.
            ..*.*.*.*...*.*...*.*.*..
            .*...*.*.*.*.*.*.....*.*.
            *.*...*.*.*.*.*.*...*.*.*
            .*.*.*.*.*.*.*.*.......*.
            *.*.*.*.*.*.*.*.*.*...*..
            .*.*.*.*.*.*.*.*.....*.*.
            *.*...*.*.*.*.*.*.*.*....
            .*.*.*.*.*.*.*.*.*.*.*.*.
            *.*.*.*.*.*.*.*.*.*.*.*.*
            .*.*.*.*.*.*.*.*.*...*.*.
            *.*.*.*.*.*.*.*.*...*.*.*
            .*.*.*.*.*.*.*.*.....*.*.
            *.*.*.*.*.*.*.*...*...*.*
            .*.*.*.*.*.*.*.*.*.*.*.*.
            *.*.*...*.*.*.*.*.*.*.*.*
            .*...*.*.*.*...*.*.*...*.
            *.*.*.*.*.*.*.*.*.*.*.*.*
            .*.*.*.*.*.*.*.*.*.*.*.*.

            RRRLLRRRLLRLRRLLLRLR
            RRRRRRRRRRLRRRRRLLRR
            LLLLLLLLRLRRLLRRLRLL
            RRRLLRRRLLRLLRLLLRRL
            RLRLLLRRLRRRLRRLRRRL
            LLLLLLLLRLLRRLLRLLLL
         """,
        'expected': '25 66',
    },
    {
        'name': 'Example III',
        'notes': """
            *.*.*.*.*.*.*.*.*.*.*.*.*.*.*.*.*.*.*.*
            .*.*.*.*.*.*.*.*.*.*.*.*.*.*.*.*.*.*.*.
            ..*.*.*.*.*.*.........*.*.*.*.....*.*.*
            .*.*...*.*.*.*.*.*.*.*.*.*.*...*.*.*.*.
            *.*.*.*...*.*.*.*.*.....*.*.*.*...*.*..
            .*...*.*...*.*.*.*.*.*.*.....*.*.*.*.*.
            *.*.*.*.*.....*.*.*.*.*.*.*.*.*.*.*.*.*
            .*.*.*.*.*.*...*.*.*.*.....*.*.*.*...*.
            *.*...*.*.*.*.*.*.*.*...*.*.*...*.*.*.*
            .*...*.*.*.*.*.*.*.*...*.*.*.*.*.*.*.*.
            *.*.*.*.*.*...*.....*.*...*...*.*.*.*.*
            .*...*.*.*.*.*...*.*.*.*.*...*.*...*.*.
            *.*.*.*.*...*.*.*.*.*.*.*.*...*.*.*.*.*
            .*.*.*.*.*.*.*.*...*.*.*.*.*.*.*.*.*.*.
            ....*.*.*.*...*.*.*.*.*.*.*...*.*.*...*
            .*.*.*...*.*.*.*.*...*.*.*.*.*.*.*.*...
            *.*.*.*.*.*.*.....*...*...*.*.*.*.*.*.*
            .*.*...*.....*.*.*.*.*.*.*...*.*.*.*.*.
            *.*.*.*.*.*.*.*.*.*.*.*.*.*.*.*.*.*.*.*
            .*.*.*.*.*.*.*.*.*.*.*.*.*.*.*.*.*.*.*.

            RRRRLLRRLLLLLLLRLLRL
            RRRRRRRLRRLRRLRRRLRR
            RRRLLRRRRRLRRRRRLRRR
            LLLLRRLLRRLLLLLRRLLL
            LRRRRLRRLRLLRLLRRLRR
            RRRRRRRRLRRRRLLRRRLR
         """,
        'expected': '39 122',
    },
    # cSpell:enable
]


class Diagram_III(Diagram_I):
    def minmax_total_number_of_coins(self) -> tuple[int, int]:
        @cache
        def rec(token, tosses) -> tuple[int, int]:
            if token == len(self.tokens) - 1:
                return minmax(coins_won[token][toss] for toss in tosses)

            return minmax(
                coins_won[token][toss] + value
                for toss in tosses
                for value in rec(token + 1, tosses - {toss})
            )

        coins_won = self.create_coins_table()

        if self.debug_total_number_of_coins:
            print(
                tabulate(
                    [[token] + row for token, row in zip(self.tokens, coins_won)],
                    headers=['token'] + [str(tn) for tn in range(len(self.diagram[0]) // 2 + 2)],
                )
            )

        return rec(0, frozenset(range(1, len(self.diagram[0]) // 2 + 2)))

    def create_coins_table(self):
        coins_won = [[0] * (len(self.diagram[0]) // 2 + 2) for _ in self.tokens]

        for token_nr, token in enumerate(self.tokens):
            for toss_number in range(1, len(self.diagram[0]) // 2 + 2):
                col = self.total_number_of_coins_per_token(toss_number, token)

                final_slot = col // 2 + 1
                coins_won[token_nr][toss_number] = max(0, 2 * final_slot - toss_number)

        return coins_won


@test(tests=test_cases[:])
def partIII(notes: str, debug: bool = False, debug_slots: bool = False) -> str:
    diagram = Diagram_III(notes, debug_total=debug, debug_slots=debug_slots)
    return ' '.join(str(value) for value in diagram.minmax_total_number_of_coins())
```
