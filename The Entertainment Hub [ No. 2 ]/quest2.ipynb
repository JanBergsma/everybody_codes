Of course, the crossbow doesn't shoot real arrows. It is a design specifically created for shooting totally harmless fluffbolts. First, the crossbowman shoots a red fluffbolt, then a green one, followed by a blue one, after which he returns to the red one, executing the next **sequences of shots: red, green, blue; as many times as needed to pop all the balloons.**

Balloons also come in one of three colors, just like fluffbolts, but their arrangement is random (or at least, people think so). When a fluffbolt hits a balloon, a loud "POP!" is heard and colorful glitter scatters all around. Kids will pay to join this play for sure, but there's also a way to win some coins. The goal is to guess **how many fluffbolts need to be shot to destroy all the balloons**. The closest guesses will be rewarded.

Simple? Very simple! Well, maybe not entirely because if the fluffbolt hits a balloon of its own color, it continues flying and can pop subsequent balloons. If the colors of the fluffbolt and the balloon it hits are different, the fluffbolt pops along with it.
The simplest way to understand this is based on an example list of red `R`, green `G` and blue `B` balloons <span style="color:#eed699">( your notes )</span>. For now, we assume that the balloons are **lined up one behind the other** and the fluffbolt can pass through all of them if the colors match.

**Example based on the following notes:**

<pre style="border: 2px dotted lime;border-radius: 5px;padding:1rem;width: max-content;">
GRBGGGBBBRRRRRRRR
</pre>

Fluffbolts are shot from the left, in the order of red `R`, green `G` and blue `B`. Here is how each shoot affects the balloons:

<pre style="border: 1pt solid DimGray;border-radius: 5px;padding:1rem;width: max-content;">
Shoot     Fluffbolt         Ballons
    -                 GRBGGGBBBRRRRRRRR
    1            R    -RBGGGBBBRRRRRRRR
    2            G     -BGGGBBBRRRRRRRR
    3            B      --GGBBBRRRRRRRR
    4            R        -GBBBRRRRRRRR
    5            G         --BBRRRRRRRR
    6            B           ---RRRRRRR
    7            R              -------
</pre>

- The first fluffbolt is red `R` and it hits the green `G` balloon, so it stops there.
- The second fluffbolt is green `G` and it hits the red `R` balloon, so it also stops there.
- The third fluffbolt is blue `B` and it hits the blue `B` balloon, so it flies passing through it, popping the green `G` balloon after it.
- The next fluffbolt is red `R` and it hits the green `G` balloon.
- The next fluffbolt is green `G` and it hits the green `G` balloon, so it flies passing through it.
- The next fluffbolt is blue `B` and it hits the blue `B` balloon, so it flies passing through it, and through the next blue B balloon after it. Finally it pops the red `R` balloon.
- The last fluffbolt is red `R` and it pops the last seven red `R` balloons.

In total, <code style:"color:white"> 7</code> fluffbolts are needed to pop all the balloons.

<span style="color:#eed699"How many fluffbolts need to be shot to pop all the balloons?<span>


```python
from collections import deque
from itertools import cycle

from test_utilities import test

test_cases = [
    # cSpell:disable
    {
        'name': 'Example I',
        'notes': 'GRBGGGBBBRRRRRRRR',
        'expected': 7,
    },
    # cSpell:enable
]


@test(tests=test_cases[:])
def partI(notes: str, debug: bool = False) -> int:
    balloons = deque(notes.strip())
    fluffbolts = 0

    for fluffbolt in cycle('RGB'):
        fluffbolts += 1

        while balloons and balloons[0] == fluffbolt:
            balloons.popleft()

        if balloons:
            balloons.popleft()

        if not balloons:
            return fluffbolts

    raise AssertionError('Should not get here!')
```

```python
with open('./inputs/everybody_codes_e2_q02_p1.txt') as f:
    notes1 = f.read()

print(f'Part I: {partI(notes1)}')
```

During grand festivals, **balloons are arranged at perfectly equal intervals in the shape of a circle**. Fluffbolts are fired in such a way that they fly perfectly through the center of this circle. If the fluffbolt pops the first balloon and continues flying, it can pop the balloon directly opposite it, but only if the number of balloons before the shot was even. Otherwise, it flies between the two balloons directly opposite without popping them. After each shot, the balloons are rearranged in perfectly equal intervals in a circular shape.

- If we repeat it <code style="border: 2px dotted orange;border-radius: 5px;padding:0.1rem 1rem;width: max-content;"> 100</code> times, we will need <code style="color:white"> 2955</code> shots to pop all the balloons.

```python
import math

import drawsvg as draw
from IPython.display import SVG, display

test_cases = [
    # cSpell:disable
    {
        'name': 'Example I',
        'notes': 'GGBR',
        'repeat': 5,
        # 'debug': True,
        'expected': 14,
    },
    {
        'name': 'Example II repeat 10',
        'notes': 'BBRGGRRGBBRGGBRGBBRRBRRRBGGRRRBGBGG',
        'repeat': 10,
        'expected': 304,
    },
    {
        'name': 'Example II repeat 50',
        'notes': 'BBRGGRRGBBRGGBRGBBRRBRRRBGGRRRBGBGG',
        'repeat': 50,
        'expected': 1464,
    },
    {
        'name': 'Example II repeat 100',
        'notes': 'BBRGGRRGBBRGGBRGBBRRBRRRBGGRRRBGBGG',
        'repeat': 100,
        'expected': 2955,
    },
    # cSpell:enable
]


def deque_to_svg_circle(dq: deque[str], radius: int = 360) -> None:
    """
    Render a deque's elements arranged in a circle and save as an SVG file.

    :param dq: collections.deque object
    :param filename: Output SVG filename
    :param radius: Radius of the circle in pixels
    """
    # SVG canvas size (extra padding)
    canvas_size = (radius * 2 + 60, radius * 2 + 60)
    # center_x, center_y = canvas_size[0] / 2, canvas_size[1] / 2

    # Create SVG drawing
    d = draw.Drawing(canvas_size[0], canvas_size[1], origin='center')

    # Draw outer circle
    d.append(draw.Circle(0, 0, radius, fill='none', stroke='black', stroke_width=2))

    # Place deque elements evenly around the circle
    if not dq:
        return

    n = len(dq)
    angle_step = 2 * math.pi / n

    for i, value in enumerate(dq):
        angle = i * angle_step - math.pi / 2  # start from top
        x = radius * math.cos(angle)
        y = radius * math.sin(angle)

        # Draw node circle
        stroke = {'R': 'red', 'G': 'green', 'B': 'blue'}[value]
        d.append(draw.Circle(x, y, 15, fill='black', stroke=stroke, stroke_width=5))

        # Draw text (centered)
        d.append(draw.Text(str(value), 12, x, y, center=True, fill='white'))

    display(SVG(d.as_svg()))


@test(tests=test_cases[:])
def partII_naive(notes: str, repeat: int = 100, debug: bool = False) -> int:
    balloons = deque(repeat * notes.strip())
    fluffbolts = 0

    if debug:
        steps = 0
        deque_to_svg_circle(balloons)
        print(f'{balloons=}')

    fluffbolt_cycle = cycle('RGB')

    while balloons:
        fluffbolt = next(fluffbolt_cycle)
        fluffbolts += 1

        if debug:
            steps += 1  # type: ignore
            print(f'\n{steps=}, {fluffbolt=}')
            deque_to_svg_circle(balloons)
            print(f'{balloons=}')

        n = len(balloons)
        mid = n // 2
        balloon = balloons.popleft()

        if balloons and balloon == fluffbolt and n & 1 == 0:
            del balloons[mid - 1]

    if debug:
        steps += 1  # type: ignore
        print(f'\n{steps=}, {fluffbolt=}')  # type: ignore
        deque_to_svg_circle(balloons)
        print(f'{balloons=}')

    return fluffbolts
```

```python
test_cases = [
    # cSpell:disable
    {
        'name': 'Example I',
        'notes': 'GGBR',
        'repeat': 5,
        'expected': 14,
    },
    {
        'name': 'Example II repeat 10',
        'notes': 'BBRGGRRGBBRGGBRGBBRRBRRRBGGRRRBGBGG',
        'repeat': 10,
        'expected': 304,
    },
    {
        'name': 'Example II repeat 50',
        'notes': 'BBRGGRRGBBRGGBRGBBRRBRRRBGGRRRBGBGG',
        'repeat': 50,
        'expected': 1464,
    },
    {
        'name': 'Example II repeat 100',
        'notes': 'BBRGGRRGBBRGGBRGBBRRBRRRBGGRRRBGBGG',
        'repeat': 100,
        'expected': 2955,
    },
    {
        'name': 'Part 2',
        'notes': notes2,
        'repeat': 100,
        'expected': 21609,
    },
    # cSpell:enable
]


@test(tests=test_cases[:])
def partII(notes: str, repeat: int = 100) -> int:
    right_balloons = deque(repeat // 2 * notes)
    left_balloons = deque((repeat - repeat // 2) * notes)

    while len(left_balloons) > len(right_balloons):
        right_balloons.append(left_balloons.popleft())

    fluffbolts = 0

    fluffbolt_cycle = cycle('RGB')

    while right_balloons:
        fluffbolt = next(fluffbolt_cycle)
        fluffbolts += 1

        if right_balloons[0] == fluffbolt and len(right_balloons) == len(left_balloons):
            left_balloons.popleft()

        right_balloons.popleft()

        if len(left_balloons) > len(right_balloons):
            right_balloons.append(left_balloons.popleft())

    return fluffbolts
```
