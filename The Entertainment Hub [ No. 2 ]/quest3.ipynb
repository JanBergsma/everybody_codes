> ## <span style="color:#eed699"> <img src='./images/book.svg' height=40 width=40 style="display:inline-block height:fit-content;vertical-align: middle; margin: 5px; ">Story section</span>
>
> Some inventions are born out of necessity. Some inventions are the result of sheer brilliance. And a few… well, a few are just born, simply because I was bored one rainy afternoon and had access to a lathe. That’s how I ended up designing the world’s first truly rollable die.
>
> Not the usual cube nonsense, mind you. Those things don’t roll - they ricochet, they leap, and they spin like caffeinated goats. Whoever coined the phrase “roll a die” clearly never tried it. I mean, come on - you don’t roll a cube, you unleash it and pray it lands somewhere legal.
>
> My creation, on the other hand? Oh, it rolls. A finely cut polygonal cylinder - a kind of angular log with numbers on the sides. It moves with purpose, with grace… with physics! And when it stops, you can trust what it shows. No funny business. No airborne acrobatics. Just a fair roll.
>
> Now I only need to design the game around this magnificent invention. It has to be fun to play, hard to predict, and - of course - wildly addictive. Coins will follow. They always do.
>
> <br/>

My version of **the die can have any number of faces, with any values, even negative ones**. On one hand, rolling the die must return a random result. If that weren't the case, the Golden Duck Order's inspection would have noticed it very quickly. On the other hand, the ability to predict the next results is the only way to ensure the safety of my coins.

This is why I placed a mechanism inside each die that fully determines its subsequent results, even though they appear to be random. Let's take a closer look at the sample die configuration:

The first number indicates the **id**, and it is used to identify each die in the game. The second parameter is a **list of the face**s arranged in the order they are placed on the die. This is the list of possible results the die can return. The next element is the seed parameter, which is used by the algorithm to generate the subsequent results. Each die also has an inner **pulse** and **counter of rolls**, also used by the algorithm:

- First, initialise the inner pulse of the die with the seed parameter:
  `pulse = seed `.
- For each roll, calculate how many faces the die has to spin as:
  `spin = roll_number * pulse` . For the first roll, we start from the first face from the faces list. For the next roll, we start from the previously rolled face. Roll number starts at 1 and is incremented each time the die is rolled.
- Now you know the next result rolled with the die! The only thing left is to update the pulse.
- Add "spin" value to the pulse:
  `pulse = pulse + spin` .
- Divide the pulse by the seed parameter and set the reminder as the new pulse (modulo operation):
  `pulse = pulse % seed` .
- Increment the pulse by one and add the roll_number and the seed parameter to it:
  `pulse = pulse + 1 + roll_number + seed `.
- Increment roll_number and go back to the second step for the next roll.

```python
import re
from dataclasses import dataclass
from typing import Self

from test_utilities import test
from util import Str

test_cases = [
    # cSpell:disable
    {
        'name': 'Example I',
        'notes': '1: faces=[1,2,4,-1,5,7,9] seed=3',
        'expected': 2363,
    },
    {
        'name': 'Example II',
        'notes': """
            1: faces=[1,2,3,4,5,6] seed=7
            2: faces=[-1,1,-1,1,-1] seed=13
            3: faces=[9,8,7,8,9] seed=17
        """,
        'expected': 844,
    },
    # cSpell:enable
]


@dataclass
class Dice:
    id: int
    faces: list[int]
    seed: int
    pulse: int = 0
    roll_number: int = 0
    spin: int = 0
    face: int = 0

    def roll(self) -> None:
        self.roll_number += 1
        self.spin = self.roll_number * self.pulse

        self.pulse += self.spin
        self.pulse %= self.seed
        self.pulse += 1 + self.roll_number + self.seed

        self.face += self.spin
        self.face %= len(self.faces)

    def sample(self, first_nr: int) -> None:
        self.initialize_seed()

        print(' Roll number      Spin    Result    Pulse after roll')
        print(f'           -         -         -  {self.pulse:>18}')

        for _ in range(first_nr):
            self.roll()
            if self.roll_number < 10 or self.roll_number % 10 == 0:
                if self.roll_number == 20:
                    print()
                    print('~' * 100)
                    print()
                result = self.faces[self.face]
                print(f'{self.roll_number:>12} {self.spin:>9} {result:>10} {self.pulse:>18}')

    def initialize_seed(self):
        self.pulse = self.seed

    @classmethod
    def parse_line(cls, line: str) -> Self:
        id, *faces, seed = map(int, re.findall(r'-?\d+', line.strip()))
        return cls(id, faces, seed)


class Die(Str):
    DICE = Dice

    def __init__(self, notes: str) -> None:
        self.die = [self.DICE.parse_line(line) for line in notes.strip().splitlines()]
        self.initialize_die()

    def goal_reached_rounds(self, goal: int, debug: bool = False) -> int:
        if debug:
            self.print_header()

        total_points = 0
        roll_number = 1

        while total_points < goal:
            roll_points = 0

            for dice in self.die:
                dice.roll()
                result = dice.faces[dice.face]
                roll_points += result
                total_points += result

            if debug and (roll_number < 10 or roll_number % 10 == 0):
                self.print_roll(total_points, roll_number, roll_points)

            roll_number += 1

        return roll_number - 1

    def initialize_die(self) -> None:
        for dice in self.die:
            dice.initialize_seed()

    def print_roll(self, total_points, roll_number, roll_points):
        if roll_number == 20:
            print()
            print('~' * 100)
            print()

        print(
            ''.join(
                (
                    f'{roll_number:>11}',
                    ''.join(f'{d.faces[d.face]:>10}' for d in self.die),
                    f'{roll_points:>16}',
                    f'{total_points:>18}',
                )
            )
        )

    def print_header(self):
        space = 5 * ' '
        print(
            ''.join(
                (
                    'Roll Number',
                    space,
                    space.join(f'Die {i}' for i in range(1, len(self.die) + 1)),
                    space,
                    'Roll points',
                    space,
                    'Total points',
                )
            )
        )

    def sample(self, idx: int, rounds: int = 100) -> None:
        self.die[idx].sample(rounds)


@test(tests=test_cases[:])
def partI(notes: str, goal: int = 10000, debug: bool = False) -> int:
    die = Die(notes)
    # print(die)
    # die.sample(0)
    return die.goal_reached_rounds(goal, debug)
```

```python
from more_itertools import first, last
from test_utilities import test

test_cases = [
    # cSpell:disable
    {
        'name': 'Example I',
        'notes': """
            1: faces=[1,2,3,4,5,6,7,8,9] seed=13
            2: faces=[1,2,3,4,5,6,7,8,9] seed=29
            3: faces=[1,2,3,4,5,6,7,8,9] seed=37
            4: faces=[1,2,3,4,5,6,7,8,9] seed=43

            51257284
        """,
        'expected': '1,3,4,2',
    },
    # cSpell:enable
]


class DieII(Die):
    def __init__(self, notes: str) -> None:
        super().__init__(first(re.split(r'(?:\r?\n){2,}', notes)))
        self.track = [int(d) for d in last(re.split(r'(?:\r?\n){2,}', notes)).strip()]

    def race(self, debug: bool) -> str:
        if debug:
            self.print_race_header()

        placements_old = [0] * len(self.die)
        placements_new = [0] * len(self.die)

        finish = len(self.track) - 1
        finished_order = []

        count = 1

        while len(finished_order) < len(self.die):
            for i, dice in enumerate(self.die):
                if placements_old[i] > finish:
                    placements_new[i] = placements_old[i]
                else:
                    dice.roll()

                    if dice.faces[dice.face] == self.track[placements_old[i]]:
                        placements_new[i] = placements_old[i] + 1
                        if placements_old[i] == finish:
                            finished_order.append(dice.id)
                    else:
                        placements_new[i] = placements_old[i]

            if debug:
                self.print_row(placements_old)

            placements_old, placements_new = placements_new, placements_old
            count += 1

        return ','.join(map(str, finished_order))

    def print_row(self, placements: list[int]) -> None:
        if not any(
            placements[i] < len(self.track) and self.track[placements[i]] == dice.faces[dice.face]
            for i, dice in enumerate(self.die)
        ):
            return

        line = [f'{max(dice.roll_number for dice in self.die):>3}']
        for i, dice in enumerate(self.die):
            if (
                0 <= placements[i] < len(self.track)
                and self.track[placements[i]] == dice.faces[dice.face]
            ):
                line.append(f'[{dice.faces[dice.face]}]')
            elif placements[i] < len(self.track):
                line.append(f' {dice.faces[dice.face]} ')
            else:
                line.append(' _ ')

        print('       '.join(line))

    def print_race_header(self) -> None:
        space = 5 * ' '
        print(
            ''.join(
                (
                    'Turn',
                    space,
                    space.join(f'Die {i}' for i in range(1, len(self.die) + 1)),
                )
            )
        )


@test(tests=test_cases[:])
def partII(notes: str, debug: bool = False) -> str:
    die = DieII(notes)
    return die.race(debug)
```
