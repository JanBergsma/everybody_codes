## <span style="color:#eed699">Part I</span>

There's something remarkable about the eggshell you hatched from. Faintly glowing dots and lines cover its entire surface. Even stranger is the fact that you understand what they mean! Is it a unique language that all members of your species are born knowing?

After taking a closer look, it turns out that the eggshell has three separate sections. <span style="color:ivory">Each section contains a list of names and a list of instructions specifying how to navigate through the list of names.</span> Before executing the first instruction, you are at the first name of the list. Each instruction begins with the letter <code> R </code> or <code> L </code>, followed by a certain number, which simply indicates how many positions to move to the (R)ight or to the (L)eft from your current position.

You're taking a closer look at the first section (<span style="color:#eed699">your notes</span> ). The description says that <span style="color:ivory">if you can't follow the full instruction without going off the list, you do as many moves as you can to stay on it.</span> For example, if the next instruction is <code> L3 </code> and you are at the second name of the list, you move only one position to the left. The description additionally states that the name you land on at the end of the entire process represents the name of the dragonduck inside the egg! So, you're a dragonduck, and you'll find out your name soon!

<span style="color:ivory">Example based on the following notes:</span>

<pre style=" border: 2px dotted lime;border-radius: 5px;padding:2px;width: max-content;">
Vyrdax,Drakzyph,Fyrryn,Elarzris

R3,L2,R3,L1
</pre>

Start at the first name on the list: <span style="color:ivory;font-weight:bolder">Vyrdax</span>:

<pre style=" border: 1px solid ivory;border-radius: 5px;padding:2px;width: max-content;">
  ↓
Vyrdax  Drakzyph  Fyrryn  Elarzris
</pre>

The first instruction is <code> R3 </code> , so you move three positions to the right, landing on <span style="color:ivory;font-weight:bolder">Elarzris</span>:

<pre style=" border: 1px solid white;border-radius: 5px;padding:2px;width: max-content;">
                            ↓
Vyrdax  Drakzyph  Fyrryn  Elarzris
</pre>

The next instruction is <code> L2 </code> , so you move three positions to the right, landing on <span style="color:ivory;font-weight:bolder">Drakzyph</span>:

<pre style=" border: 1px solid white;border-radius: 5px;padding:2px;width: max-content;">
           ↓
Vyrdax  Drakzyph  Fyrryn  Elarzris
</pre>

The next instruction is <code> R3 </code> , so you move three positions to the right, landing on <span style="color:ivory;font-weight:bolder">Elarzris</span>:

<pre style=" border: 1px solid white;border-radius: 5px;padding:2px;width: max-content;">
                             ↓
Vyrdax  Drakzyph  Fyrryn  Elarzris
</pre>

The final instruction is <code> L1 </code> , so you move three positions to the right, landing on <span style="color:ivory;font-weight:bolder">Fyrryn</span>. This is where the process ends:

<pre style=" border: 1px solid white;border-radius: 5px;padding:2px;width: max-content;">
                   ↓
Vyrdax  Drakzyph  Fyrryn  Elarzris
</pre>

<span style="color:#eed699">What is your name?</span>


```python
import re

from test_utilities import test

tests = [
    {
        'name': 'Example',
        's': """
            Vyrdax,Drakzyph,Fyrryn,Elarzris

            R3,L2,R3,L1
        """,
        'expected': 'Fyrryn',
    },
]


@test(tests=tests)
def partI(s: str) -> str:
    names, instructions = [part.split(',') for part in re.split(r'(?:\r?\n\s*){2,}', s.strip())]
    pointer = 0
    for instruction in instructions:
        direction = 1 if instruction[0] == 'R' else -1
        delta = int(instruction[-1])
        pointer += direction * delta
        if pointer >= len(names):
            pointer = len(names) - 1
        elif pointer < 0:
            pointer = 0

    return names[pointer]
```

```python
with open('../inputs/everybody_codes_e2025_q01_p1.txt') as f:
    notes1 = f.read()

print(f'Part I: {partI(notes1)}')
```

tart at the first name on the list: <span style="color:ivory;font-weight:bolder">Vyrdax.</span> The first instruction is <code> R3 </code> , so you move three positions to the right (clockwise), landing on <span style="color:ivory;font-weight:bolder">Elarzris</span>.

The next instruction is <code> L2 </code> , so you move two positions to the left (counterclockwise), stopping at <span style="color:ivory;font-weight:bolder">Drakzyph</span>.

The following instruction is <code> R3 </code> , moving you three places to the right, bringing you back to <span style="color:ivory;font-weight:bolder">Vyrdax</span>.

The final instruction is <code> L1 </code> , shifting you one position to the left, ending the process at <span style="color:ivory;font-weight:bolder">Elarzris</span> . This would be the name of your first parent for this example.

<span style="color:#eed699">What is the name of your first parent?</span>


```python
import re

from test_utilities import test

tests = [
    {
        'name': 'Example',
        's': """
            Vyrdax,Drakzyph,Fyrryn,Elarzris

            R3,L2,R3,L1
        """,
        'expected': 'Elarzris',
    },
]


@test(tests=tests)
def partII(s: str) -> str:
    names, instructions = [part.split(',') for part in re.split(r'(?:\r?\n\s*){2,}', s.strip())]
    pointer = 0
    for instruction in instructions:
        direction = 1 if instruction[0] == 'R' else -1
        delta = int(instruction[-1])
        pointer = (pointer + direction * delta) % len(names)
    return names[pointer]
```

## <span style="color:#eed699">Part III</span>

The final section also requires arranging the list of names in a closed loop. However, this time, the instructions work in a different way. <span style="color:ivory;font-weight:bolder">The indicator <code> R </code> or <code> L </code> points to the name relative to the one at the top of the circle (the first on the list) that must swap places with it.</span>

<span style="color:#eed699">What is the name of your second parent?</span>


```python
import re

from test_utilities import test

tests = [
    {
        'name': 'Example',
        's': """
            Vyrdax,Drakzyph,Fyrryn,Elarzris

            R3,L2,R3,L3
        """,
        'expected': 'Drakzyph',
    },
]


@test(tests=tests)
def partIII(s: str) -> str:
    names, instructions = [part.split(',') for part in re.split(r'(?:\r?\n\s*){2,}', s.strip())]

    for instruction in instructions:
        direction = 1 if instruction[0] == 'R' else -1
        delta = int(instruction[1:])

        swap_idx = (direction * delta) % len(names)
        names[0], names[swap_idx] = names[swap_idx], names[0]

    return names[0]
```

```python
with open('../inputs/everybody_codes_e2025_q01_p3.txt') as f:
    notes3 = f.read()


print(f'Part III: {partIII(notes3)}')
```
