<span style="color:#eed699">How many barrels can the entire chain reaction destroy?</span>


```python
import re
from collections import deque
from collections.abc import Iterator

from test_utilities import test

tests = [
    {
        'name': 'Example 1',
        'notes': """
            989601
            857782
            746543
            766789
        """,
        'expected': 16,
    },
]

type Cell = tuple[int, int]


class Grid:
    MOVES = (-1, 0), (0, 1), (1, 0), (0, -1)

    def __init__(self, notes: str) -> None:
        self.grid = [[int(barrel) for barrel in barrels] for barrels in re.findall(r'\S+', notes)]

    def ignite(self, start_barrels: list[Cell]) -> int:
        queue = deque(start_barrels)

        seen = set()

        while queue:
            row, col = queue.popleft()

            if (row, col) in seen:
                continue

            seen.add((row, col))

            queue.extend((r, c) for r, c in self.valid_moves(row, col))

        return len(seen)

    def valid_moves(self, row, col) -> Iterator[Cell]:
        rows, cols = len(self.grid), len(self.grid[0])

        for dr, dc in self.MOVES:
            if not (0 <= row + dr < rows and 0 <= col + dc < cols):
                continue

            if self.grid[row][col] < self.grid[row + dr][col + dc]:
                continue

            yield row + dr, col + dc

    def __str__(self) -> str:
        return '\n'.join(' '.join(f'{b:=2d}' for b in row) for row in self.grid)


@test(tests=tests)
def partI(notes: str) -> int:
    grid = Grid(notes)
    return grid.ignite([(0, 0)])
```

<span style="color:#eed699">How many barrels at most can be ultimately destroyed by setting the three barrels chosen with the greedy approach on fire simultaneously?</span>


```python
from itertools import product

from matplotlib import pyplot as plt
from test_utilities import test

tests = [
    {
        'name': 'Example 1',
        'notes': """
            5411
            3362
            5235
            3112
        """,
        'expected': 14,
    },
    {
        'name': 'Example 2',
        'notes': """
            41951111131882511179
            32112222211518122215
            31223333322115122219
            31234444432147511128
            91223333322176121892
            61112222211166431583
            14661111166111111746
            11111119142122222177
            41222118881233333219
            71222127839122222196
            56111126279711111517
        """,
        'expected': 136,
    },
]


class GridIII(Grid):
    def ignite_once(self, start: Cell, seen_before: set[Cell]) -> int:
        if start in seen_before:
            return 0

        queue = deque([start])
        seen = set()

        while queue:
            row, col = queue.popleft()

            if (row, col) in seen:
                continue

            seen.add((row, col))

            queue.extend((r, c) for r, c in self.valid_moves(row, col))

        seen_before |= seen
        return len(seen)

    def ignite_destroy(self, start: Cell) -> None:
        queue = deque([start])
        seen = set()

        while queue:
            row, col = queue.popleft()

            if (row, col) in seen:
                continue

            seen.add((row, col))

            queue.extend((r, c) for r, c in self.valid_moves(row, col))

        for r, c in seen:
            self.grid[r][c] = -1

    def valid_moves(self, row, col) -> Iterator[Cell]:
        rows, cols = len(self.grid), len(self.grid[0])

        for dr, dc in self.MOVES:
            if not (0 <= row + dr < rows and 0 <= col + dc < cols):
                continue

            if self.grid[row + dr][col + dc] < 0:
                continue

            if self.grid[row][col] < self.grid[row + dr][col + dc]:
                continue

            yield row + dr, col + dc

    def display_grid(self) -> None:
        plt.figure(figsize=(10, 10))
        plt.matshow(
            [[10 * (n - 1) for n in row] for row in self.grid], cmap='tab20b', fignum=1
        )  # 'gray' colormap: 0=black, 1=white

        plt.axis('off')  # Hide axis ticks
        plt.show()


@test(tests=tests)
def partIII(notes: str) -> int:
    grid = GridIII(notes)
    rows, cols = len(grid.grid), len(grid.grid[0])

    seen_before = set()
    first_destructor = max(
        sorted(filter(lambda c: grid.grid[c[0]][c[1]] > 3, product(range(rows), range(cols)))),
        key=lambda cell: grid.ignite_once(cell, seen_before),
    )
    grid.ignite_destroy(first_destructor)

    seen_before = set()
    second_destructor = max(
        sorted(filter(lambda c: grid.grid[c[0]][c[1]] > -1, product(range(rows), range(cols)))),
        key=lambda cell: grid.ignite_once(cell, seen_before),
    )
    grid.ignite_destroy(second_destructor)

    seen_before = set()
    third_destructor = max(
        sorted(filter(lambda c: grid.grid[c[0]][c[1]] > -1, product(range(rows), range(cols)))),
        key=lambda cell: grid.ignite_once(cell, seen_before),
    )
    grid.ignite_destroy(third_destructor)

    return sum(1 for r, c in product(range(rows), range(cols)) if grid.grid[r][c] == -1)
```
