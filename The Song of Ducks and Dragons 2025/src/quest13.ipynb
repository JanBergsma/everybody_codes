## <span style="color:#eed699">Part I</span>

> ## <span style="color:#eed699"> <img src='./images/book.svg' height=40 width=40 style="display:inline-block height:fit-content;vertical-align: middle; margin: 5px; ">Story section</span>
>
> Explosions shake the air, and pillars of fire light up the sky. The barrels that haven't exploded topple over, their contents flowing down the slope like a river of lava, heading straight for the castle! Shouts echo through the courtyard. People spring into action, trying to stop the impending disaster. Soldiers bring sandbags; others dig trenches to redirect the substance - everyone is busy.
>
> The pulsing signal still flickers in the distance, guiding you to the entrance of a cave on the mountainside. You glide towards it. In front of the massive iron gates stands a lone guard. He seems more concerned about reading documents than actually doing any guarding.
>
> You land a few steps away and, without hesitation, spit a fireball that lands at his feet. The guard jumps, his face turning pale. He doesn’t even try to fight. Instead, he throws down the papers in his hands and runs away as fast as his legs will carry him.
>
> You pick up the papers and begin to look through them. It appears that these documents contain instructions for crafting the lock to the gate in front of you! If your suspicions are correct, the dragonducks will soon be free.
>
> <br/>

You approach the massive door and see three enormous wheels mounted at its centre. Each one has a dial in the middle and a series of numbers written around the edge. Above each dial is a small arrow pointing to a number on the wheel. It looks like you need to set all three dials correctly to open the gate.

The instructions for crafting the first wheel include a list of numbers <span style="color:#eed699">( your notes )</span>. Initially, the wheel has only one number written at the highest point: <code> 1 </code>, and as many free spots as there are numbers on the list. The numbers from the list should be placed on the wheel in a clockwise and counterclockwise pattern from this point. The first number is placed at the first free spot to the right (clockwise) of <code> 1 </code>. The second number is placed at the first free spot to the left (counterclockwise) of <code> 1 </code>. The third number is placed to the right again, and so on.

The instructions state that the first lock should open after turning the dial <code style="color:ivory;border: 1px dotted #eed699;border-radius: 5px;padding:2px;width: max-content;"> 2025 </code> positions clockwise, starting from the initial number <code style="color:white"> 1 </code>.

61
67
</pre>

The steps for constructing the wheel based on the given list are shown below. The first number from the list is placed on the first available position clockwise from <code> 1 </code>. The second number from the list is placed counterclockwise from <code> 1 </code>. The third number is again placed clockwise, and so on, alternating directions.

If we turn the dial by a very large number of positions, we will pass through exactly the same spots multiple times along the way. For example, turning the dial from <code> 1 </code> by 6 or 12 positions will cause the pointer to return to the number <code> 1 </code>.

If we turn the dial <code style="border: 1px dotted #eed699;border-radius: 5px;padding:2px;width: max-content;"> 2025 </code> positions clockwise, the pointer will land on the number <code style="color:white;font-weight:bolder"> 67 </code>.

<span style="color:#eed699">What number should the pointer of the first dial be set to?</span>


```python
import re

from more_itertools import grouper
from test_utilities import test

tests = [
    {
        'name': 'Example 1',
        'notes': """
            72
            58
            47
            61
            67
        """,
        'turn': 2025,
        'expected': 67,
    },
]


@test(tests=tests)
def partI(notes: str, turn: int = 2025) -> int:
    numbers = [int(num) for num in re.findall(r'\d+', notes)]
    n = len(numbers) + 1

    wheel = [-1] * n
    wheel[0] = 1

    for i, (right, left) in enumerate(grouper(numbers, 2, incomplete='fill', fillvalue=-1)):
        wheel[i + 1] = right

        if left > -1:
            wheel[(-i - 1) % n] = left

    return wheel[turn % n]
```

```python
from collections import deque

from more_itertools import minmax
from test_utilities import test

tests = [
    {
        'name': 'Example 1',
        'notes': """
            10-15
            12-13
            20-21
            19-23
            30-37
        """,
        'turn': 20252025,
        'expected': 30,
    },
    {
        'name': 'Part II',
        'notes': notes2,
        'turn': 20252025,
        'expected': 7281,
    },
    {
        'name': 'Part III',
        'notes': notes3,
        'turn': 202520252025,
        'expected': 489628,
    },
]


@test(tests=tests[:2])
def partII_bf(notes: str, turn: int = 20252025) -> int:
    # slow 40s on Part III
    numbers = (int(num) for num in re.findall(r'\d+', notes))

    ranges = grouper(numbers, 2, incomplete='fill', fillvalue=-1)
    pairwise_ranges = grouper(ranges, 2, incomplete='fill', fillvalue=(-1, -1))

    wheel = deque([1])

    for (right_fr, right_to), (left_fr, left_to) in pairwise_ranges:
        wheel.extend(range(right_fr, right_to + 1))
        if left_fr != -1:
            wheel.extendleft(range(left_fr, left_to + 1))

    wheel.rotate(-wheel.index(1))
    n = len(wheel)

    return wheel[turn % n]


@test(tests=tests)
def partII(notes: str, turn: int = 20252025) -> int:
    numbers = (int(num) for num in re.findall(r'\d+', notes))

    ranges = grouper(numbers, 2, incomplete='fill', fillvalue=-1)
    pairwise_ranges = grouper(ranges, 2, incomplete='fill', fillvalue=(-1, -1))

    wheel = deque([(1, 1)])
    n = 1

    for (right_fr, right_to), (left_fr, left_to) in pairwise_ranges:
        wheel.append((right_fr, right_to))
        n += right_to - right_fr + 1
        if left_fr != -1:
            wheel.appendleft((left_to, left_fr))
            n += left_to - left_fr + 1

    wheel.rotate(-wheel.index((1, 1)))

    turn %= n
    i = 0
    for fr, to in wheel:
        fr1, to1 = minmax(fr, to)
        j = i
        i += to1 - fr1 + 1
        if turn < i:
            if fr <= to:
                return fr + (turn - j)
            else:
                return fr - (turn - j)

    raise ValueError(f'{turn=}, huh!')
```
