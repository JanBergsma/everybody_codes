<span style="color:#eed699">What is the total number of active tiles across all <code style="color:#eed699;font-weight:bolder"> 10 </code> rounds?</span>


```python
import re
from itertools import product

from test_utilities import test

tests = [
    {
        'name': 'Example 1',
        'notes': """
            .#.##.
            ##..#.
            ..##.#
            .#.##.
            .###..
            ###.##
        """,
        'rounds': 10,
        'expected': 200,
    },
]


class Tiles:
    DIAGONALS = (-1, 1), (1, 1), (1, -1), (-1, -1)

    def __init__(self, notes: str) -> None:
        self.grid = [list(row) for row in re.findall(r'\S+', notes)]

    def perform_rounds(self, rounds: int, debug: bool = False) -> int:
        rows, cols = len(self.grid), len(self.grid[0])

        counts = 0
        other_grid = [[' '] * cols for _ in range(rows)]

        if debug:
            print('Initial')
            print(self)
            print()

        for round in range(rounds):
            round_count = self.perform_round(other_grid)
            self.grid, other_grid = other_grid, self.grid
            counts += round_count

            if debug:
                print(f'Round {round + 1}')
                print(self)
                print(f'active {round_count}')
                print()

        return counts

    def perform_round(self, new_grid: list[list[str]]) -> int:
        rows, cols = len(self.grid), len(self.grid[0])
        count_active = 0

        for r, c in product(range(rows), range(cols)):
            active_diagonal_neighbors = self.count_active_diagonal_neighbors(r, c)

            if self.grid[r][c] == '#':
                if active_diagonal_neighbors & 1 == 1:
                    new_grid[r][c] = '#'
                    count_active += 1
                else:
                    new_grid[r][c] = '.'
            else:
                if active_diagonal_neighbors & 1 == 0:
                    new_grid[r][c] = '#'
                    count_active += 1
                else:
                    new_grid[r][c] = '.'

        return count_active

    def count_active_diagonal_neighbors(self, row: int, col: int) -> int:
        rows, cols = len(self.grid), len(self.grid[0])
        return sum(
            1
            for dr, dc in self.DIAGONALS
            if 0 <= row + dr < rows and 0 <= col + dc < cols
            if self.grid[row + dr][col + dc] == '#'
        )

    def __str__(self) -> str:
        return '\n'.join(''.join(row) for row in self.grid)


@test(tests=tests)
def partI(notes: str, rounds: int = 10, debug: bool = False) -> int:
    tiles = Tiles(notes)
    return tiles.perform_rounds(rounds, debug=debug)
```

<span style="color:#eed699">How many active tiles are there when you simulate <code style="color:#eed699;font-weight:bolder"> 1000000000 </code> rounds and count only those that match your pattern?</span>


```python
import pickle
from pprint import pprint

from matplotlib import pyplot as plt
from test_utilities import test

tests = [
    {
        'name': 'Example 1',
        'notes': """
            #......#
            ..#..#..
            .##..##.
            ...##...
            ...##...
            .##..##.
            ..#..#..
            #......#
        """,
        'rounds': 1_000_000_000,
        'expected': 278388552,
    },
    {
        'name': 'Part III',
        'notes': """
                #.#..#.#
                .#.##.#.
                #.#..#.#
                .#.##.#.
                .#.##.#.
                #.#..#.#
                .#.##.#.
                #.#..#.#
        """,
        'rounds': 1_000_000_000,
        'expected': 971917140,
    },
]


class TilesII(Tiles):
    def __init__(self, notes: str, width: int = 34) -> None:
        self.pattern = [list(row) for row in re.findall(r'\S+', notes)]
        self.grid = [[' '] * width for _ in range(width)]

    def perform_rounds_until_repeat(self, debug: bool = False) -> tuple[int, int, int]:
        rows, cols = len(self.grid), len(self.grid[0])

        counts = 0
        other_grid = [[' '] * cols for _ in range(rows)]

        if debug:
            print('Initial')
            print(self)
            print()

        seen = {pickle.dumps(self.grid, -1): 0}

        round = 0

        while True:
            round_count = self.perform_round(other_grid)
            self.grid, other_grid = other_grid, self.grid

            round += 1

            p = pickle.dumps(self.grid, -1)
            if p in seen:
                return seen[p], round, counts

            seen[p] = round

            if self.center_equals_pattern():
                counts += round_count

            if debug:
                print(f'Round {round + 1}')
                print(self)
                print(f'active {round_count}')
                print()

    def count_pattern_occurrences(self, rounds: int, debug: bool = False) -> int:
        rows, cols = len(self.grid), len(self.grid[0])

        counts = 0
        other_grid = [[' '] * cols for _ in range(rows)]

        if debug:
            print('Initial')
            print(self)
            print()

        for round in range(rounds):
            round_count = self.perform_round(other_grid)
            self.grid, other_grid = other_grid, self.grid

            round += 1
            if self.center_equals_pattern():
                counts += round_count

                if debug:
                    print(f'Round {round + 1}')
                    print(self)
                    print(f'active {round_count}')
                    print()

        return counts

    def center_equals_pattern(self) -> bool:
        offset = len(self.grid) // 2 - len(self.pattern) // 2
        return all(
            self.grid[offset + r][offset + c] == self.pattern[r][c]
            for r, c in product(range(len(self.pattern)), repeat=2)
        )

    def display_pattern(self) -> None:
        n, m = len(self.grid), len(self.pattern)
        self._display(self.pattern, (4 * m / n, 4 * m / n))

    def display_grid(self) -> None:
        self._display(self.grid)

    @classmethod
    def _display(cls, grid: list[list[str]], size: tuple[float, float] = (4, 4)) -> None:
        plt.figure(figsize=size)
        plt.matshow(
            [[0 if cell == '#' else 100 for cell in row] for row in grid],
            cmap='Greys',
            fignum=1,
        )

        plt.axis('off')  # Hide axis ticks
        plt.show()


@test(tests=tests)
def partIII(notes: str, rounds: int, debug: bool = False) -> int:
    tiles = TilesII(notes)
    start, end, count = tiles.perform_rounds_until_repeat()
    q, m = divmod(rounds, end - start)
    return count * q + tiles.count_pattern_occurrences(m)
```
