`(Xv - Xc) * (Xv - Xc) + (Yv - Yc) * (Yv - Yc) <= R * R`

```python
from itertools import product
from re import findall

from more_itertools import first
from test_utilities import test

tests = [
    {
        'name': 'Example 1',
        'notes': """
            189482189843433862719
            279415473483436249988
            432746714658787816631
            428219317375373724944
            938163982835287292238
            627369424372196193484
            539825864246487765271
            517475755641128575965
            685934212385479112825
            815992793826881115341
            1737798467@7983146242
            867597735651751839244
            868364647534879928345
            519348954366296559425
            134425275832833829382
            764324337429656245499
            654662236199275446914
            317179356373398118618
            542673939694417586329
            987342622289291613318
            971977649141188759131
        """,
        'expected': 1573,
    },
]


def log(on_circle, grid, rows, yv, xv):
    print(f'{xv, yv =}')
    print('\n'.join(' '.join(f'{c}' for c in row) for row in grid))
    print()
    print(
        '\n'.join(
            ' '.join(f'{grid[yc][xc]}' if on_circle(yc, xc) else '.' for xc in range(rows))
            for yc in range(rows)
        )
    )


@test(tests=tests)
def partI(notes: str, radius: int = 10, debug: bool = False) -> int:
    def on_circle(yc, xc):
        return (xv - xc) * (xv - xc) + (yv - yc) * (yv - yc) <= r_squared

    grid = [
        [c if c == '@' else int(c) for c in findall(r'[\d{1}@]', row)]
        for row in notes.strip().splitlines()
    ]

    rows, cols = len(grid), len(grid[0])
    yv, xv = first((r, c) for r in range(rows) for c in range(cols) if grid[r][c] == '@')

    r_squared = radius * radius

    if debug:
        log(on_circle, grid, rows, yv, xv)

    total = 0
    for yc, xc in product(range(rows), range(cols)):
        value = grid[yc][xc]
        if on_circle(yc, xc) and isinstance(value, int):
            total += value

    return total
```

<span style="color:#eed699">Find the step in which the destruction is the greatest. What is the result of multiplying that sum of destruction by the lava radius at that step?</span>


```python
from test_utilities import test
from util import hexcolor_str

tests = [
    {
        'name': 'Example 1',
        'notes': """
            4547488458944
            9786999467759
            6969499575989
            7775645848998
            6659696497857
            5569777444746
            968586@767979
            6476956899989
            5659745697598
            6874989897744
            6479994574886
            6694118785585
            9568991647449
        """,
        'expected': 1090,
    },
]


def log(grid, fringe, round):  # noqa: F811
    rows, cols = len(grid), len(grid[0])
    print(hexcolor_str('#ffffff', f'R:{round}  Lava:{sum(grid[y][x] for y, x in fringe)}'))
    print(
        '\n'.join(
            ''.join(
                hexcolor_str('#ffffff' if (y, x) in fringe else '#696969', f'{grid[y][x]}')
                for x in range(cols)
            )
            for y in range(rows)
        )
    )
    print()


type Cell = tuple[int, int]
type Grid = list[list[int | str]]


def get_circle(xv: int, yv: int, r: int, grid: Grid) -> list[Cell]:
    def on_circle(yc, xc):
        return (xv - xc) * (xv - xc) + (yv - yc) * (yv - yc) <= r_squared

    rows, cols = len(grid), len(grid[0])

    r_squared = r * r

    circle = []
    for yc, xc in product(range(rows), range(cols)):
        value = grid[yc][xc]
        if on_circle(yc, xc) and isinstance(value, int):
            circle.append((yc, xc))
    return circle


@test(tests=tests)
def partII(notes: str, debug: bool = False) -> int:

    grid = [
        [c if c == '@' else int(c) for c in findall(r'[\d{1}@]', row)]
        for row in notes.strip().splitlines()
    ]

    rows, cols = len(grid), len(grid[0])
    yv, xv = first((r, c) for r in range(rows) for c in range(cols) if grid[r][c] == '@')

    round = 1
    circle = get_circle(yv, xv, round, grid)

    greatest_destruction, greatest_destruction_round = 0, 0

    while circle:
        if debug:
            log(grid, circle, round)

        destruction = sum(grid[y][x] for y, x in circle)  # type: ignore
        if destruction > greatest_destruction:
            greatest_destruction = destruction
            greatest_destruction_round = round

        for y, x in circle:
            grid[y][x] = '.'

        round += 1
        circle = get_circle(yv, xv, round, grid)

    return greatest_destruction_round * greatest_destruction
```

```python
from heapq import heapify, heappop, heappush
from itertools import takewhile

from matplotlib import pyplot as plt
from more_itertools import one
from test_utilities import test

tests = [
    {
        'name': 'Example 1',
        'notes': """
            2645233S5466644
            634566343252465
            353336645243246
            233343552544555
            225243326235365
            536334634462246
            666344656233244
            6426432@2366453
            364346442652235
            253652463426433
            426666225623563
            555462553462364
            346225464436334
            643362324542432
            463332353552464
        """,
        'expected': 592,
    },
    {
        'name': 'Example 2',
        'notes': """
            545233443422255434324
            5222533434S2322342222
            523444354223232542432
            553522225435232255242
            232343243532432452524
            245245322252324442542
            252533232225244224355
            523533554454232553332
            522332223232242523223
            524523432425432244432
            3532242243@4323422334
            542524223994422443222
            252343244322522222332
            253355425454255523242
            344324325233443552555
            423523225325255345522
            244333345244325322335
            242244352245522323422
            443332352222535334325
            323532222353523253542
            553545434425235223552
        """,
        'expected': 330,
    },
    {
        'name': 'Example 3',
        'notes': """
            5441525241225111112253553251553
            133522122534119S911411222155114
            3445445533355599933443455544333
            3345333555434334535435433335533
            5353333345335554434535533555354
            3533533435355443543433453355553
            3553353435335554334453355435433
            5435355533533355533535335345335
            4353545353545354555534334453353
            4454543553533544443353355553453
            5334554534533355333355543533454
            4433333345445354553533554555533
            5554454343455334355445533453453
            4435554534445553335434455334353
            3533435453433535345355533545555
            534433533533535@353533355553345
            4453545555435334544453344455554
            4353333535535354535353353535355
            4345444453554554535355345343354
            3534544535533355333333445433555
            3535333335335334333534553543535
            5433355333553344355555344553435
            5355535355535334555435534555344
            3355433335553553535334544544333
            3554333535553335343555345553535
            3554433545353554334554345343343
            5533353435533535333355343333555
            5355555353355553535354333535355
            4344534353535455333455353335333
            5444333535533453535335454535553
            3534343355355355553543545553345
        """,
        'expected': 3180,
    },
]


INF = 10**18


def get_circle_II(xv: int, yv: int, r: int) -> set[Cell]:
    def on_circle(yc, xc):
        return (xv - xc) * (xv - xc) + (yv - yc) * (yv - yc) <= r_squared

    r_squared = r * r

    return {
        (yc, xc)
        for yc, xc in product(range(yv - r, yv + r + 1), range(xv - r, xv + r + 1))
        if on_circle(yc, xc)
    }


def dijkstra(grid, max_time, circle, barrier, heap, goals):
    rows, cols = len(grid), len(grid[0])

    seen = {(y, x) for _, y, x in heap}

    heapify(heap)

    results = []
    while heap:
        t, y, x = heappop(heap)

        if (y, x) in goals:
            results.append((t, y, x))

        for dy, dx in ((-1, 0), (0, 1), (1, 0), (0, -1)):
            y1, x1 = y + dy, x + dx

            if not 0 <= y1 < rows:
                continue

            if not 0 <= x1 < cols:
                continue

            if (y1, x1) in barrier:
                continue

            if (y1, x1) in circle:
                continue

            if (y1, x1) in seen:
                continue

            seen.add((y1, x1))

            t1 = (t + grid[y1][x1]) if isinstance(grid[y1][x1], int) else t

            if t1 >= max_time:
                continue

            heappush(heap, (t1, y1, x1))

    return results


def display_grid(grid: Grid, size: tuple[float, float] = (4, 4)) -> None:
    plt.figure(figsize=size)
    plt.matshow(
        [[cell if isinstance(cell, int) and cell < INF else 0 for cell in row] for row in grid],
        cmap='bwr',
        fignum=1,
    )

    plt.axis('off')  # Hide axis ticks
    plt.show()


@test(tests=tests[:])
def partIII(notes: str, debug: bool = False) -> int:
    def on_circle(y):
        return (y, xv) not in circle

    grid = [
        [c if c in '@S' else int(c) for c in findall(r'[\d{1}@S]', row)]
        for row in notes.strip().splitlines()
    ]

    if debug:
        display_grid(grid)

    rows, cols = len(grid), len(grid[0])

    yv, xv = first((y, x) for y in range(rows) for x in range(cols) if grid[y][x] == '@')
    ys, xs = first((y, x) for y in range(rows) for x in range(cols) if grid[y][x] == 'S')

    assert xv == xs

    r = -1
    barrier_right = {(yv, x) for x in range(xv, rows)}
    barrier_left = {(yv, x) for x in range(xv)}

    while True:
        r += 1
        max_time = (r + 1) * 30

        circle = get_circle_II(yv, xv, r)

        start = [(0, ys, xs)]
        goals = {(y, xv) for y in takewhile(on_circle, range(rows - 1, -1, -1))}
        result = dijkstra(grid, 2 * max_time // 3, circle, barrier_right, start, goals)

        if not result:
            continue

        result = dijkstra(grid, max_time, circle, barrier_left, result, {(ys, xs)})

        if result:
            return r * one(result)[0]
```
