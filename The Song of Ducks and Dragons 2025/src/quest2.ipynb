The first page of the instructions describes a series of mathematical operations involving numbers called "complex". However, you don’t find them complex at all. Interestingly, you grasp the formulas instantly, as if they were instinctive. <span style="color:ivory">Each "complex" number consists of a pair of integers, X and Y</span>, written in square brackets as <code> [X,Y] </code>. These numbers can be added, multiplied, and divided, with the result always being another "complex" number.

To <span style="color:ivory">add two "complex" numbers</span>, simply sum the X and Y values separately, following this formula:

<pre><code>
[X1,Y1] + [X2,Y2] = [X1 + X2, Y1 + Y2]
</code></pre>

A few examples of addition:

<pre><code>
[1,1] + [2,2] = [1 + 2, 1 + 2] = [3,3]
[2,5] + [3,7] = [2 + 3, 5 + 7] = [5,12]
[-2,5] + [10,-1] = [-2 + 10, 5 + -1] = [8,4]
[-1,-2] + [-3,-4] = [-1 + -3, -2 + -4] = [-4,-6]
</code></pre>

<span style="color:ivory">Multiplying two "complex" numbers </span>follows a slightly more advanced formula:

<pre><code>
[X1,Y1] * [X2,Y2] = [X1 * X2 - Y1 * Y2, X1 * Y2 + Y1 * X2]
</code></pre>

Some examples of multiplication:

<pre><code>
[1,1] * [2,2] = [1 * 2 - 1 * 2, 1 * 2 + 1 * 2] = [2 - 2, 2 + 2] = [0,4]
[2,5] * [3,7] = [2 * 3 - 5 * 7, 2 * 7 + 5 * 3] = [6 - 35, 14 + 15] = [-29,29]
[-2,5] * [10,-1] = [-2 * 10 - 5 * -1, -2 * -1 + 5 * 10] = [-20 + 5, 2 + 50] = [-15,52]
[-1,-2] * [-3,-4] = [-1 * -3 - -2 * -4, -1 * -4 + -2 * -3] = [3 - 8, 4 + 6] = [-5,10]
</code></pre>

<span style="color:ivory">Dividing two "complex" numbers</span> involves dividing both X and Y pairs while ignoring the division rest. This means that when performing the division, we focus solely on the integer parts of each number, <span style="color:ivory">ignoring (not rounding!) any fractional components that might arise</span>

<pre><code>
[X1,Y1] / [X2,Y2] = [X1 / X2, Y1 / Y2]
</code></pre>

Some examples of division:

<pre><code>
[10,12] / [2,2] = [10 / 2, 12 / 2] = [5,6]
[11,12] / [3,5] = [11 / 3, 12 / 5] = [3,2]
[-10,-12] / [2,2] = [-10 / 2, -12 / 2] = [-5,-6]
[-11,-12] / [3,5] = [-11 / 3, -12 / 5] = [-3,-2]
</code></pre>

To ensure you understand these calculations correctly, the instructions provide a sample number, referred to as: A ( <span style="color:#eed699">your notes</span> ), along with a sequence of operations to perform on it.

Begin by setting the result to <code> [0,0] </code>. Then, complete <span style="color:ivory">three cycles of the following operations:</span>

- Multiply the result by itself.
- Divide the result by <code> [10,10] </code>.
- Add <code> A </code> to the result.

After three cycles, <span style="color:ivory">the final result is your answer.</span>

<span style="color:ivory">Example based on the following notes:</span>

<pre style=" border: 2px dotted lime;border-radius: 5px;padding:2px;width: max-content;">
A=[25,9]
</pre>

Here are the steps of the calculation, with the result stored as R:

<pre style=" border: 1px solid white;border-radius: 5px;padding:2px;width: max-content;">
R = [0,0]

Cycle I
R = R * R = [0,0]
R = R / [10,10] = [0,0]
R = R + A = [25,9]

Cycle II
R = R * R = [544,450]
R = R / [10,10] = [54,45]
R = R + A = [79,54]

Cycle III
R = R * R = [3325,8532]
R = R / [10,10] = [332,853]
R = R + A = [357,862]
</pre>

The final result is:<code style="color:ivory;font-weight:bolder"> [357,862] </code>.

<span style="color:#eed699">What number will you get at the end of the process?
The answer is a "complex" number, written in square brackets as [X,Y] without any spaces.</span>


```python
class ComplexNumber:
    def __init__(self, X: int, Y: int) -> None:
        self.X = X
        self.Y = Y

    def __add__(self, other: ComplexNumber) -> ComplexNumber:
        X = self.X + other.X
        Y = self.Y + other.Y
        return ComplexNumber(X, Y)

    def __floordiv__(self, other: ComplexNumber) -> ComplexNumber:
        X = int(self.X / other.X)
        Y = int(self.Y / other.Y)
        return ComplexNumber(X, Y)

    def __mul__(self, other: ComplexNumber) -> ComplexNumber:
        X = self.X * other.X - self.Y * other.Y
        Y = self.X * other.Y + self.Y * other.X
        return ComplexNumber(X, Y)

    def __eq__(self, other: ComplexNumber) -> bool:
        return self.X == other.X and self.Y == other.Y

    def __repr__(self) -> str:
        return f'[{self.X},{self.Y}]'


assert ComplexNumber(1, 1) + ComplexNumber(2, 2) == ComplexNumber(3, 3)
assert ComplexNumber(2, 5) + ComplexNumber(3, 7) == ComplexNumber(5, 12)
assert ComplexNumber(-2, 5) + ComplexNumber(10, -1) == ComplexNumber(8, 4)
assert ComplexNumber(-1, -2) + ComplexNumber(-3, -4) == ComplexNumber(-4, -6)

assert ComplexNumber(1, 1) * ComplexNumber(2, 2) == ComplexNumber(0, 4)
assert ComplexNumber(2, 5) * ComplexNumber(3, 7) == ComplexNumber(-29, 29)
assert ComplexNumber(-2, 5) * ComplexNumber(10, -1) == ComplexNumber(-15, 52)
assert ComplexNumber(-1, -2) * ComplexNumber(-3, -4) == ComplexNumber(-5, 10)

assert ComplexNumber(10, 12) // ComplexNumber(2, 2) == ComplexNumber(5, 6)
assert ComplexNumber(11, 12) // ComplexNumber(3, 5) == ComplexNumber(3, 2)
assert ComplexNumber(-10, -12) // ComplexNumber(2, 2) == ComplexNumber(-5, -6)
assert ComplexNumber(-11, -12) // ComplexNumber(3, 5) == ComplexNumber(-3, -2)
```

```python
import re

from test_utilities import test

tests = [
    {
        'name': 'Example',
        's': 'A=[25,9]',
        'expected': ComplexNumber(357, 862),
    },
]


@test(tests=tests)
def partI(s: str) -> ComplexNumber:
    X, Y = map(int, re.findall(r'\d+', s))
    A = ComplexNumber(X, Y)
    R = ComplexNumber(0, 0)
    D = ComplexNumber(10, 10)

    for _ in range(3):
        R = R * R
        R = R // D
        R = R + A

    return R
```

<span style="color:#eed699">How many points will be engraved according to the blacksmith’s plan?</span>


```python
from itertools import product

from matplotlib import pyplot as plt
from test_utilities import test

tests = [
    {
        'name': 'Example',
        's': 'A=[35300,-64910]',
        'width': 101,
        'expected': 4076,
    },
]


class Grid:
    def __init__(self, width: int = 101) -> None:
        self.width = width

    def count_valid(self, A: ComplexNumber, debug: bool) -> int:
        count = 0

        E = A + ComplexNumber(1_000, 1_000)
        fx = (E.X - A.X) // (self.width - 1)
        fy = (E.Y - A.Y) // (self.width - 1)

        for y, x in product(range(self.width), repeat=2):
            C = A + ComplexNumber(fx * x, fy * y)

            if self.valid(C, debug=debug):
                count += 1

        return count

    def display_grid(self, A: ComplexNumber) -> None:
        grid = [[0] * self.width for _ in range(self.width)]

        E = A + ComplexNumber(1_000, 1_000)
        fx = (E.X - A.X) // (self.width - 1)
        fy = (E.Y - A.Y) // (self.width - 1)

        for y, x in product(range(self.width), repeat=2):
            C = A + ComplexNumber(fx * x, fy * y)

            if self.valid(C):
                grid[y][x] = 1

        plt.figure(figsize=(6, 6))
        plt.matshow(grid, cmap='gray', fignum=1)  # 'gray' colormap: 0=black, 1=white

        plt.axis('off')  # Hide axis ticks
        plt.show()

    def valid(self, P: ComplexNumber, cycles: int = 100, debug: bool = False) -> bool:
        R = ComplexNumber(0, 0)
        D = ComplexNumber(100_000, 100_000)

        for c in range(cycles):
            R = R * R
            R = R // D
            R = R + P

            if not (-1_000_000 <= R.X <= 1_000_000 and -1_000_000 <= R.Y <= 1_000_000):
                if debug:
                    print(f'{P=}, {R=} C={c + 1} False')

                return False

        if debug:
            print(f'{P=}, {R=} C={c + 1} True')

        return True


@test(tests=tests)
def partII(s: str, width=101, debug: bool = False) -> int:
    X, Y = map(int, re.findall(r'[-+]?\d+', s))
    A = ComplexNumber(X, Y)
    g = Grid(width)

    if debug:
        g.display_grid(A)

    return g.count_valid(A, debug=debug)
```

## <span style="color:#eed699">Part III</span>

The symbol is too faint to be clearly visible, but you offer the blacksmith a solution. <span style="color:ivory">You can quickly calculate a far more detailed 1001x1001 grid</span> and map it onto the breastplate. The smaller points will actually be easier to engrave, and with the time saved on calculations, the plan should work.

Just to be certain, the blacksmith asks you to determine exactly how many points will need engraving so he can assess whether he is capable of completing such a task.

<span style="color:ivory">Example based on the following notes:</span>

<pre style=" border: 2px dotted lime;border-radius: 5px;padding:2px;width: max-content;">
A=[35300,-64910]
</pre>

After increasing the precision of the symbol to a 1001x1001 grid, the total number of engraved points for the same example as in Part II is <code style="color:ivory;font-weight:bolder"> 406954 </code>.

<span style="color:#eed699">What is the number of points you plan to engrave?</span>


```python
from typing import override

from test_utilities import test

tests = [
    {
        'name': 'Example',
        's': 'A=[35300,-64910]',
        'width': 1001,
        'expected': 406954,
    },
]


class GridII(Grid):
    @override
    def count_valid(self, A: ComplexNumber, debug: bool) -> int:
        count = 0

        XA, YA = A.X, A.Y
        XE, YE = XA + 1_000, YA + 1_000
        fx = (XE - XA) // (self.width - 1)
        fy = (YE - YA) // (self.width - 1)

        for y, x in product(range(self.width), repeat=2):
            XC, YC = XA + fx * x, YA + fy * y

            if self.valid(XC, YC, debug=debug):
                count += 1

        return count

    def valid(self, XP: int, YP: int, cycles: int = 100, debug: bool = False) -> bool:
        D = 100_000
        X, Y = 0, 0

        for c in range(cycles):
            # R = R * R =? [X1,Y1] * [X2,Y2] = [X1 * X2 - Y1 * Y2, X1 * Y2 + Y1 * X2]
            X, Y = X * X - Y * Y, X * Y + Y * X

            # R = R // D => [X1,Y1] / [X2,Y2] = [X1 / X2, Y1 / Y2]
            X, Y = int(X / D), int(Y / D)

            # R = R + P => [X1,Y1] + [X2,Y2] = [X1 + X2, Y1 + Y2]
            X, Y = X + XP, Y + YP

            if not (-1_000_000 <= X <= 1_000_000 and -1_000_000 <= Y <= 1_000_000):
                if debug:
                    print(f'P=[{XP},{YP}], R=[{X},{Y}] C={c + 1} False')

                return False

        if debug:
            print(f'P=[{XP},{YP}], R=[{X},{Y}] C={c + 1} True')

        return True

    def display_grid(self, A: ComplexNumber) -> None:
        grid = [[0] * self.width for _ in range(self.width)]

        XA, YA = A.X, A.Y
        XE, YE = XA + 1_000, YA + 1_000
        fx = (XE - XA) // (self.width - 1)
        fy = (YE - YA) // (self.width - 1)

        for y, x in product(range(self.width), repeat=2):
            C = A + ComplexNumber(fx * x, fy * y)

            XC, YC = XA + fx * x, YA + fy * y

            if self.valid(XC, YC):
                grid[y][x] = 1

        plt.figure(figsize=(10, 10))
        plt.matshow(grid, cmap='gray', fignum=1)  # 'gray' colormap: 0=black, 1=white

        plt.axis('off')  # Hide axis ticks
        plt.show()


@test(tests=tests)
def partIII(s: str, width=1001, debug: bool = False) -> int:
    X, Y = map(int, re.findall(r'[-+]?\d+', s))

    A = ComplexNumber(X, Y)
    g = GridII(width)

    return g.count_valid(A, debug=debug)
```
