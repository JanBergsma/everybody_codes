## <span style="color:#eed699">Part I</span>

> ## <span style="color:#eed699"> <img src='./images/book.svg' height=40 width=40 style="display:inline-block height:fit-content;vertical-align: middle; margin: 5px; ">Story section</span>
>
> After a long and exhausting flight, you reach the final passage. Your wings ache so much that you definitely won't be able to fly for a while. Beyond the last opening, you find a plant formation similar to the one that transported you here from the garden. You approach the free branches and rest your head against them.
>
> Once again, a golden flash erupts. The blinding light covers everything around you, and upon regaining your vision, you discover that you have once again found yourself in a completely unfamiliar place.
>
> You wander through a winding corridor, eventually reaching a room that surpasses your imagination of what the builders of this place were capable of. It looks like... a maze? But totally unlike any you've seen before.
>
> The space is completely open, allowing you to clearly see everything; however, each individual cell of this labyrinth is either a trampoline or an empty space that leads into a void. So you'll have to move by jumping from one trampoline to another. Luckily, your legs have rested during the flight, so that shouldn't be a problem. You'd rather fly over it all, but you don't want to wait for your wings to recover.
>
> <br/>

The maze resembles a triangle, with each side having an equal length. **Each trampoline is also a triangle, so the entire space can be divided into triangular cells.** You've never dealt with a space of this shape before. You mentally note down the whole map, but for some reason your brain does it in square cells rather than triangular ones.

You decide to take a moment to examine a small section of this map more closely <span style="color:#eed699">( your notes )</span>.. You focus on trampolines that share edges with one another, as you can definitely jump between them. Trampolines that only touch at triangle tips seem too far apart.

**You decide to count pairs of trampolines that you can jump between.** To do this, however, you need to convert the square map your brain suggests into a space made of triangles. The trampolines are marked as `T`, and the empty spaces you must avoid are marked as `#`. Probably out of habit, everything else has been filled with `.`, but in practice that area doesn't exist.

**Example based on the following notes:**

<pre style="border: 2px dotted lime;border-radius: 5px;padding:1rem;width: max-content;">
T#TTT###T##
.##TT#TT##.
..T###T#T..
...##TT#...
....T##....
.....#.....
</pre>

Let's convert the pixels generated by your brain into triangles, as they appear in practice. Each row of the pixel map represents a row of triangles arranged in such a way that adjacent triangles share one side, ultimately forming a straight line.
```python
from itertools import product

from test_utilities import test
from util import Str

test_cases = [
    {
        'name': 'Example I',
        'notes': """
            T#TTT###T##
            .##TT#TT##.
            ..T###T#T..
            ...##TT#...
            ....T##....
            .....#.....
        """,
        'expected': 7,
    },
    {
        'name': 'Example II',
        'notes': """
            T#T#T#T#T#T
            .T#T#T#T#T.
            ..T#T#T#T..
            ...T#T#T...
            ....T#T....
            .....T.....
        """,
        'expected': 0,
    },
    {
        'name': 'Example III',
        'notes': """
            T#T#T#T#T#T
            .#T#T#T#T#.
            ..#T###T#..
            ...##T##...
            ....#T#....
            .....#.....
        """,
        'expected': 0,
    },
]


class TriangleI(Str):
    def __init__(self, notes: str) -> None:
        self.grid = [list(row.strip()) for row in notes.strip().splitlines()]

    def trampoline_jumps(self) -> int:
        rows, cols = len(self.grid), len(self.grid[0])
        jumps = 0
        # horizontal links
        for row, col in product(range(rows), range(cols - 1)):
            if self.grid[row][col] != '.' and self.grid[row][col + 1] == '.':
                continue

            if self.grid[row][col] == self.grid[row][col + 1] == 'T':
                jumps += 1

        # vertical
        for row in range(1, rows):
            triangles = 0
            for col in range(cols):
                if self.grid[row][col] == '.':
                    continue

                if (
                    triangles & 1 == 0
                    and self.grid[row][col] == 'T'
                    and self.grid[row - 1][col] == 'T'
                ):
                    jumps += 1

                triangles += 1

        return jumps

    def __str__(self) -> str:
        s = ''
        for row in self.grid:
            s1, s2 = '', ''
            for ch in row:
                if ch == '.':
                    s1 += '   '
                    s2 += ' . '
                else:
                    s1 += '---'
                    s2 += f' {ch} '

            s += f'{s1}\n{s2}\n'
        return s


@test(tests=test_cases)
def partI(notes: str, debug: bool = False) -> int:
    triangle = TriangleI(notes)
    return triangle.trampoline_jumps()
```

<span style="color:#eed699">What is the minimum number of jumps needed to reach the golden trampoline?</span>


```python
from collections import defaultdict, deque
from collections.abc import Iterator

from more_itertools import first
from test_utilities import test
from util import Str

test_cases = [
    {
        'name': 'Example I',
        'notes': """
            TTTTTTTTTTTTTTTTT
            .TTTT#T#T#TTTTTT.
            ..TT#TTTETT#TTT..
            ...TT#T#TTT#TT...
            ....TTT#T#TTT....
            .....TTTTTT#.....
            ......TT#TT......
            .......#TT.......
            ........S........
        """,
        'expected': 32,
    },
]
type Cell = tuple[int, int]
type Jump = tuple[Cell, Cell]


class TriangleII(TriangleI):
    def __init__(self, notes: str) -> None:
        super().__init__(notes)
        self.E = self._get_symbol('E')
        self.S = self._get_symbol('S')
        self.graph = self._create_graph()

    def minimum_jumps(self, fr: Cell, to: Cell) -> int:
        queue = deque([fr])
        jumps = 0
        seen = set()

        while queue:
            for _ in range(len(queue)):
                node = queue.popleft()

                if node == to:
                    return jumps

                if node in seen:
                    continue

                seen.add(node)

                queue.extend(neighbor for neighbor in self.graph[node])

            jumps += 1
        return -1

    def _create_graph(self) -> dict[Cell, set[Cell]]:
        graph = defaultdict(set)

        for fr, to in self._valid_neighbors():
            graph[fr].add(to)
            graph[to].add(fr)

        return dict(graph)

    def _get_symbol(self, symbol):
        return first(
            (row, col)
            for row in range(len(self.grid))
            for col in range(len(self.grid[0]))
            if self.grid[row][col] == symbol
        )

    def _valid_neighbors(self) -> Iterator[Jump]:
        rows, cols = len(self.grid), len(self.grid[0])
        # horizontal links
        for row, col in product(range(rows), range(cols - 1)):
            if self.grid[row][col] in 'TSE' and self.grid[row][col + 1] in 'TSE':
                yield (row, col), (row, col + 1)

        # vertical links
        for row in range(1, rows):
            triangles = 0
            for col in range(cols):
                if self.grid[row][col] == '.':
                    continue

                if (
                    (triangles & 1 == 0 or row == rows - 1)
                    and self.grid[row][col] in 'TSE'
                    and self.grid[row - 1][col] in 'TSE'
                ):
                    yield (row - 1, col), (row, col)

                triangles += 1


@test(tests=test_cases)
def partII(notes: str, debug: bool = False) -> int:
    triangle = TriangleII(notes)
    return triangle.minimum_jumps(triangle.S, triangle.E)
```

```python
from math import sqrt
from typing import TypedDict

from IPython.display import HTML, display
from test_utilities import test
from util import Str

test_cases = [
    {
        'name': 'Example I',
        'notes': """
            T####T#TTT##T##T#T#
            .T#####TTTT##TTT##.
            ..TTTT#T###TTTT#T..
            ...T#TTT#ETTTT##...
            ....#TT##T#T##T....
            .....#TT####T#.....
            ......T#TT#T#......
            .......T#TTT.......
            ........TT#........
            .........S.........
        """,
        'expected': 32,
    },
]


class Triangle(TypedDict):
    i: int
    j: int
    up: bool
    value: str


class TriangleIII(Str):
    def __init__(self, notes: str) -> None:
        self.triangles = self._parse(notes)
        self.S = first((t['i'], t['j']) for t in self.triangles if t['value'] == 'S')
        self.E = first((t['i'], t['j']) for t in self.triangles if t['value'] == 'E')
        self.n = max(t['j'] for t in self.triangles)

    @staticmethod
    def _parse(notes: str) -> list[Triangle]:
        grid = [list(row.strip()) for row in notes.strip().splitlines()]
        rows, cols = len(grid), len(grid[0])
        triangles = []

        for i, j in product(range(rows), range(cols)):
            ch = grid[i][j]
            if ch == '.':
                continue  # empty triangle

            up = (i + j) % 2 == 1

            triangles.append({'i': i, 'j': j, 'up': up, 'value': ch})

        return triangles

    def rotate120(self, i, j):
        """
        Rotate an up/down triangle coordinate (i, j)
        by 120 degrees clockwise.
        """
        return j, -i - j

    def triangles_to_svg(self, i_me: int, j_me: int, size: int = 20, padding: int = 10) -> str:
        """
        Create an SVG string from a list of triangles.
        Each triangle is a dict with:
            { "i": int, "j": int, "up": bool, "value": any }

        Geometry:
            Up triangle:
                apex at (x, y)
                base at y + h
            Down triangle:
                apex at (x, y + h)
                base at y

        Where:
            h = size * sqrt(3)/2
        """
        print(f'{i_me, j_me=}')

        h = size * sqrt(3) / 2

        # Convert (i, j) to geometric coordinates
        def tri_center(i, j):
            # UpDownTriGrid axial geometry:
            # Each step in j shifts x by size/2
            # Each step in i shifts y by h
            x = j * (size / 2)
            y = i * h
            return x, y

        # Build SVG elements
        svg_elements = []

        for i, j, up, value in (t.values() for t in self.triangles):
            cx, cy = tri_center(i, j)

            if up:
                # Up triangle
                p1 = (cx, cy)  # top
                p2 = (cx - size / 2, cy + h)  # bottom-left
                p3 = (cx + size / 2, cy + h)  # bottom-right
            else:
                # Down triangle
                p1 = (cx, cy + h)  # bottom
                p2 = (cx - size / 2, cy)  # top-left
                p3 = (cx + size / 2, cy)  # top-right

            # Color: filled if value != '.'
            if value == 'T':
                fill = 'grey'
            elif value == 'S':
                fill = 'Blue'
            elif value == 'E':
                fill = 'lime'
            else:
                fill = 'black'

            if (i, j) == (i_me, j_me):
                fill = 'Orange'

            svg_elements.append(
                f'<polygon points="{p1[0]},{p1[1]} {p2[0]},{p2[1]} {p3[0]},{p3[1]}" '
                f'stroke="white" fill="{fill}" />'
            )

        # Compute bounding box
        all_x = []
        all_y = []
        for tri in self.triangles:
            cx, cy = tri_center(tri['i'], tri['j'])
            all_x.extend([cx - size, cx + size])
            all_y.extend([cy - size, cy + size])

        min_x = min(all_x) - padding
        max_x = max(all_x) + padding
        min_y = min(all_y) - padding
        max_y = max(all_y) + padding

        width = max_x - min_x
        height = max_y - min_y

        # Wrap in SVG
        svg = [
            f'<svg xmlns="http://www.w3.org/2000/svg" width="{width}" height="{height}" '
            f'viewBox="{min_x} {min_y} {width} {height}">'
        ]
        svg.extend(svg_elements)
        svg.append('</svg>')

        return '\n'.join(svg)

    def __str__(self) -> str:
        rows, cols = (
            max(t['i'] for t in self.triangles) + 1,
            max(t['j'] for t in self.triangles) + 1,
        )
        triangles = {(t['i'], t['j']) for t in self.triangles}

        return '\n'.join(
            ''.join(f'({i:>3},{j:>3})' if (i, j) in triangles else 9 * ' ' for j in range(cols))
            for i in range(rows)
        )


#
@test(tests=test_cases)
def partIII(notes: str, debug: bool = False) -> int:
    triangle = TriangleIII(notes)
    print(triangle)
    print()
    display(HTML(triangle.triangles_to_svg(*triangle.S)))
    print(me102 := triangle.rotate120(*triangle.S))
    display(HTML(triangle.triangles_to_svg(0, 0)))
    print(triangle.rotate120(*me102))
    display(HTML(triangle.triangles_to_svg(0, 18)))

    return -1
```
