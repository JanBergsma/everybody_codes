The knight came to the village to get supplies for the army. He has a number of horse-drawn wagons that are full of different-sized wooden crates. The carts are already filled with some goods. There are many empty crates still, but it's clear that not all of them will fit on the carts. But the knight has an idea. For as long as possible, he wants to put smaller crates inside bigger ones and bigger crates inside even bigger ones. Therefore, the crates will take up a lot less space since only the biggest one will be left from each set.

The knight gives you a list of numbers ( <span style="color:#eed699">your notes</span> ). Each number represents the size of a single crate. If two crates differ in size, the smaller one can be packed within the larger one. A sequence of crates packed one inside another is called a set. <span style="color:ivory">The knight is curious about the largest possible set that can be formed, in terms of the total size of all crates forming it.</span>

<span style="color:ivory">Example based on the following notes:</span>

<pre style=" border: 2px dotted lime;border-radius: 5px;padding:2px;width: max-content;">
10,5,1,10,3,8,5,2,2
</pre>

A few sample sets of crates that can be packed one inside the other are listed below. For example, the notation <code> 10 > 8 > 5 </code> means that inside a crate of size 10, there is a crate of size 8, and inside that, there is a crate of size 5.

<pre style=" border: 1px solid white;border-radius: 5px;padding:2px;width: max-content;">
10 > 8 > 5 > 3 > 2 > 1   the sum of the sizes: 10 + 8 + 5 + 3 + 2 + 1 = 29
10 > 8 > 5 > 2 > 1       the sum of the sizes: 10 + 8 + 5 + 2 + 1 = 26
10 > 5 > 3 > 2           the sum of the sizes: 10 + 5 + 3 + 2 = 20
10 > 5 > 2               the sum of the sizes: 10 + 5 + 2 = 17
</pre>

There are many ways to pack crates; however, out of all the packing options, the largest possible set of crates is <code style:color:ivory; font-weight:bolder> 29 </code>.

<span style="color:#eed699">What is the largest possible set of crates that can be formed from a given list?</span>


```python
import re

from test_utilities import test

tests = [
    {
        'name': 'Example',
        's': '10,5,1,10,3,8,5,2,2',
        'expected': 29,
    },
]


@test(tests=tests)
def partI(s: str) -> int:
    crates = sorted(int(crate) for crate in re.findall(r'\d+', s))
    prev_crate = 0
    max_set = 0

    for crate in crates:
        if crate > prev_crate:
            max_set += crate
            prev_crate = crate

    return max_set
```

```python
with open('../inputs/everybody_codes_e2025_q03_p1.txt') as f:
    notes1 = f.read()


print(f'Part I: {partI(notes1)}')
```

To pack the mushroom into exactly 20 crates according to the rules, the following sizes should be used:

<pre style=" border: 1px solid white;border-radius: 5px;padding:2px;width: max-content;">
75 > 72 > 65 > 64 > 63 > 61 > 57 > 51 > 49 > 48 > 43 > 40 > 32 > 16 > 14 > 13 > 9 > 4 > 3 > 2

The total size of this set is 781, and it is the smallest possible set of 20 crates for this list of crates.

<span style="color:#eed699">Pack the mushroom into exactly <code style="color:#eed699"> 20 </code> crates. What is the smallest possible set of the crates that can be used for this purpose?</span>


```python
from functools import cache

from test_utilities import test

tests = [
    {
        'name': 'Example',
        's': '4,51,13,64,57,51,82,57,16,88,89,48,32,49,49,2,84,65,49,43,9,13,2,3,75,72,63,48,61,14,40,77',
        'expected': 781,
    },
]

INF = 10**18


def partII_recursive(s: str) -> int:
    # ridiculous slow 4 to 5 mins. Has to plow through way too much combinations
    # 20 choose 152 = 4820043373396040980854930
    @cache
    def smallest_set(i: int, prev_crate: int, size: int, nr_crates: int) -> int:
        if nr_crates == 0:
            nonlocal smallest
            smallest = min(size, smallest)
            return size

        if size >= smallest:
            return INF

        if i >= n - nr_crates:
            return INF

        result = smallest_set(i + 1, prev_crate, size, nr_crates)

        if crates[i] > prev_crate:
            return min(smallest_set(i + 1, crates[i], size + crates[i], nr_crates - 1), result)

        return result

    crates = sorted({int(crate) for crate in re.findall(r'\d+', s)})
    n = len(crates)
    smallest = INF
    return smallest_set(0, -INF, 0, 20)


@test(tests=tests)
def partII(s: str) -> int:
    # greedy
    crates = sorted({int(crate) for crate in re.findall(r'\d+', s)})  # sorted set
    return sum(crates[:20])
```

## <span style="color:#eed699">Part III</span>

The mushroom and other goods are loaded onto the carts. The only task left is to pack the empty crates ( <span style="color:#eed699">your notes</span> ).

Your final task is to pack these crates as efficiently as possible. The rules for packing remain unchanged. There is also no limit on the number of crates that can be packed into one set. <span style="color:ivory">The goal is to pack everything into as few sets as possible.</span>

<span style="color:ivory">Example based on the following notes:</span>

<pre style=" border: 2px dotted lime;border-radius: 5px;padding:2px;width: max-content;">
4,51,13,64,57,51,82,57,16,88,89,48,32,49,49,2,84,65,49,43,9,13,2,3,75,72,63,48,61,14,40,77
</pre>

The given list can be packed into <code style="color:ivory"> 3 </code> sets.

Here is an example of such packaging:

<pre style=" border: 1px solid white;border-radius: 5px;padding:2px;width: max-content;">
88 > 82 > 64 > 57 > 51 > 49 > 16 > 13 > 4 > 2
89 > 84 > 65 > 61 > 57 > 51 > 49 > 48 > 43 > 32 > 2
77 > 75 > 72 > 63 > 49 > 48 > 40 > 14 > 13 > 9 > 3</pre>

<span style="color:#eed699">Pack the given list as efficiently as possible. How many sets do you need for this?</span>


```python
from collections import Counter

from test_utilities import test

tests = [
    {
        'name': 'Example',
        's': '4,51,13,64,57,51,82,57,16,88,89,48,32,49,49,2,84,65,49,43,9,13,2,3,75,72,63,48,61,14,40,77',
        'expected': 3,
    },
]

INF = 10**18


@test(tests=tests)
def partIII(s: str) -> int:
    crates = Counter(int(crate) for crate in re.findall(r'\d+', s))
    return max(crates.values())
```

```python
with open('../inputs/everybody_codes_e2025_q03_p3.txt') as f:
    notes3 = f.read()

print(f'Part III: {partIII(notes3)}')
```
