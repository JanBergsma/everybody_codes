The miller hands you a list of numbers representing the gears of the first mill ( <span style="color:#eed699">your notes</span> ). Each number corresponds to the number of teeth on the given gear. The first gear on the list is connected directly to the windmill. The last gear is responsible for driving the millstones. The gears are connected to each other in the order on the list.

To calibrate the first mill, you need to </span style="color:ivory">calculate the number of full turns the last gear will make if the first one makes exactly <code style="color:ivory;border: 1px dotted white;border-radius: 5px;padding:2px;width: max-content;"> 2025 </code> full rotations.<span>

<span style="color:ivory;">Example based on the following notes:</span>

<pre style=" border: 2px dotted lime;border-radius: 5px;padding:2px;width: max-content;">
128
64
32
16
8
</pre>

The second gear (64) has half as many teeth as the first one (128). For this reason, for every full turn of the first gear, the second gear will make two full turns.

The same relationship occurs between successive pairs of gears. The gear (32) will therefore turn twice for one turn of the gear (64) and thus 4 times for each turn of the gear (128).

Analyzing the movement of gears (16) and (8) analogously, it can be observed that for each single turn of gear (128), the last gear (8) will make exactly 16 full turns.

Multiplying 16 by 2025 (the number of full turns of the first gear) gives you the final result for this example: <code style="color:ivory;font-weight:bolder"> 32400 </span>.

<pre style=" border: 2px dotted lime;border-radius: 5px;padding:2px;width: max-content;">
102
75
50
35
13
</pre>

In the second example, the relationship between the sizes of the gears no longer fits together perfectly, but you notice a faster and equally accurate method for calculating the required value. In this example, after 2025 full turns of the first gear, the last one will complete <code style="color:ivory;font-weight:bolder"> 15888 </span> full turns.

<span style="color:#eed699">How many full turns will the last gear make if the first one turns exactly <code style="color:ivory;border: 1px dotted #eed699;border-radius: 5px;padding:2px;width: max-content;"> 2025</code> times?</span>


```python
import re

from test_utilities import test

tests = [
    {
        'name': 'Example 1',
        'notes': """
            128
            64
            32
            16
            8
        """,
        'expected': 32400,
    },
    {
        'name': 'Example 2',
        'notes': """
            102
            75
            50
            35
            13
        """,
        'expected': 15888,
    },
]


@test(tests=tests)
def partI(notes: str, rotations: int = 2025) -> int:
    prev, *gears = [int(gear) for gear in re.findall(r'\d+', notes)]

    factor = 1

    for gear in gears:
        factor *= prev / gear
        prev = gear

    return int(factor * rotations)
```

```python
with open('../inputs/everybody_codes_e2025_q04_p1.txt') as f:
    notes1 = f.read()


print(f'Part I: {partI(notes1)}')
```

## <span style="color:#eed699">Part II</span>

The first mill is starting to operate and produce flour! You promptly proceed to calibrate the next mill ( <span style="color:#eed699">your notes</span> ). The gears' operation and list format are the same as for the first mill. Only the method of calibrating is different.
To calibrate the second mill, you need to provide the minimum number of full turns that the first gear must make, so that the last one turns at least <code> 10000000000000 </code> times.

<span style="color:ivory">Example based on the following notes:</span>

For the last gear to turn at least <code> 10000000000000 </code> times, the first gear must make at least <code style="color:ivory;font-weight:bolder"> 625000000000 </code> full turns.

<pre style=" border: 2px dotted lime;border-radius: 5px;padding:2px;width: max-content;">
102
75
50
35
13
</pre>

For the last gear to turn at least <code> 10000000000000 </code> times, the first gear must make at least <code style="color:ivory;font-weight:bolder"> 1274509803922 </code> full turns.

<span style="color:#eed699">What is the minimum number of full turns for the first gear, after which the last one turns at least<code style="color:#eed699"> 10000000000000 </code> times?</span>


```python
from math import ceil

from test_utilities import test

tests = [
    {
        'name': 'Example 1',
        'notes': """
            128
            64
            32
            16
            8
        """,
        'expected': 625000000000,
    },
    {
        'name': 'Example 2',
        'notes': """
            102
            75
            50
            35
            13
        """,
        'expected': 1274509803922,
    },
]


@test(tests=tests)
def partII(notes: str, rotations: int = 10_000_000_000_000) -> int:
    prev, *gears = [int(gear) for gear in re.findall(r'\d+', notes)]

    factor = 1

    for gear in gears:
        factor *= gear / prev
        prev = gear

    return int(ceil(factor * rotations))
```

```python
with open('../inputs/everybody_codes_e2025_q04_p2.txt') as f:
    notes2 = f.read()

print(f'Part II: {partII(notes2)}')
```

## <span style="color:#eed699">Part III</span>

The second mill is also starting up at full capacity! The miller is regaining his positive mood. Only the third mill remains to be calibrated.

The mechanism of the last mill differs from the previous ones. The list of gears provided by the miller( <span style="color:#eed699">your notes</span> ) tarts and ends with single gears, just like before. <span style="color:ivory">However, between them are pairs of gears mounted on common shafts.</span> Thanks to this, they always turn at the same speed, meaning that if one of the gears makes, for example, 10 full turns, the other will also make exactly 10 full turns.

The list represents gear pairs as numbers separated by a pipe character, such as <code> 15|45</code> . The number on the left represents the gear connected to the previous gear on the list. The number on the right represents the gear connected to the subsequent gear on the list.
Sample list of gears:

<pre style=" border: 1px solid white;border-radius: 5px;padding:2px;width: max-content;">
10
7|14
8|16
6
</pre>

is connected in the following way:

<code>10 > 7|14 > 8|16 > 6</code>

and it looks like this in practice:

<img src="./images/q4_p3_cogs.svg" alt="Description" width="200">

To calibrate the last mill, it is necessary to <span style="color:ivory">calculate how many full turns the last gear makes, if the first one turns exactly <code style="color:ivory;border: 1px dotted white;border-radius: 5px;padding:2px;width: max-content;"> 100</code> times.</span>

<span style="color:ivory">Example based on the following notes:</span>

<pre style=" border: 2px dotted lime;border-radius: 5px;padding:2px;width: max-content;">
5
5|10
10|20
5
</pre>

The second gear turns once after the first gear turns once because the left side gear of the first pair has the same number of teeth as the first gear on the list (5).

The next gear also turns once because its gear on the left side has exactly the same number of teeth as the gear of the previous pair of gears on the right side (10).

The last gear with 5 teeth turns 4 times because it is connected to a gear with 20 teeth, which has only turned once.

After 1 turn of the first gear, the last gear turns exactly 4 times; therefore, after <code style="color:ivory;border: 1px dotted white;border-radius: 5px;padding:2px;width: max-content;"> 100</code> full turns of the first gear, the last gear will turn exactly <code style="color:ivory;font-weight:bolder;"> 400</code> times.

<pre style=" border: 2px dotted lime;border-radius: 5px;padding:2px;width: max-content;">
5
7|21
18|36
27|27
10|50
10|50
11
</pre>

In the second example, after <code style="color:ivory;border: 1px dotted white;border-radius: 5px;padding:2px;width: max-content;"> 100</code> full turns of the first gear, the last gear will make <code style="color:ivory;font-weight:bolder;"> 6818</code> full turns.

<span style="color:#eed699">How many full turns will the last gear make if the first one turns exactly <code style="color:ivory;border: 1px dotted #eed699;border-radius: 5px;padding:2px;width: max-content;"> 100</code> times?</span>


```python
from test_utilities import test

tests = [
    {
        'name': 'Example 1',
        'notes': """
            5
            5|10
            10|20
            5
        """,
        'expected': 400,
    },
    {
        'name': 'Example 2',
        'notes': """
            5
            7|21
            18|36
            27|27
            10|50
            10|50
            11
        """,
        'expected': 6818,
    },
]


@test(tests=tests)
def partIII(notes: str, rotations: int = 100) -> int:
    prev, *gears = [
        [int(gear) for gear in re.findall(r'\d+', line)] for line in notes.strip().splitlines()
    ]

    factor = 1

    for gear in gears:
        factor *= prev[-1] / gear[0]
        prev = gear

    return int(factor * rotations)
```

```python
with open('../inputs/everybody_codes_e2025_q04_p3.txt') as f:
    notes3 = f.read()

print(f'Part III: {partIII(notes3)}')
```
