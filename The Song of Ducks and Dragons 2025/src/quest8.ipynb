The artwork referenced in the book is called string-art. To create such a work, you need a wooden board, tiny nails, and a very long piece of thread. First, you need to draw a circle on the board. Next, hammer nails around this circle at equal distances. The more nails you use, the more details you will ultimately achieve. Once you've finished hammering, number the nails in a clockwise direction, starting from `1`.

The process of creating a string-art image involves threading a string between nails according to a specified sequence of numbers. You should execute the entire sequence with a single thread, connecting the nails with straight lines, without breaking the thread at any point.

The first part of the challenge allows you to verify your understanding of all the rules. Assuming you have <code style="color:ivory;border: 1px dotted #eed699;border-radius: 5px;padding:2px;width: max-content;">32</code> nails, construct string-art according to the given nails sequence ( <span style="color:#eed699">your notes</span> ) and count how many times the thread passes through the exact centre of the circle.

<span style="color:ivory">Example based on the following notes:</span>

<pre style=" border: 2px dotted lime;border-radius: 5px;padding:2px;width: max-content;">
1,5,2,6,8,4,1,7,3
</pre>

For the given example, the thread passes exactly through the centre of the circle <code style="color:ivory;border: 1px dotted white;border-radius: 5px;padding:2px;width: max-content;"> 4</code> times, in the first, third, fifth, and eighth steps.

<span style="color:#eed699">How many times does the thread from the given sequence pass exactly through the centre of the circle?</span>


```python
import re

from more_itertools import pairwise
from test_utilities import test

tests = [
    {
        'name': 'Example 1',
        'notes': '1,5,2,6,8,4,1,7,3',
        'nails': 8,
        'expected': 4,
    },
]


@test(tests=tests)
def partI(notes: str, nails: int) -> int:
    nail_sequence = (int(nail) - 1 for nail in re.findall(r'\d+', notes))

    return sum(
        1 for prev_nail, nail in pairwise(nail_sequence) if nail == (prev_nail + nails // 2) % nails
    )
```

```python
with open('../inputs/everybody_codes_e2025_q08_p1.txt') as f:
    notes1 = f.read()


print(f'Part I: {partI(notes1, 32)}')
```

The number of knots needed to create the entire string-art above equals:

<span style="color:#eed699">How many knots need to be tied to create the given string-art?</span>


```python
from more_itertools import minmax
from test_utilities import test

tests = [
    {
        'name': 'Example 1',
        'notes': '1,5,2,6,8,4,1,7,3,5,7,8,2',
        'nails': 8,
        'expected': 21,
    },
]


@test(tests=tests)
def partII_bf(notes: str, nails: int, debug: bool = False) -> int:
    # slow 7-8s
    chords = [minmax(map(int, line)) for line in pairwise(re.findall(r'\d+', notes))]
    lines = []
    knots = 0

    for step, line1 in enumerate(chords, start=1):
        temp = 0

        for line2 in lines:
            (nail1, nail2), (nail3, nail4) = minmax(line1, line2)

            if nail1 < nail3 < nail2 < nail4:
                temp += 1

        if debug:
            print(f'{step=} requires knots {temp}')

        knots += temp

        lines.append(line1)

    return knots


@test(tests=tests)
def partII(notes: str, nails: int, debug: bool = False) -> int:
    chords = [minmax(map(int, line)) for line in pairwise(re.findall(r'\d+', notes))]

    chords_per_nail = [[] for _ in range(nails + 1)]

    total_knots = 0

    for step, (nail_fr, nail_to) in enumerate(chords, start=1):
        knots = sum(
            1
            for fr in range(nail_fr + 1, nail_to)
            for to in chords_per_nail[fr]
            if not nail_fr <= to <= nail_to
        )

        if debug:
            print(f'{step=} requires knots {knots}')

        total_knots += knots

        chords_per_nail[nail_fr].append(nail_to)
        chords_per_nail[nail_to].append(nail_fr)

    return total_knots
```

<span style="color:#eed699">How many threads can be cut at most with a single sword strike between any two nails?</span>


```python
from itertools import combinations, product
from pprint import pprint

from test_utilities import test

tests = [
    {
        'name': 'Example 1',
        'notes': '1,5,2,6,8,4,1,7,3,6',
        'nails': 8,
        'expected': 7,
    },
]


@test(tests=tests)
def partIII_bf(notes: str, nails: int, debug: bool = False) -> int:
    # too slow runs around 3 to 4 minutes
    chords = [minmax(map(int, line)) for line in pairwise(re.findall(r'\d+', notes))]
    cuts_max = 0

    for line1 in combinations(range(1, nails + 1), 2):
        cuts = 0

        for line2 in chords:
            (nail1, nail2), (nail3, nail4) = minmax(line1, line2)

            if nail1 < nail3 < nail2 < nail4 or line1 == line2:
                cuts += 1

        if debug:
            print(f'{line1=} requires knots {cuts}')

        if cuts > cuts_max:
            cuts_max = cuts

    return cuts_max
@test(tests=tests)
def partIII_semi_bf(notes: str, nails: int, debug: bool = False) -> int:
    # Still slow between 9-10s
    chords = [minmax(map(int, line)) for line in pairwise(re.findall(r'\d+', notes))]

    chords_per_nail = [[] for _ in range(nails + 1)]
    for nail_fr, nail_to in chords:
        chords_per_nail[nail_fr].append(nail_to)
        chords_per_nail[nail_to].append(nail_fr)

    max_cuts = 0

    for nail_fr in range(1, nails + 1):
        for nail_to in range(nail_fr + 2, nails + 1):
            cuts = sum(
                1
                for fr in range(nail_fr + 1, nail_to)
                for to in chords_per_nail[fr]
                if not (nail_fr <= to <= nail_to)
            )
            cuts += chords_per_nail[nail_fr].count(nail_to)
            max_cuts = max(max_cuts, cuts)

    return max_cuts

@test(tests=tests)
def partIII(notes: str, nails: int, debug: bool = False) -> int:
    #  solution from https://www.reddit.com/user/i_have_no_biscuits/
    # because my own solution did run in 9 to 10 seconds
    chords = [minmax(map(int, line)) for line in pairwise(re.findall(r'\d+', notes))]

    chords_freq = [[0] * (nails + 1) for _ in range(nails + 1)]

    for fr, to in chords:
        chords_freq[fr][to] += 1
        chords_freq[to][fr] += 1

    acc_chords_freq = [[0] * (nails + 1) for _ in range(nails + 1)]

    for fr, to in product(range(1, nails + 1), repeat=2):
        acc_chords_freq[fr][to] = acc_chords_freq[fr][to - 1] + chords_freq[fr][to]

    best_cuts = 0
    for fr in range(1, nails + 1):
        cuts = 0
        for to in range(fr + 2, nails + 1):
            cuts += acc_chords_freq[to][fr] - acc_chords_freq[to][to - 2]
            cuts += (
                acc_chords_freq[to - 1][fr - 1]
                + acc_chords_freq[to - 1][nails]
                - acc_chords_freq[to - 1][to]
            )
            best_cuts = max(best_cuts, cuts + chords_freq[fr][to])

    return best_cuts


```
