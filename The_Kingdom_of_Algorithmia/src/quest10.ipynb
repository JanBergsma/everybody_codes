# 🐦 Everybody Codes 2024 — Quest 9, Part I

## The Runic Grid of the Memory Stack Desert

The knights seeking entry into the **Order of the Golden Duck** have reached the **Memory Stack Desert**, where construction of the shrine to **Nullpointer** is underway.  
The shrine will be decorated with **runic symbols**, and the knights must learn how to complete the missing parts of these runes.

Each knight receives an **8×8 grid** as a practice sample.

---

## 🧩 Grid Rules

- The **four corners** of the grid contain `*` and **cannot** hold runic symbols.
- The **edges** (connecting the corners) are already engraved with runic symbols.
- This leaves a **4×4 empty square** in the center, filled with `.` characters.
- Each `.` must be replaced with **exactly one runic symbol**.

### How to fill each dot

For each empty cell:

1. Look at the **row** it belongs to.
2. Look at the **column** it belongs to.
3. The correct symbol is the **only symbol that appears in both** the row and the column.

There is **always exactly one matching symbol**.

---

## 📜 Reading the Final Runic Word

Once all missing symbols are filled in:

- Read the grid **row by row**, left to right, top to bottom.
- Extract **only the runic symbols** (ignore `*`).
- The resulting sequence forms the **runic word**.
- The runic word contains **no repeated symbols**.

Your task:

> **Determine the runic word formed after completing your sample grid (your notes).**

---

## 📘 Example

Given the sample grid:

```
**PCBS**
**RLNW**
BV....PT
CR....HZ
FL....JW
SG....MN
**FTZV**
**GMJH**
```

### Step-by-step filling

For the first dot:

- Row contains: `BVPT`
- Column contains: `PRFG`
- Intersection: `P`

Continue this process for all dots.

Final completed grid:

```
**PCBS**
**RLNW**
BVPTBVPT
CRRCZHHZ
FLFLJWJW
SGGMNSMN
**FTZV**
**GMJH**
```

### Final runic word

```
PTBVRCZHFLJWGMNS
```

---

## 🎯 Your Task

> **What runic word will be formed upon solving your sample grid?**


````python
import re
from itertools import product
from string import ascii_uppercase
from typing import Self

from more_itertools import one
from test_utilities import test

tests = [
    {
        "name": "Example",
        "s": """
            **PCBS**
            **RLNW**
            BV....PT
            CR....HZ
            FL....JW
            SG....MN
            **FTZV**
            **GMJH**
        """,
        "expected": "PTBVRCZHFLJWGMNS",
    },
]


class Grid:
    BASE_POWER = {c: i for i, c in enumerate(ascii_uppercase, start=1)}

    def __init__(self, grid: list[list[str]]) -> None:
        self.grid = grid
        self._rune()

    def _rune(self) -> None:
        padding = self.grid[0].count("*") // 2
        grid = [set(c for c in row if c not in ".*") for row in self.grid]
        grid_T = [set(c for c in col if c not in ".*") for col in zip(*self.grid)]
        dimension = len(grid)

        for r, c in product(range(padding, dimension - padding), repeat=2):
            intersect = grid[r] & grid_T[c]
            if len(intersect) == 1:
                self.grid[r][c] = one(intersect)

    def get_rune(self) -> str:
        padding = self.grid[0].count("*") // 2
        dimension = len(self.grid)

        return "".join(
            self.grid[r][c]
            for r, c in product(range(padding, dimension - padding), repeat=2)
        )

    def power(self) -> int:
        return sum(
            p * self.BASE_POWER[c] for p, c in enumerate(self.get_rune(), start=1)
        )

    def __str__(self) -> str:
        return "\n".join("".join(r) for r in self.grid)

    @classmethod
    def create_from_str(cls, s: str) -> Self:
        grid = [list(row) for row in re.findall(r"\S+", s)]
        return cls(grid)


@test(tests=tests[:])
def part_I(s: str) -> str:
    g = Grid.create_from_str(s)
    return g.get_rune()
````

````python
with open("../inputs/everybody_codes_e2024_q10_p1.txt") as f:
    notes1 = f.read()

print(f"Part I: {part_I(notes1)}")
````

# 🐦 Everybody Codes 2024 — Quest 9, Part II

## Calculating the Power of Runic Words

After discovering the runic word from your grid, you advance to the next stage of shrine‑building training.

Every runic word created using the rules from Part I has a **power value**, determined by the positions and alphabetical values of its symbols.

---

## 🔡 Base Power of Symbols

Each letter has a **base power** equal to its alphabetical index:

| Letter | Power | Letter | Power |
| ------ | ----- | ------ | ----- |
| A      | 1     | N      | 14    |
| B      | 2     | O      | 15    |
| C      | 3     | P      | 16    |
| D      | 4     | Q      | 17    |
| E      | 5     | R      | 18    |
| F      | 6     | S      | 19    |
| G      | 7     | T      | 20    |
| H      | 8     | U      | 21    |
| I      | 9     | V      | 22    |
| J      | 10    | W      | 23    |
| K      | 11    | X      | 24    |
| L      | 12    | Y      | 25    |
| M      | 13    | Z      | 26    |

---

## ⚡ Effective Power of a Symbol

For a runic word:

- Positions are numbered starting from **1**.
- The **effective power** of a symbol is:

$$
\text{effective power} = \text{base power} \times \text{position}
$$

The **total power** of the word is the **sum of all effective powers**.

---

## 📘 Example

For the runic word from Part I:

```
PTBVRCZHFLJWGMNS
```

Its power is:

```
1 * 16 (P)
+ 2 * 20 (T)
+ 3 * 2  (B)
+ ...
+ 16 * 19 (S)
= 1851
```

---

## 🎯 Your Task

You are given a **stack of training samples** (your notes).  
For each sample:

1. Solve the 8×8 grid (as in Part I).
2. Extract the runic word.
3. Compute its power.
4. Sum the powers of **all** runic words.

> **What is the total power of all runic words in your training samples?**

---


````python
from test_utilities import test

tests = [
    {
        "name": "Example",
        "s": """
            **PCBS**
            **RLNW**
            BV....PT
            CR....HZ
            FL....JW
            SG....MN
            **FTZV**
            **GMJH**
        """,
        "expected": 1851,
    },
]


class Grids:
    def __init__(self, s: str) -> None:
        self.grids = self._parse(s)

    @classmethod
    def _parse(cls, s: str) -> list[Grid]:
        temp = [
            [re.findall(r"\S+", r) for r in l.splitlines()]
            for l in re.split(r"(?:\r?\n){2,}", s.strip())
        ]
        grids = [Grid([list(r) for r in g]) for t1 in temp for g in zip(*t1)]
        return grids

    def power(self) -> int:
        return sum(grid.power() for grid in self.grids)

    def __str__(self) -> str:
        return "\n\n".join(f"{grid}" for grid in self.grids)


@test(tests=tests[:])
def part_II(s: str) -> int:
    gs = Grids(s)
    return gs.power()
````

# 🐦 Everybody Codes 2024 — Quest 9, Part III

## Binding the Shrine With Runic Power

Your training is complete — you are now ready to join the shrine builders.

It turns out that covering the shrine with runic words is not merely decorative.  
Each **platinum block** used in construction contains a **6×6 grid** of symbols on one face.  
This 6×6 area corresponds to the **center** of the 8×8 grids you practiced on earlier.

The outer symbols of each 8×8 grid appear on **adjacent blocks**, meaning:

- Every 8×8 grid overlaps with **nine** blocks.
- The runic word produced from an 8×8 grid **binds all nine blocks together**.
- Only blocks that produce a valid runic word contribute to the shrine’s structural integrity.

You are now given a **section of the actual shrine wall**:

```
(your notes)
```

Not every 8×8 region will yield a runic word, but it is guaranteed that the entire wall can be secured.

---

## ❓ Unknown Symbols

Some cells contain a `?` symbol.  
Each `?` represents **exactly one runic symbol**, which must be deduced using the same rules as before:

- Each `.` or `?` must be replaced with the **unique symbol** that appears in both its row and its column.

Once all missing symbols in an 8×8 region are resolved, you can:

1. Extract the runic word (row‑by‑row, left to right, ignoring `*`).
2. Compute its power (as in Part II).
3. Add it to the total.

Your goal:

> **Identify all valid runic words in the wall section and compute the sum of their powers.**

---

## 📘 Example

Given the following wall section:

```
**XFZB**DCST**
**LWQK**GQJH**
?G....WL....DQ
BS....H?....CN
P?....KJ....TV
NM....Z?....SG
**NSHM**VKWZ**
**PJGV**XFNL**
WQ....?L....YS
FX....DJ....HV
?Y....WM....?J
TJ....YK....LP
**XRTK**BMSP**
**DWZN**GCJV**
```

### Step 1 — Identify 8×8 blocks

The wall is composed of overlapping 8×8 regions.  
The example shows how the top‑left block is isolated and solved first.

### Step 2 — Fill missing symbols

Using the row/column intersection rule, the missing symbols are deduced:

- First missing symbol → `V`
- Next → `X`
- Next → `Q`
- Last → `F`

This completes the first block and reveals additional symbols needed for the next block.

### Step 3 — Solve all solvable blocks

Two blocks in the example produce valid runic words.  
Their powers are:

```
1900
1989
```

Total:

```
1900 + 1989 = 3889
```

---

## 🎯 Your Task

You are given a large wall section:

> **Find all solvable 8×8 blocks, extract their runic words, compute their powers, and return the sum of all powers.**

Some blocks may be unsolvable — that is expected.  
Only include blocks that yield a complete, valid runic word.

---


````python
from itertools import chain

from test_utilities import test

tests = [
    {
        "name": "Example",
        "s": """
            **XFZB**DCST**
            **LWQK**GQJH**
            ?G....WL....DQ
            BS....H?....CN
            P?....KJ....TV
            NM....Z?....SG
            **NSHM**VKWZ**
            **PJGV**XFNL**
            WQ....?L....YS
            FX....DJ....HV
            ?Y....WM....?J
            TJ....YK....LP
            **XRTK**BMSP**
            **DWZN**GCJV**
        """,
        "expected": 3889,
    },
]


class GridView:
    BASE_POWER = {c: i for i, c in enumerate(ascii_uppercase, start=1)}
    OUTSIDE = 0, 1, 6, 7
    INSIDE = 2, 3, 4, 5
    PADDING = 2
    DIMENSION = 8

    def __init__(self, row: int, col: int, grid: list[list[str]]) -> None:
        self.row, self.col = row, col
        self.grid = grid

        self._rune()

    def _rune(self) -> bool:
        cols, rows = self.get_proposed_letters()
        change = False

        for r, c in product(
            range(self.PADDING, self.DIMENSION - self.PADDING), repeat=2
        ):
            if self.grid[self.row + r][self.col + c] == ".":
                cols_s = set(cols[c - self.PADDING])
                rows_s = set(rows[r - self.PADDING])

                intersect = cols_s & rows_s

                if len(intersect) == 1:
                    self.grid[self.row + r][self.col + c] = one(intersect)
                    change = True
                else:
                    symm_dif = self.symm_diff(r, c, cols_s, rows_s)

                    if len(symm_dif) == 2 and "?" in symm_dif:
                        symm_dif.discard("?")

                        self.grid[self.row + r][self.col + c] = one(symm_dif)

                        idx = self.get_questionmarks(r, c)
                        if len(idx) == 1:
                            rr, cc = one(idx)
                            self.grid[rr][cc] = self.grid[self.row + r][self.col + c]
                            change = True

        return change

    def get_proposed_letters(self) -> tuple[list[list[str]], list[list[str]]]:
        cols = [
            [self.grid[self.row + dr][self.col + dc] for dr in self.OUTSIDE]
            for dc in self.INSIDE
        ]
        rows = [
            [self.grid[self.row + dr][self.col + dc] for dc in self.OUTSIDE]
            for dr in self.INSIDE
        ]

        return cols, rows

    def symm_diff(self, r: int, c: int, cols_s: set[str], rows_s: set[str]) -> set[str]:
        symm_dif = cols_s - {
            s for dr in self.INSIDE for s in self.grid[self.row + dr][self.col + c]
        }
        symm_dif ^= rows_s - {
            s for dc in self.INSIDE for s in self.grid[self.row + r][self.col + dc]
        }

        return symm_dif

    def get_questionmarks(self, r: int, c: int) -> list[tuple[int, int]]:
        idx = list(
            chain(
                (
                    (self.row + dr, self.col + c)
                    for dr in self.OUTSIDE
                    if self.grid[self.row + dr][self.col + c] == "?"
                ),
                (
                    (self.row + r, self.col + dc)
                    for dc in self.OUTSIDE
                    if self.grid[self.row + r][self.col + dc] == "?"
                ),
            )
        )
        return idx

    def get_rune(self) -> str:
        return "".join(
            self.grid[self.row + r][self.col + c]
            for r, c in product(
                range(self.PADDING, self.DIMENSION - self.PADDING),
                repeat=2,
            )
        )

    def power(self) -> int:
        try:
            return sum(
                p * self.BASE_POWER[c] for p, c in enumerate(self.get_rune(), start=1)
            )
        except KeyError:
            return 0

    def __str__(self) -> str:
        return "\n".join(
            "".join(
                self.grid[self.row + r][self.col + c] for c in range(self.DIMENSION)
            )
            for r in range(self.DIMENSION)
        )


class GridsIII(Grids):
    def __init__(self, s: str) -> None:
        super().__init__(s)
        while any(grid._rune() for grid in self.grids):
            ...

    @classmethod
    def _parse(cls, s: str) -> list[GridView]:
        grid = [list(l.strip()) for l in s.strip().splitlines()]
        return cls._conv(grid)

    @classmethod
    def _conv(cls, grid: list[list[str]]) -> list[GridView]:
        rows, cols, grid_dim, step_size = len(grid), len(grid[0]), 8, 6
        return [
            GridView(r, c, grid)
            for r in range(0, rows + (step_size - grid_dim - 1), step_size)
            for c in range(0, cols + (step_size - grid_dim - 1), step_size)
        ]

    def power(self) -> int:
        return sum(grid.power() for grid in self.grids)


@test(tests=tests[:])
def part_III(s: str) -> int:
    gs = GridsIII(s)
    return gs.power()
````
