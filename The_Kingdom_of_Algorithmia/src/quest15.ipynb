```python
from collections import deque

from test_utilities import test

type Cell = tuple[int, int]

tests = [
    {
        "name": "Example Part I",
        "notes": """
            #####.#####
            #.........#
            #.######.##
            #.........#
            ###.#.#####
            #H.......H#
            ###########
        """,
        "expected": 26,
    },
]


class Map:
    DIRECTIONS = (-1, 0), (0, 1), (1, 0), (0, -1)

    def __init__(self, notes: str) -> None:
        self.grid = [line.strip() for line in notes.strip().splitlines()]
        self.rows, self.cols = len(self.grid), len(self.grid[0])
        self.E = self.find_entry_pont()

    def shortest_possible_path_to_collect_one_herb(self, herb: str = "H") -> int:
        queue = deque([self.E])
        seen = set()
        distance = 0

        while queue:
            for _ in range(len(queue)):
                r, c = queue.popleft()

                if not (0 <= r < self.rows and 0 <= c < self.cols):
                    continue

                if self.grid[r][c] == herb:
                    return 2 * distance

                if self.grid[r][c] != ".":
                    continue

                if (r, c) in seen:
                    continue

                seen.add((r, c))

                queue.extend((r + dr, c + dc) for dr, dc in self.DIRECTIONS)

            distance += 1

        raise ValueError(f"No {herb} found.")

    def find_entry_pont(self) -> Cell:
        for col in range(self.cols):
            if self.grid[0][col] == ".":
                return 0, col
            if self.grid[-1][col] == ".":
                return self.rows - 1, col

        for row in range(self.rows):
            if self.grid[row][0] == ".":
                return row, 0
            if self.grid[row][-1] == ".":
                return row, self.cols - 1

        raise ValueError("No Entry Point")

    def __str__(self) -> str:
        return "\n".join(self.grid)


@test(tests=tests[:])
def part_I(notes: str) -> int:
    map = Map(notes)
    return map.shortest_possible_path_to_collect_one_herb()
```

The shortest path to collect at least one plant of each kind takes <span style="color:Ivory">38 steps</span>:

<span style="color:yellow; font:bold">What is the minimum number of steps required to collect at least one plant of each type of herb on your map and then return to the starting point?</span>.


```python
from collections import defaultdict
from functools import cache
from itertools import permutations, product


from test_utilities import test

type Edge = tuple[Cell, Cell, int]


tests = [
    {
        "name": "Example Part Ii",
        "notes": """
            ##########.##########
            #...................#
            #.###.##.###.##.#.#.#
            #..A#.#..~~~....#A#.#
            #.#...#.~~~~~...#.#.#
            #.#.#.#.~~~~~.#.#.#.#
            #...#.#.B~~~B.#.#...#
            #...#....BBB..#....##
            #C............#....C#
            #####################
        """,
        "expected": 38,
    },
]

INF = 10**38


class MapII(Map):
    def __init__(self, notes: str) -> None:
        super().__init__(notes)

        self.herbs = self.get_herbs()
        self.graph = self.create_graph()
        self.groups = self.create_groups()

    def shortest_path_to_select_one_of_all_herbs(self) -> int:
        min_distance = INF

        for perm in permutations(self.groups.keys()):
            d = self.distance(self.E, perm)
            min_distance = min(d, min_distance)

        return min_distance

    @cache
    def distance(self, fr: Cell, perms: tuple[str, ...]) -> int:
        if not perms:
            min_dist = self.graph[fr][self.E]
        else:
            min_dist = INF
            for to in self.groups[perms[0]]:
                if to in self.graph[fr] and self.graph[fr][to] < min_dist:
                    d = self.graph[fr][to] + self.distance(to, perms[1:])
                    min_dist = min(d, min_dist)
        return min_dist

    def get_herbs(self) -> dict[Cell, str]:
        herbs = {}

        for r, c in product(range(self.rows), range(self.cols)):
            if self.grid[r][c].isupper():
                herbs[(r, c)] = self.grid[r][c]

        herbs[self.E] = "XX"

        return herbs

    def create_groups(self) -> dict[str, list[Cell]]:
        groups = defaultdict(list)

        for r_herb, c_herb in self.graph.keys():
            if (r_herb, c_herb) != self.E:
                groups[self.grid[r_herb][c_herb]].append((r_herb, c_herb))

        return groups

    def bfs(self, start: Cell, graph: dict[Cell, dict[Cell, int]]) -> None:
        queue = deque([start])
        seen = set()

        distance = 0

        while queue:
            for _ in range(len(queue)):
                r, c = queue.popleft()

                if not (0 <= r < self.rows and 0 <= c < self.cols):
                    continue

                if self.grid[r][c] in "~#":
                    continue

                if (r, c) != start and self.grid[r][c] == self.herbs[start]:
                    continue

                if (r, c) in seen:
                    continue

                seen.add((r, c))

                if (r, c) != start and (r, c) in self.herbs:
                    graph[start][(r, c)] = distance

                queue.extend((r + dr, c + dc) for dr, dc in self.DIRECTIONS)

            distance += 1

    def create_graph(self) -> dict[Cell, dict[Cell, int]]:
        # as an exercise make a graph from the grid and the points of the herbs
        graph = defaultdict(dict)

        for h in self.herbs:
            self.bfs(h, graph)

        return graph

    def print_graph(self) -> None:
        for (rf, cf), tos in self.graph.items():
            print(f"From node {rf,cf}, {self.herbs[(rf,cf)]}:")

            for (rt, ct), d in tos.items():
                print(f"\t to{rt,ct}, {self.herbs[rt,ct]} => {d=}")

            print()


@test(tests=tests[:])
def part_II(notes: str) -> int:
    map = MapII(notes)
    return map.shortest_path_to_select_one_of_all_herbs()
```

## Part III

The final round of the competition seems to be the same as it was in the previous two rounds. All the rules remain unchanged. The map is slightly larger, the number of herbs to collect has increased again, but some areas seem familiar, so it should go smoothly... right?

<span style="color:yellow; font:bold">What is the minimum number of steps required to collect at least one plant of each type of herb on your map and then return to the starting point?</span>


```python
# BFS
from collections import deque

from more_itertools import first

from test_utilities import test

type Cell = tuple[int, int]

tests = [
    {
        "name": "Example Part I",
        "notes": """
            #####.#####
            #.........#
            #.######.##
            #.........#
            ###.#.#####
            #H.......H#
            ###########
        """,
        "expected": 26,
    },
    {
        "name": "Challenge Part I",
        "notes": notes1,
        "expected": 184,
    },
    {
        "name": "Example Part Ii",
        "notes": """
            ##########.##########
            #...................#
            #.###.##.###.##.#.#.#
            #..A#.#..~~~....#A#.#
            #.#...#.~~~~~...#.#.#
            #.#.#.#.~~~~~.#.#.#.#
            #...#.#.B~~~B.#.#...#
            #...#....BBB..#....##
            #C............#....C#
            #####################
        """,
        "expected": 38,
    },
    {
        "name": "Challenge Part II",
        "notes": notes2,
        "expected": 520,
    },
]


class Grid:
    DIRECTIONS = (-1, 0), (0, 1), (1, 0), (0, -1)

    def __init__(self, notes: str) -> None:
        self.grid = [line.strip() for line in notes.strip().splitlines()]
        self.rows, self.cols = len(self.grid), len(self.grid[0])
        self.E = self.find_entry_pont()
        self.herbs = self.get_herbs()

    def get_herbs(self) -> set[str]:
        herbs = set()

        for r, c in product(range(self.rows), range(self.cols)):
            if self.grid[r][c].isupper():
                herbs.add(self.grid[r][c])

        return herbs

    def shortest_possible_path_to_collect_one_herb(self, herb: str = "H") -> int:
        queue = deque([(*self.E, frozenset(self.herbs))])
        seen = set()
        distance = 0

        while queue:
            for _ in range(len(queue)):
                r, c, to_be_collected = queue.popleft()

                if not (0 <= r < self.rows and 0 <= c < self.cols):
                    continue

                if self.grid[r][c] in "#~":
                    continue

                if (r, c) == self.E and not to_be_collected:
                    return distance

                if (r, c, to_be_collected) in seen:
                    continue

                seen.add((r, c, to_be_collected))

                if self.grid[r][c] in self.herbs:
                    to_be_collected = to_be_collected - {self.grid[r][c]}

                queue.extend(
                    (r + dr, c + dc, to_be_collected) for dr, dc in self.DIRECTIONS
                )

            distance += 1

        raise ValueError(f"No {herb} found.")

    def find_entry_pont(self) -> Cell:
        return first((0, col) for col in range(self.cols) if self.grid[0][col] == ".")

    def __str__(self) -> str:
        return "\n".join(self.grid)


@test(tests=tests[:])
def part_III(notes: str) -> int:
    g = Grid(notes)
    return g.shortest_possible_path_to_collect_one_herb()
```

```python
# convert grid to graph and use Dijkstra

from collections import deque
from heapq import heappop, heappush

from more_itertools import first

from test_utilities import test

type Cell = tuple[int, int]

tests = [
    {
        "name": "Example Part I",
        "notes": """
            #####.#####
            #.........#
            #.######.##
            #.........#
            ###.#.#####
            #H.......H#
            ###########
        """,
        "expected": 26,
    },
    {
        "name": "Challenge Part I",
        "notes": notes1,
        "expected": 184,
    },
    {
        "name": "Example Part Ii",
        "notes": """
            ##########.##########
            #...................#
            #.###.##.###.##.#.#.#
            #..A#.#..~~~....#A#.#
            #.#...#.~~~~~...#.#.#
            #.#.#.#.~~~~~.#.#.#.#
            #...#.#.B~~~B.#.#...#
            #...#....BBB..#....##
            #C............#....C#
            #####################
        """,
        "expected": 38,
    },
    {
        "name": "Challenge Part II",
        "notes": notes2,
        "expected": 520,
    },
]


class GridII:
    DIRECTIONS = (-1, 0), (0, 1), (1, 0), (0, -1)

    def __init__(self, notes: str) -> None:
        self.grid = [line.strip() for line in notes.strip().splitlines()]
        self.rows, self.cols = len(self.grid), len(self.grid[0])
        self.E = self.find_entry_pont()
        self.points_of_interest = self.get_points_of_interest()
        self.herbs = {h for h in self.points_of_interest.values() if h != "XX"}
        self.graph = self.create_graph()

    def create_graph(self) -> dict[Cell, dict[Cell, int]]:
        graph = defaultdict(dict)

        for h in self.points_of_interest:
            self.bfs(h, graph)

        return graph

    def bfs(self, start: Cell, graph: dict[Cell, dict[Cell, int]]) -> None:
        queue = deque([start])
        seen = set()

        distance = 0

        while queue:
            for _ in range(len(queue)):
                r, c = queue.popleft()

                if not (0 <= r < self.rows and 0 <= c < self.cols):
                    continue

                if self.grid[r][c] in "~#":
                    continue

                if (r, c) != start and self.grid[r][c] == self.points_of_interest[
                    start
                ]:
                    continue

                if (r, c) in seen:
                    continue

                seen.add((r, c))

                if (r, c) != start and (r, c) in self.points_of_interest:
                    graph[start][(r, c)] = distance

                queue.extend((r + dr, c + dc) for dr, dc in self.DIRECTIONS)

            distance += 1

    def get_points_of_interest(self) -> dict[Cell, str]:
        herbs = {}

        for r, c in product(range(self.rows), range(self.cols)):
            if self.grid[r][c].isupper():
                herbs[(r, c)] = self.grid[r][c]

        herbs[self.E] = "XX"

        return herbs

    def shortest_possible_path_to_collect_one_herb(self) -> int:
        heap = [(0, *self.E, frozenset(self.herbs))]
        n = len(self.herbs)
        seen = set()

        while heap:
            d, r, c, to_be_collected = heappop(heap)

            if (r, c) == self.E and not to_be_collected:
                return d

            if (r, c) == self.E and len(to_be_collected) < n:
                continue

            if not to_be_collected:
                continue

            if (r, c, to_be_collected) in seen:
                continue

            seen.add((r, c, to_be_collected))

            if self.points_of_interest[(r, c)] in to_be_collected:
                to_be_collected = to_be_collected - {self.points_of_interest[(r, c)]}
            elif (r, c) != self.E:
                continue

            for (nr, nc), dd in self.graph[(r, c)].items():
                heappush(heap, (d + dd, nr, nc, to_be_collected))

        raise ValueError(f"No {self.herbs} found.")

    def find_entry_pont(self) -> Cell:
        return first((0, col) for col in range(self.cols) if self.grid[0][col] == ".")

    def __str__(self) -> str:
        return "\n".join(self.grid)


@test(tests=tests[:])
def part_III(notes: str) -> int:
    grid = GridII(notes)
    return grid.shortest_possible_path_to_collect_one_herb()
```

```python
# convert grid to graph and use Lexisearch Approach to Traveling Salesman Problem
# https://www.iosrjournals.org/iosr-jm/papers/Vol6-issue4/A0640108.pdf


from test_utilities import test

type Cell = tuple[int, int]

tests = [
    {
        "name": "Example Part I",
        "notes": """
            #####.#####
            #.........#
            #.######.##
            #.........#
            ###.#.#####
            #H.......H#
            ###########
        """,
        "expected": 26,
    },
    {
        "name": "Challenge Part I",
        "notes": notes1,
        "expected": 184,
    },
    {
        "name": "Example Part Ii",
        "notes": """
            ##########.##########
            #...................#
            #.###.##.###.##.#.#.#
            #..A#.#..~~~....#A#.#
            #.#...#.~~~~~...#.#.#
            #.#.#.#.~~~~~.#.#.#.#
            #...#.#.B~~~B.#.#...#
            #...#....BBB..#....##
            #C............#....C#
            #####################
        """,
        "expected": 38,
    },
    {
        "name": "Challenge Part II",
        "notes": notes2,
        "expected": 520,
    },
]

INF = 10**18


class GridIII(GridII):
    def shortest_possible_path_to_collect_one_herb(self) -> int:
        """
        Simplified Lexicographic Search for TPP.
        Explores herb combinations in dictionary order with pruning.
        """
        shortest = INF
        points_per_herb = self.create_groups()

        for perm in permutations(points_per_herb.values()):
            for values in product(*perm):
                d = self.graph[self.E].get(values[0], INF) + self.graph[values[-1]].get(
                    self.E, INF
                )
                i = 1
                while i < len(values) and d <= shortest:
                    d += self.graph[values[i - 1]].get(values[i], INF)
                    i += 1

                if d < shortest:
                    shortest = d

        return shortest

    def create_groups(self):
        groups = defaultdict(list)
        for cell, herb in self.points_of_interest.items():
            if herb != "XX":
                groups[herb].append(cell)
        return groups


@test(tests=tests[:])
def part_III(notes: str) -> int:
    grid = GridIII(notes)
    return grid.shortest_possible_path_to_collect_one_herb()
```

```python
# This parses the data into a dictionary of allowed places to walk, and another dictionary of Points of Interest.
# The 'Start' POI is the only free path on the first row (if it exists!)
from collections import defaultdict


def parse(data):
    path, poi = dict(), defaultdict(set)
    for y, line in enumerate(data):
        for x, c in enumerate(line):
            if c in "#~":
                continue
            path[y, x] = c
            if y == 0 and c == ".":
                poi["Start"].add((y, x))
                path[y, x] = "Start"
            if c != ".":
                poi[c].add((y, x))
    return poi, path


# Find the distances from src to all the Points of Interest
def find_distances(path, src, poi):
    distances = {}
    boundary, seen, dist = set([src]), set(), 0
    while boundary:
        newboundary = set()
        for y, x in boundary:
            if (y, x) in path and path[y, x] in poi:
                distances[(path[y, x], (y, x))] = dist
            seen.add((y, x))
            for dy, dx in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
                ny, nx = y + dy, x + dx
                if (ny, nx) not in seen and (ny, nx) in path:
                    newboundary.add((ny, nx))
        dist += 1
        boundary = newboundary
    return distances


# Finds the minimum cost of following a specified route
def follow_route(route, distances):
    cur = {(c, loc): 0 for (c, loc) in distances if c == route[0]}
    for target in route[1:]:
        dests = {(c, loc): 10**9 for (c, loc) in distances if c == target}
        for src in cur:
            for dest in dests:
                dests[dest] = min(dests[dest], cur[src] + distances[src][dest])
        cur = dests
    return min(cur.values())


# Tries all the possible permutations of picking up items, starting and ending at a specified POI.
from itertools import permutations


def find_best_route(data, start="Start", split=None, choose=None):
    poi, path = parse(data)

    # We may have to split a POI into individual points that all need visiting.
    if split:
        locs = poi.pop(split)
        for i, loc in enumerate(locs):
            poi[split + str(i)] = set([loc])
            path[loc] = split + str(i)

    # We may have to choose just one of the start points based on a choice function.
    if choose:
        locs = poi.pop(start)
        for loc in locs:
            path[loc] = "."
        choice = choose(locs)
        poi[start] = set([choice])
        path[choice] = start

    distances = {}
    for c in poi:
        for loc in poi[c]:
            distances[(c, loc)] = find_distances(path, loc, poi)

    targets = set(poi) - {start}
    best_dist = 10**9
    for route in permutations(targets):
        route = [start] + list(route) + [start]
        dist = follow_route(route, distances)
        best_dist = min(best_dist, dist)
    return best_dist


def main():
    print("Part 1:", find_best_route(notes1.split("\n")))
    print("Part 2:", find_best_route(notes2.split("\n")))

    # Part 3 is too big to do naively (15! is in the trillions!)
    # It's formed of 3 columns which we can solve separately, and then add the scores together (+12 for the connection)

    data = notes3.split("\n")

    col_width = len(data[0]) // 3
    left = [line[:col_width] for line in data]
    middle = [line[col_width : 2 * col_width] for line in data]
    right = [line[2 * col_width :] for line in data]

    l = find_best_route(left, "E", choose=max)  # left route starts&ends at E
    m = find_best_route(
        middle, "Start", split="K"
    )  # middle route has to go through both K's
    r = find_best_route(right, "R", choose=min)  # end route starts&ends at R

    print("Part 3:", l + m + r + 6 * 2)


main()
```
