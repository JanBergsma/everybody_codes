## Part I

> ### :book: Story section
>
> Visiting the alchemists' workshop is always a fascinating experience. The shelves are filled with colourful herbs, each meticulously cared for, all housed in its own pot. Many of them you are seeing for the first time in your life, which seems odd as you know the nearby forests like the back of your hand. The only place you couldn't explore is the Infinite Loop Forest, guarded day and night by the knights of the Order. You have suspected for a long time that these unique plants can be found there, and today you will have the chance to verify that by yourself!
> <br/><br/>

This time, the candidates for joining the Order must demonstrate their ability to navigate unfamiliar terrain. Each knight receives a map with the locations of various herbs <span style="color:Yellow">( your notes )</span>. <span style="color:Ivory">Each type of herb is represented by a unique capital letter and can appear in multiple locations.</span> To collect the herb, you must step on the field where this herb grows. Fields marked with <span style="color:Ivory;font-size:1.2em;padding: 0 1em  0 1em; background:black"> # </span> are poison bushes (and thus, should be avoided).

Every map represents an enclosed area with a <span style="color:Ivory">single entry point at the top edge</span>. The race begins from this point. The first task is to find the herb marked on the map with <span style="color:Ivory;font-size:1.2em;padding: 0 1em  0 1em; background:black"> H </span> and return <span style="color:Ivory">to the starting point. Only a single plant is enough</span>. You don't have to visit all the places where the herb grows. Time is of the essence, so you need to plan the shortest possible route to achieve this goal!

<span style="color:Yellow">What is the minimum number of steps required to collect the herb and then return to the starting point?</span>


```python
from collections import deque

from test_utilities import test

type Cell = tuple[int, int]

tests = [
    {
        "name": "Example Part I",
        "notes": """
            #####.#####
            #.........#
            #.######.##
            #.........#
            ###.#.#####
            #H.......H#
            ###########
        """,
        "expected": 26,
    },
]


class Map:
    DIRECTIONS = (-1, 0), (0, 1), (1, 0), (0, -1)

    def __init__(self, notes: str) -> None:
        self.grid = [line.strip() for line in notes.strip().splitlines()]
        self.rows, self.cols = len(self.grid), len(self.grid[0])
        self.E = self.find_entry_pont()

    def shortest_possible_path_to_collect_one_herb(self, herb: str = "H") -> int:
        queue = deque([self.E])
        seen = set()
        distance = 0

        while queue:
            for _ in range(len(queue)):
                r, c = queue.popleft()

                if not (0 <= r < self.rows and 0 <= c < self.cols):
                    continue

                if self.grid[r][c] == herb:
                    return 2 * distance

                if self.grid[r][c] != ".":
                    continue

                if (r, c) in seen:
                    continue

                seen.add((r, c))

                queue.extend((r + dr, c + dc) for dr, dc in self.DIRECTIONS)

            distance += 1

        raise ValueError(f"No {herb} found.")

    def find_entry_pont(self) -> Cell:
        for col in range(self.cols):
            if self.grid[0][col] == ".":
                return 0, col
            if self.grid[-1][col] == ".":
                return self.rows - 1, col

        for row in range(self.rows):
            if self.grid[row][0] == ".":
                return row, 0
            if self.grid[row][-1] == ".":
                return row, self.cols - 1

        raise ValueError("No Entry Point")

    def __str__(self) -> str:
        return "\n".join(self.grid)


@test(tests=tests[:])
def part_I(notes: str) -> int:
    map = Map(notes)
    return map.shortest_possible_path_to_collect_one_herb()
```

In the example map above, there are three different types of herbs: <span style="color:Ivory">A</span>, <span style="color:Ivory">B</span>, and <span style="color:Ivory">C</span>.
The shortest path to collect at least one plant of each kind takes <span style="color:Ivory">38 steps</span>:

- <span style="color:Ivory">12 steps</span> to reach the A on the left side,
- <span style="color:Ivory">7 more steps</span> to move to the C in the left corner,
- <span style="color:Ivory">9 more steps</span> to move to the first B from the left,
- and<span style="color:Ivory"> 10 more steps</span> to return to the entry point.

<span style="color:yellow; font:bold">HWhat is the minimum number of steps required to collect at least one plant of each type of herb on your map and then return to the starting point?</span>.


```python
from itertools import product
from pprint import pprint

import networkx as nx
import matplotlib.pyplot as plt
from tabulate import tabulate

from test_utilities import test

type Edge = tuple[Cell, Cell, int]


plt.style.use("dark_background")  # makes axes, text, and background dark

tests = [
    {
        "name": "Example Part Ii",
        "notes": """
            ##########.##########
            #...................#
            #.###.##.###.##.#.#.#
            #..A#.#..~~~....#A#.#
            #.#...#.~~~~~...#.#.#
            #.#.#.#.~~~~~.#.#.#.#
            #...#.#.B~~~B.#.#...#
            #...#....BBB..#....##
            #C............#....C#
            #####################
        """,
        "expected": 38,
    },
]


class MapII(Map):
    def __init__(self, notes: str) -> None:
        super().__init__(notes)

        self.herbs = self.get_herbs()
        self.graph = self.create_graph()

    def get_herbs(self) -> dict[Cell, str]:
        herbs = {}

        for r, c in product(range(self.rows), range(self.cols)):
            if self.grid[r][c].isupper():
                herbs[(r, c)] = self.grid[r][c]

        herbs[self.E] = "XX"

        return herbs

    def bfs(self, start: Cell, graph: dict[Cell, dict[Cell, int]]) -> None:
        queue = deque([start])
        seen = set()
        local = {}
        graph[start] = local

        distance = 0

        while queue:
            for _ in range(len(queue)):
                r, c = queue.popleft()

                if not (0 <= r < self.rows and 0 <= c < self.cols):
                    continue

                if (r, c) != start and self.grid[r][c] == self.herbs[start]:
                    continue

                if (r, c) in seen:
                    continue

                seen.add((r, c))

                if (r, c) != start and (r, c) in self.herbs:
                    local[(r, c)] = distance

                queue.extend((r + dr, c + dc) for dr, dc in self.DIRECTIONS)

            distance += 1

    def create_graph(self) -> dict[Cell, dict[Cell, int]]:
        graph = {}

        for h in self.herbs:
            self.bfs(h, graph)

        return graph

    def print_graph(self) -> None:
        for (rf, cf), tos in self.graph.items():
            print(f"From node {rf,cf}, {self.herbs[(rf,cf)]}:")

            for (rt, ct), d in tos.items():
                print(f"\t to{rt,ct}, {self.herbs[rt,ct]} => {d=}")

            print()


@test(tests=tests[:])
def part_II(notes: str) -> int:
    map = MapII(notes)
    pprint(map.graph)
    # return map.shortest_path_to_select_one_of_all_herbs()
```
