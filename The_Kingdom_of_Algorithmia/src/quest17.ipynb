First, you need to know how to measure <span style="color:Ivory">the distance between two stars</span>. Each star has coordinates that define its position in space. To calculate the distance between two stars based on their coordinates, take the absolute difference of their X coordinates and the absolute difference of their Y coordinates. Then sum these differences. For example, if you have two stars with coordinates at `2, 4` and at ` 7, 3` , the distance between them is calculated as `abs(2 - 7) + abs(4 - 3) = 5 + 1 = 6 `.

This method of measurement is known as [<span style="color:lime">the algorithm from Manhattan's Realm</span>](https://en.wikipedia.org/wiki/Taxicab_geometry)

A constellation is a group of stars that forms a single structure such that <span style="color:Ivory">the sum of all distances between connected stars is as low as possible. There are also as few connections as necessary to maintain connectivity.</span>

Once the stars are connected in this manner, <span style="color:Ivory">the size of the constellation</span> is determined, which is <span style="color:Ivory">the number of stars forming the constellation combined with the sum of the distances between connected stars.</span>

The sky is exceptionally clear and dark tonight. These are ideal conditions for the tournament's astronomy competition. Your goal is <span style="color:Ivory">to create a constellation from these stars and determine its size.</span>
