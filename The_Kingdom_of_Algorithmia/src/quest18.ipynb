```python
import re
from collections import deque
from collections.abc import Iterator

from test_utilities import test
from util import Str

tests = [
    {
        'name': 'Example Part I',
        'notes': """
            ##########
            ..#......#
            #.P.####P#
            #.#...P#.#
            ##########
        """,
        'expected': 11,
    },
    {
        'name': 'Example Part II',
        'notes': """
            #######################
            ...P..P...#P....#.....#
            #.#######.#.#.#.#####.#
            #.....#...#P#.#..P....#
            #.#####.#####.#########
            #...P....P.P.P.....P#.#
            #.#######.#####.#.#.#.#
            #...#.....#P...P#.#....
            #######################
        """,
        'expected': 21,
    },
]


class IrrigationSystem(Str):
    def __init__(self, notes: str) -> None:
        self.channels = re.findall(r'\S+', notes)
        self.palms = sum(1 for line in self.channels for c in line if c == 'P')
        self.entries = self.get_entries()

    def get_entries(self) -> list[tuple[int, int]]:
        entries = []
        for r in range(len(self.channels)):
            if self.channels[r][0] == '.':
                entries.append((r, 0))

            if self.channels[r][-1] == '.':
                entries.append((r, len(self.channels[0]) - 1))

        for c in range(len(self.channels[0])):
            if self.channels[0][c] == '.':
                entries.append((0, c))

            if self.channels[-1][c] == '.':
                entries.append((len(self.channels) - 1, c))

        return entries

    def time_all_palms_watered(self) -> int:
        time = 0
        queue = deque(self.entries)
        seen = set()
        palms = self.palms

        while queue:
            for _ in range(len(queue)):
                r, c = queue.popleft()

                if (r, c) in seen:
                    continue

                seen.add((r, c))

                if self.channels[r][c] == 'P':
                    palms -= 1

                if palms == 0:
                    return time

                queue.extend(self.valid_neighbors(r, c))  # type: ignore

            time += 1

        raise ValueError('he?')

    def valid_neighbors(self, r: int, c: int) -> Iterator[tuple[int, int]]:
        for dr, dc in ((-1, 0), (0, 1), (1, 0), (0, -1)):
            rr, cc = r + dr, c + dc

            if not (0 <= rr < len(self.channels) and 0 <= cc < len(self.channels[0])):
                continue

            if self.channels[rr][cc] == '#':
                continue

            yield rr, cc


@test(tests=tests[:])
def part_I(notes: str) -> int:
    irs = IrrigationSystem(notes)
    return irs.time_all_palms_watered()
```

```python
from collections import defaultdict
from itertools import product
from pprint import pprint

from tabulate import tabulate

from test_utilities import test

tests = [
    {
        'name': 'Example Part III',
        'notes': """
            ##########
            #.#......#
            #.P.####P#
            #.#...P#.#
            ##########
        """,
        'expected': 12,
    },
]


class IrrigationSystemIII(IrrigationSystem):
    def optimal_point(self) -> int:
        """Instead processing per '.' process per 'P'"""
        sum_of_times = defaultdict(int)

        for palm in self.points_of_interest('P'):
            self.added_distances_from_palm(palm, sum_of_times)

        return min(sum_of_times.values())

    def optimal_point_bf(self) -> int:
        """Runs in 9 minutes and 16 seconds. Way too slow"""
        points = self.points_of_interest('.')
        return min(self.sum_of_times(p) for p in points)

    def points_of_interest(self, kind: str) -> list[tuple[int, int]]:
        rows = len(self.channels)
        cols = len(self.channels[0])
        return [(r, c) for r, c in product(range(rows), range(cols)) if self.channels[r][c] == kind]

    def added_distances_from_palm(
        self,
        palm: tuple[int, int],
        sum_of_times: defaultdict[tuple[int, int], int],
    ) -> None:
        time = 0
        queue = deque([palm])
        seen = set()

        while queue:
            for _ in range(len(queue)):
                r, c = queue.popleft()

                if (r, c) in seen:
                    continue

                seen.add((r, c))

                if self.channels[r][c] == '.':
                    sum_of_times[(r, c)] += time

                queue.extend(self.valid_neighbors(r, c))

            time += 1

    def sum_of_times(self, point: tuple[int, int]) -> int:
        time = time_sum = 0

        queue = deque([point])
        seen = set()
        palms = self.palms

        while queue:
            for _ in range(len(queue)):
                r, c = queue.popleft()

                if (r, c) in seen:
                    continue

                seen.add((r, c))

                if self.channels[r][c] == 'P':
                    palms -= 1
                    time_sum += time

                if palms == 0:
                    return time_sum

                queue.extend(self.valid_neighbors(r, c))

            time += 1

        raise ValueError('he?')


@test(tests=tests[:])
def part_III(notes: str) -> int:
    irs = IrrigationSystemIII(notes)
    return irs.optimal_point()
```
