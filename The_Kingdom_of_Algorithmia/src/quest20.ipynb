To better understand the new machinery, Sir Greg presents a map of the test flight. There is a single starting point `S`, which is the cliff you are standing on. You can start your flight in any direction from that point. Warm streams of air are marked with `+`, cold streams of air with `-`, and rocky obstacles with `#`. They are located at exactly the same coordinates at every altitude above sea level (where altitude equals 0), infinitely upward.

The glider moves at a constant speed, but it flies quite fast, covering each segment of the map in one second.

- If there are no air streams (`.`), the glider loses 1 meter of altitude while passing through such a segment.
- If it passes through a cold air stream (`-`), it loses 2 meters of altitude instead.
- If it encounters a warm air stream (`+`), it gains 1 meter of altitude.

<span style="color:Ivory">The altitude at which the glider is located is always an integer value and changes by the rules above exactly at the center of the segment it is currently flying over.</span>

After 7 seconds of flight, you can reach an area of warm air, which is large enough to circle around, gaining 2 meters of altitude with each loop. After 100 seconds of flight, the glider will be at <span style="color:ivory">1045</span> meters,
significantly above the starting point!

<span style="color:yellow">What is the highest altitude you can reach after the first 100 seconds of flight?</span>


```python
import re
from collections import deque

from more_itertools import first
from test_utilities import test
from util import Str, hexcolor_str

tests = [
    {
        'name': 'Example Part I',
        'notes': """
            #....S....#
            #.........#
            #---------#
            #.........#
            #..+.+.+..#
            #.+-.+.++.#
            #.........#
        """,
        'expected': 1045,
    },
]


class TrainingMap(Str):
    AIR_STREAMS = {'.': -1, '-': -2, '+': 1}

    def __init__(self, notes: str, altitude: int = 1_000) -> None:
        self.altitude = altitude
        self.map = [list(line) for line in re.findall(r'\S+', notes)]
        self.S = self.get_location_of('S')

    def get_location_of(self, character: str) -> tuple[int, int]:
        return first(
            (r, c) for r, row in enumerate(self.map) for c, ch in enumerate(row) if ch == character
        )

    def max_altitude_after_flight_of(self, seconds: int = 100, debug: bool = False) -> int:
        rows, cols = len(self.map), len(self.map[0])

        queue = deque([(self.altitude, *self.S, 1, 0)])

        max_values = {}

        for t in range(seconds + 1):
            for _ in range(len(queue)):
                alt, r, c, dr, dc = queue.popleft()

                if alt <= max_values.get((r, c, dr, dc), 0):
                    continue

                max_values[(r, c, dr, dc)] = alt

                if debug:
                    print(self.debug_str(r, c))
                    print()
                    print(f'{t=},{alt=}, {r,c=}, {dr,dc=}')
                    print()

                for ddr, ddc in ((dr, dc), (dc, -dr), (-dc, dr)):
                    r1, c1 = r + ddr, c + ddc

                    if not (0 <= r1 < rows and 0 <= c1 < cols):
                        continue

                    if self.map[r1][c1] != 'S' and self.map[r1][c1] not in self.AIR_STREAMS:
                        continue

                    altitude = alt + self.AIR_STREAMS.get(self.map[r1][c1], -1)
                    queue.append((altitude, r1, c1, ddr, ddc))

        return max(max_values.values())

    def debug_str(self, row: int, col: int) -> str:
        return '\n'.join(
            ''.join(
                self.map[r][c] if (r, c) != (row, col) else hexcolor_str('#FFFF00', 'G')
                for c in range(len(self.map[0]))
            )
            for r in range(len(self.map))
        )


@test(tests=tests[:])
def part_I(notes: str) -> int:
    tm = TrainingMap(notes)
    return tm.max_altitude_after_flight_of(seconds=100, debug=False)
```

```python
with open('../inputs/everybody_codes_e2024_q20_p1.txt') as f:
    notes1 = f.read()

print(f'Part I: {part_I(notes1)}')
```

```python
from test_utilities import test

tests = [
    {
        'name': 'Example Part II A',
        'notes': """
            ####S####
            #-.+++.-#
            #.+.+.+.#
            #-.+.+.-#
            #A+.-.+C#
            #.+-.-+.#
            #.+.B.+.#
            #########
        """,
        'expected': 24,
    },
    {
        'name': 'Example Part II B',
        'notes': """
            ###############S###############
            #+#..-.+.-++.-.+.--+.#+.#++..+#
            #-+-.+-..--..-+++.+-+.#+.-+.+.#
            #---.--+.--..++++++..+.-.#.-..#
            #+-+.#+-.#-..+#.--.--.....-..##
            #..+..-+-.-+.++..-+..+#-.--..-#
            #.--.A.-#-+-.-++++....+..C-...#
            #++...-..+-.+-..+#--..-.-+..-.#
            #..-#-#---..+....#+#-.-.-.-+.-#
            #.-+.#+++.-...+.+-.-..+-++..-.#
            ##-+.+--.#.++--...-+.+-#-+---.#
            #.-.#+...#----...+-.++-+-.+#..#
            #.---#--++#.++.+-+.#.--..-.+#+#
            #+.+.+.+.#.---#+..+-..#-...---#
            #-#.-+##+-#.--#-.-......-#..-##
            #...+.-+..##+..+B.+.#-+-++..--#
            ###############################
        """,
        'expected': 78,
    },
    {
        'name': 'Example Part II C',
        'notes': """
            ###############S###############
            #-----------------------------#
            #-------------+++-------------#
            #-------------+++-------------#
            #-------------+++-------------#
            #-----------------------------#
            #-----------------------------#
            #-----------------------------#
            #--A-----------------------C--#
            #-----------------------------#
            #-----------------------------#
            #-----------------------------#
            #-----------------------------#
            #-----------------------------#
            #-----------------------------#
            #--------------B--------------#
            #-----------------------------#
            #-----------------------------#
            ###############################
        """,
        'expected': 206,
    },
]

type Cell = tuple[int, int]

INF = 10**18


class TrainingMapII(TrainingMap):
    def __init__(self, notes: str, altitude: int = 1000) -> None:
        super().__init__(notes, altitude)
        self.altitude = 10_000

    def shortest_time_to_visited_checkpoints_and_back_to_start(self, debug: bool = False) -> int:
        """too slow runs for 3-4 minutes"""
        rows, cols = len(self.map), len(self.map[0])

        waypoints = 'ABCS'
        queue = deque([(self.altitude, *self.S, 1, 0, 0)])

        max_values = {}

        time = 0

        while queue:
            for _ in range(len(queue)):
                alt, r, c, dr, dc, waypoint = queue.popleft()

                if alt >= self.altitude and self.map[r][c] == waypoints[waypoint] == 'S':
                    return time

                if alt <= max_values.get((r, c, dr, dc, waypoint), self.altitude - 100):
                    continue

                max_values[(r, c, dr, dc, waypoint)] = alt

                if (
                    self.map[r][c].isupper()
                    and self.map[r][c] != waypoints[waypoint]
                    and len(waypoints) == 1
                ):
                    continue

                if self.map[r][c] == waypoints[waypoint] != 'S':
                    waypoint += 1

                if debug:
                    print(self.debug_str(r, c))
                    print()
                    print(f'{time=},{alt=}, {r,c=}, {dr,dc=}')
                    print()

                for ddr, ddc in ((dr, dc), (dc, -dr), (-dc, dr)):
                    r1, c1 = r + ddr, c + ddc

                    if 0 <= r1 < rows and 0 <= c1 < cols and self.map[r1][c1] != '#':
                        queue.append(
                            (
                                alt + self.AIR_STREAMS.get(self.map[r1][c1], -1),
                                r1,
                                c1,
                                ddr,
                                ddc,
                                waypoint,
                            )
                        )

            time += 1

        raise ValueError('Oops!')


@test(tests=tests[:])
def part_II(notes: str) -> int:
    tm = TrainingMapII(notes)
    return tm.shortest_time_to_visited_checkpoints_and_back_to_start(debug=False)
```
