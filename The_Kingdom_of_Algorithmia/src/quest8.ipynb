# 🏛️ Everybody Codes 2024 — Quest 8, Part I

The knights of the **Order of the Golden Duck** are not only fearless warriors — they are also exceptional architects. Their next challenge is to design a new shrine in honor of **Nullpointer**, the protector of the lost.

The priests insist on two requirements:

- The structure must be **perfectly symmetrical**
- It must have **exactly one highest point**

---

## 🧱 The Pyramid Construction Rules

The proposed shrine is a **pyramid made of stone blocks**. Its shape follows a simple rule:

> From the highest point, the height decreases by **1 block** for every block you move sideways.

The construction algorithm:

1. **Place one block** at the center — this is the top.
2. Determine the thickness of the next layer.
   - For a pyramid, this is always **1**.
3. Add a new layer by placing blocks **to the left and right** on the ground.
4. Add blocks above all columns that existed in the previous layer.
5. Repeat until you run out of blocks.

---

## 📐 Example of Layer Growth

Each layer forms a wider row:

```
Layer 1: 1 block
Layer 2: 3 blocks
Layer 3: 5 blocks
Layer 4: 7 blocks
...
```

## 🎯 Your Task

Given your input (a single number: the number of available blocks), determine:

1. **The final width** of the shrine
2. **How many additional blocks** are needed to complete the final layer
3. **The product:**
   $$
   \text{missing blocks} \times \text{final width}
   $$

---


````python
from test_utilities import test

tests = [
    {
        "name": "Example",
        "n": 13,
        "expected": 3 * 7,
    },
]


@test(tests=tests[:])
def part_I(n: int) -> int:
    c, t = 1, 0

    while t <= n:
        t += c
        c += 2

    return (c - 2) * (t - n)
````

# 🏛️ Everybody Codes 2024 — Quest 8, Part II

## Recursive Layer Thickness — The Shrine Evolves

The priests decide that a simple pyramid is not glorious enough for **Nullpointer**, the god of the lost.  
A knight proposes a more elaborate idea:

> Keep the same construction rules, but **change the thickness of each layer**, computed _recursively_.

---

## 🔁 Recursive Thickness Rules

Start with:

- **Layer 1 thickness = 1**

For each next layer:

1. Multiply the previous layer’s thickness by the number of **priests of Nullpointer**  
   (this number is _your notes_).
2. Take the result **modulo 1111** (the number of acolytes).
3. The result is the **thickness of the next layer**.

The shrine is now built from **marble**, and the priests have:

```
20240000 blocks
```

The king will still buy extra blocks if needed, so your task remains:

- Determine the **final width** of the shrine
- Determine the **number of missing blocks** needed to complete the last layer
- Compute:  
  `missing_blocks * final_width`

---

## 🎯 Your Task

Using:

- **Priests = your notes**
- **Acolytes = 1111**
- **Available blocks = 20240000**

Compute:

1. The **final width** of the shrine
2. The **number of missing blocks** needed to complete the last layer
3. The product:  
   `missing_blocks * final_width`

---


````python

from test_utilities import test

tests = [
    {
        "name": "Example",
        "priests": 3,
        "acolytes": 5,
        "blocks_available": 50,
        "expected": 3 * 9,
    },
]


@test(tests=tests[:])
def part_II(priests: int, acolytes: int, blocks_available: int) -> int:
    width, blocks = 1, 1
    thickness = 1

    while blocks < blocks_available:
        thickness = (thickness * priests) % acolytes
        width += 2
        blocks += width * thickness

    return (blocks - blocks_available) * width
````

# 🏛️ Everybody Codes 2024 — Quest 8, Part III

## The Final Shrine — Adjustable Layers and Hollow Columns

Everyone agrees the shrine is _almost_ perfect.  
Only a few parameters need refinement — and the priests want **empty spaces** inside the structure for offerings.

The final design modifies the previous rules.

---

## 🔁 Recursive Layer Thickness (New Rules)

Start with:

- **Layer 1 thickness = 1**

For each next layer:

1. Multiply the previous layer’s thickness by the number of **high priests of Nullpointer**  
   (this number is _your notes_).
2. Take the result **modulo 10** (the number of high priest acolytes).
3. Add **10** to ensure a minimum thickness of 10.

This gives the **thickness of the next layer**.

---

## 🧱 After Building the Shrine

Once the shrine is constructed layer by layer, we must determine how many **bottom blocks** in each column should be removed (left empty).

### Rules for removing blocks

- The **upper frame** of the shrine cannot be removed (to avoid collapse).
- Blocks are connected **side‑to‑side**, not diagonally.

Example:

```
     #                   #
    ###                 ###
    ###                 # #
    ###                 # #
   #####               ## ##
   #####       >       #   #
   #####               #   #
   #####               #   #
  #######             ##   ##
  #######             #     #
 #########           ##     ##
```

---

## 🧮 Computing Empty Space Per Column

For each column:

1. Multiply the number of **high priests** (your notes)  
   by the **current width** of the shrine base.
2. Multiply that result by the **height of the column**.
3. Take the result **modulo 10**.

The result = **number of bottom blocks to remove** in that column.

After doing this for all columns, you can compute the **total number of blocks actually needed**.

---

## 🪙 Material: Platinum

The shrine will be built from platinum.

Available supply:

```

```

The width does **not** matter — the Memory Stack Desert is large enough.

> **Find the number of platinum blocks the king must provide (minimum) so that the shrine can be fully completed.**

You must choose the number of layers so that:

- The shrine uses **all available blocks**, or as close as possible **without exceeding**, and

## 🧹 Removing Bottom Blocks

Width = 9  
High priests = 2  
Acolytes = 5

- Columns of height **6** → remove 0
- Columns of height **14** → remove `(2×9×14) mod 5 = 2`
- Columns of height **23** → remove `(2×9×23) mod 5 = 4`
- Columns of height **30** → remove `(2×9×30) mod 5 = 0`
- Column of height **31** → remove `(2×9×31) mod 5 = 3`

Total removed = **15**

Total blocks needed =  
`original total (177) – removed (15) = 162`

King must buy:  
`162 − 160 = 2`

## 🎯 Final Task

> **How many platinum blocks must the king provide (minimum) to ensure the shrine can be completed?**

You must compute:

- The number of layers that gets closest to using **202400000** blocks
- Without exceeding it
- And then determine how many **extra blocks** the king must buy

---


````python
from test_utilities import test

tests = [
    {
        "name": "Example",
        "priests": 2,
        "acolytes": 5,
        "blocks_available": 160,
        "expected": 2,
    },
]


@test(tests=tests[:])
def part_III(priests: int, acolytes: int, blocks_available: int) -> int:
    shrine = [1]
    blocks = 1

    while blocks < blocks_available:
        thickness = (shrine[0] * priests) % acolytes + acolytes

        for i in range(len(shrine)):
            shrine[i] += thickness
            blocks += thickness

        shrine.insert(0, thickness)
        shrine.append(thickness)
        blocks += 2 * thickness

    width = len(shrine)

    for col in shrine[1:-1]:
        blocks -= col * priests * width % acolytes

    return blocks - blocks_available
````
