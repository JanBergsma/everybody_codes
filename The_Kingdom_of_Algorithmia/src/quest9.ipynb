# 🌠 Everybody Codes 2024 — Quest 9, Part I

## The Rising Stars and the Beetle‑Powered Sparkballs

The sun has just set, and the city is settling into rest when you notice something strange:  
a **shooting star rising upward**.

At its peak, it bursts into **hundreds of tiny glowing sparks**.  
Then another appears. And another.

You trace the source to the **Headquarters of Pyromancers** near the market.

As you approach, the ground glows brighter. Strange shining spheres lie scattered around, and a pyromancer in a singed apron is loading them into a trebuchet. Each sphere — a _sparkball_ — launches into the sky and explodes into a dazzling display.

The pyromancer eagerly explains the secret:

- Each sparkball contains **beetles**.
- Each beetle has **shiny dots** stamped onto its shell.
- When released, the beetles scatter glowing powder in all directions.

To achieve a specific brightness, you must stamp beetles with the correct number of dots.  
There are **4 available stamps**:

```
1-dot
3-dot
5-dot
10-dot
```

Each beetle can be stamped **only once**.

Your task is to determine the **minimum number of beetles** needed to achieve the required brightness for each sparkball.

The pyromancer hands you a list of required brightness values:

```
(your notes)
```

---

## 📘 Example

Given the following brightness list:

```
2
4
7
16
```

We want to minimize the number of beetles used for each sparkball.

### Optimal stamping:

| Brightness | Optimal Stamps Used | Beetles |
| ---------- | ------------------- | ------- |
| 2          | 1 + 1               | 2       |
| 4          | 3 + 1               | 2       |
| 7          | 5 + 1 + 1           | 3       |
| 16         | 10 + 5 + 1          | 3       |

Total beetles needed:

```
2 + 2 + 3 + 3 = 10
```

---

## 🎯 Your Task

Given a list of brightness values (your notes):

> **Compute the minimum total number of beetles needed to stamp all sparkballs.**


````python
import re
from math import inf

from test_utilities import test

tests = [
    {
        "name": "Example",
        "s": """
            2
            4
            7
            16
        """,
        "expected": 2 + 2 + 3 + 3,
    },
]


def minimum_beetles(notes: list[int], stamps: list[int]) -> int:
    n = max(notes)
    beetles = [0] + [inf] * n

    for i in range(1, n + 1):
        for stamp in stamps:
            if stamp <= i:
                beetles[i] = min(beetles[i], beetles[i - stamp] + 1)

    return int(sum(beetles[n] for n in notes))


@test(tests=tests[:])
def part_I(s: str) -> int:
    notes = [int(n) for n in re.findall(r"\d+", s)]
    stamps = [1, 3, 5, 10]
    return minimum_beetles(notes, stamps)
````

````python
with open("../inputs/everybody_codes_e2024_q09_p1.txt") as f:
    notes1 = f.read()

print(f"Part I: {part_I(notes1)}")
````

# 🌌 Everybody Codes 2024 — Quest 9, Part III

## The King’s Gargantuan Sparkball Spectacle

In the evening, you meet the same pyromancer again at the market, performing another dazzling light show.  
You thank him for his help during the tournament — and he immediately presents you with a **new problem**.

The sparkballs have impressed the **king** himself.  
He has ordered a display of **enormous brightness**, far beyond what a single sparkball can hold.  
The pyromancer’s solution:

> **Split each required brightness into two sparkballs**, tied together and launched as a pair.

To maintain symmetry, the two sparkballs must have **similar brightness**:

```
Maximum allowed difference: 100 dots
```

---

## 🔧 New Stamp Set

The pyromancer has crafted many more stamps.  
You can now stamp a beetle with any of the following dot counts:

```
1, 3, 5, 10, 15, 16, 20, 24, 25, 30,
37, 38, 49, 50, 74, 75, 100, 101
```

Each beetle can still be stamped **only once**.

Your task:

> For each target brightness, split it into two sparkballs whose brightnesses differ by at most **100**,  
> and whose combined beetle count is **minimal**.

Then sum the beetle counts for all sparkballs.

---

## 📘 Example

Given the following brightness list:

```
156488
352486
546212
```

### 1. Brightness: 156488

Best split:

- Sparkball A: **78275** brightness → **775 beetles**
- Sparkball B: **78213** brightness → **775 beetles**

Difference:  
`78275 − 78213 = 62` (valid)

### 2. Brightness: 352486

Best split:

- Sparkball A: **176245** brightness → **1745 beetles**
- Sparkball B: **176241** brightness → **1745 beetles**

Difference:  
`176245 − 176241 = 4` (valid)

### 3. Brightness: 546212

Best split:

- Sparkball A: **273156** brightness → **2705 beetles**
- Sparkball B: **273056** brightness → **2704 beetles**

Difference:  
`273156 − 273056 = 100` (valid, at the limit)

### Total beetles needed

```
775 + 775 +
1745 + 1745 +
2705 + 2704
= 10449
```

---

## 🎯 Your Task

Given a list of target brightness values:

> **Compute the minimum total number of beetles needed to stamp all sparkballs**,  
> where each brightness must be split into two sparkballs whose brightness difference is at most **100**.

You may use any combination of the available stamps, but each beetle can be stamped only once.

---


````python


from test_utilities import test

tests = [
    {
        "name": "Example",
        "s": """
            156488
            352486
            546212
        """,
        "expected": 10449,
    },
]


def minimum_beetles_two_spark_balls(
    notes: list[int], stamps: list[int], max_difference: int = 100
) -> int:
    halve_diff = max_difference // 2
    n = max(notes) // 2 + halve_diff
    beetles = [0] + [inf] * n

    for i in range(1, n + 1):
        for stamp in stamps:
            if stamp <= i:
                beetles[i] = min(beetles[i], beetles[i - stamp] + 1)

    total_beetles = 0

    for bridgess in notes:
        halve1 = bridgess // 2
        halve2 = bridgess - halve1
        odd = bridgess & 1  # if odd halve1 is 1 too high

        total_beetles += min(
            beetles[halve1 - d] + beetles[halve2 + d]
            for d in range(halve_diff - odd + 1)
        )

    return int(total_beetles)


@test(tests=tests[:])
def part_III(s: str) -> int:
    notes = [int(n) for n in re.findall(r"\d+", s)]
    stamps = [1, 3, 5, 10, 15, 16, 20, 24, 25, 30, 37, 38, 49, 50, 74, 75, 100, 101]
    return minimum_beetles_two_spark_balls(notes, stamps)
````

````python
with open("../inputs/everybody_codes_e2024_q09_p3.txt") as f:
    notes3 = f.read()


print(f"Part III {part_III(notes3)}")
````
